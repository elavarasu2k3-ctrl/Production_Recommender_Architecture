### 0.1 Locate and load MovieLens

The loader looks (in order) at the `ML_DIR` environment variable, the usual project folders and a `data/` folder next to the notebook. If it finds nothing it tries
to download `ml-latest-small.zip` from GroupLens, and if that is also impossible it stops with instructions instead of silently using other data.

```python
def _is_ml_dir(p: Path) -> bool:
    return (p / "ratings.csv").is_file() and (p / "movies.csv").is_file()

def find_movielens():
    cands = []
    if os.environ.get("ML_DIR"):
        cands.append(Path(os.environ["ML_DIR"]).expanduser())
    cwd, home = Path.cwd(), Path.home()
    bases = [cwd, cwd.parent, cwd.parent.parent, home / "Documents", home / "Downloads", home / "Desktop"]
    for b in bases:
        cands += [b, b / "ml-latest-small", b / "data", b / "data" / "ml-latest-small", b / "data" / "raw", b / "data" / "raw" / "ml-latest-small"]
    for c in cands:
        if _is_ml_dir(c):
            return c
    # bounded search (depth <= 4, time-limited, junk folders pruned) under the usual project roots
    skip = {".git", "node_modules", "venv", ".venv", "site-packages", "__pycache__", "AppData", "$Recycle.Bin", "Library"}
    t_end = time.time() + 20
    for b in bases:
        if not b.is_dir():
            continue
        base_depth = len(b.parts)
        for root, dirs, files in os.walk(b):
            dirs[:] = [d for d in dirs if d not in skip and not d.startswith(".")]
            if len(Path(root).parts) - base_depth >= 4:
                dirs[:] = []
            if "ratings.csv" in files and _is_ml_dir(Path(root)):
                return Path(root)
            if time.time() > t_end:
                return None
    return None

def download_movielens(dest=Path("data")):
    url = "https://files.grouplens.org/datasets/movielens/ml-latest-small.zip"
    dest.mkdir(parents=True, exist_ok=True)
    zpath = dest / "ml-latest-small.zip"
    print("downloading", url)
    urllib.request.urlretrieve(url, zpath)
    with zipfile.ZipFile(zpath) as z:
        z.extractall(dest)
    return dest / "ml-latest-small"

DATA_DIR = find_movielens()
if DATA_DIR is None:
    try:
        DATA_DIR = download_movielens()
    except Exception as e:
        raise FileNotFoundError(
            "MovieLens ml-latest-small was not found and could not be downloaded "
            f"({type(e).__name__}: {e}).\nDownload it from https://grouplens.org/datasets/movielens/latest/ , unzip it, and either place the "
            "'ml-latest-small' folder next to this notebook or set the ML_DIR environment variable to its path.")

ratings = pd.read_csv(DATA_DIR / "ratings.csv")
movies = pd.read_csv(DATA_DIR / "movies.csv")
assert {"userId", "movieId", "rating", "timestamp"} <= set(ratings.columns) and {"movieId", "title", "genres"} <= set(movies.columns)
ratings = ratings.drop_duplicates(["userId", "movieId"], keep="last").reset_index(drop=True)

print("data folder :", DATA_DIR.resolve().name, "(", DATA_DIR.resolve().parent.name, ")")
print(f"ratings     : {len(ratings):,} rows | users: {ratings.userId.nunique():,} | rated movies: {ratings.movieId.nunique():,} | catalog (movies.csv): {len(movies):,}")
official = (len(ratings), ratings.userId.nunique(), len(movies)) == (100836, 610, 9742)
print("matches published ml-latest-small (100,836 / 610 / 9,742):", official)
if not official:
    print("NOTE: the dataset differs from the published ml-latest-small, so absolute numbers will differ from the earlier tasks.")
```

### 0.2 Implicit feedback, a leave-one-out split, and popularity-ordered item ids

The same protocol as Task 14, so numbers are comparable with it. Every rating is a *positive* interaction (implicit feedback). For each user, in time order:

| Split | Per user | Used for |
|---|---|---|
| `test`  | the **last** interaction | final numbers only — never used to choose anything |
| `val`   | the **second-last** interaction | early stopping and every choice of a setting (λ, index parameters, …) |
| `train` | everything earlier | fitting every model |

Items are re-indexed by *training* popularity (id 0 = most-rated movie); ids `0 … n_active-1` have at least one training interaction. A held-out movie with no training interaction cannot be scored by any ID-based model, so it counts as a miss for every model
(Exercise 4 returns to exactly these movies: they are the real "new items" of this dataset).

```python
user_ids = np.sort(ratings.userId.unique())
item_ids = np.sort(movies.movieId.unique())
n_users, n_items = len(user_ids), len(item_ids)
ratings["u"] = pd.Index(user_ids).get_indexer(ratings.userId)
ratings["i"] = pd.Index(item_ids).get_indexer(ratings.movieId)
assert (ratings.i >= 0).all(), "ratings.csv contains movies missing from movies.csv"
movies = movies.set_index("movieId").loc[item_ids].reset_index()

# time order inside each user (stable, so ties keep file order), then last = test, second last = val
ratings = ratings.sort_values(["u", "timestamp"], kind="mergesort").reset_index(drop=True)
ratings["pos"] = ratings.groupby("u").cumcount()
ratings["n_u"] = ratings.groupby("u").pos.transform("size")
ratings["split"] = np.where(ratings.pos == ratings.n_u - 1, "test", np.where(ratings.pos == ratings.n_u - 2, "val", "train"))
assert ratings.n_u.min() >= 3

# re-index items by TRAIN popularity (0 = most popular); items never seen in train go last
cnt_old = np.bincount(ratings.loc[ratings.split == "train", "i"], minlength=n_items)
order = np.lexsort((np.arange(n_items), -cnt_old))               # most popular first, ties by original index
new_of_old = np.empty(n_items, dtype=np.int64); new_of_old[order] = np.arange(n_items)
ratings["i"] = new_of_old[ratings.i.values]
movies = movies.iloc[order].reset_index(drop=True)
item_ids = item_ids[order]
assert (movies.movieId.values == item_ids).all()

train = ratings[ratings.split == "train"].reset_index(drop=True)
val   = ratings[ratings.split == "val"].sort_values("u").reset_index(drop=True)
test  = ratings[ratings.split == "test"].sort_values("u").reset_index(drop=True)
assert len(val) == len(test) == n_users and (val.u.values == np.arange(n_users)).all() and (test.u.values == np.arange(n_users)).all()
VAL_ITEM, TEST_ITEM = val.i.values, test.i.values
train_cnt = np.bincount(train.i.values, minlength=n_items).astype(np.float64)          # training popularity
n_active = int((train_cnt > 0).sum())                                                  # items with >= 1 training interaction are ids 0..n_active-1
assert (train_cnt[:n_active] > 0).all() and (train_cnt[n_active:] == 0).all()

TRAIN_MAT = np.zeros((n_users, n_items), dtype=bool); TRAIN_MAT[train.u.values, train.i.values] = True
SEEN_FOR_VAL  = TRAIN_MAT.copy()                                                       # candidates excluded when ranking for validation
SEEN_FOR_TEST = TRAIN_MAT.copy(); SEEN_FOR_TEST[np.arange(n_users), VAL_ITEM] = True   # ...and for test (the val item was already "consumed")
SEEN = {"val": SEEN_FOR_VAL, "test": SEEN_FOR_TEST}
TARGET = {"val": VAL_ITEM, "test": TEST_ITEM}

per_user = ratings.groupby("u").size()
print(f"users={n_users:,}  catalog={n_items:,}  interactions={len(ratings):,}  sparsity={1 - len(ratings) / (n_users * n_items):.2%}")
print(f"train={len(train):,}  val={len(val):,}  test={len(test):,} | items with >=1 training interaction: {n_active:,} of {n_items:,}")
print(f"test items never seen in training (unscorable by any ID-based model -> counted as misses for every model): {(TEST_ITEM >= n_active).sum()} users ({(TEST_ITEM >= n_active).mean():.1%})")
print(f"interactions per user: min={per_user.min()}  median={per_user.median():.0f}  max={per_user.max()}")
tr_u, tr_i = train.u.values.astype(np.int32), train.i.values.astype(np.int32)

# ---- item metadata used by later stages: release year (from the title) and a "when did the catalog first see it" timestamp ----
REL_YEAR = movies.title.map(lambda t: float(m.group(1)) if (m := re.search(r"\((\d{4})\)\s*$", str(t))) else np.nan).values.astype(np.float64)
FIRST_TS = ratings.groupby("i").timestamp.min().reindex(np.arange(n_items)).values.astype(np.float64)     # NaN if never rated
print(f"release year parsed for {np.isfinite(REL_YEAR).mean():.1%} of titles ({np.nanmin(REL_YEAR):.0f}-{np.nanmax(REL_YEAR):.0f}); "
      f"rating timestamps span {pd.to_datetime(ratings.timestamp.min(), unit='s').year}-{pd.to_datetime(ratings.timestamp.max(), unit='s').year}")
```

### 0.3 The evaluator (Task 13 metrics for one held-out movie per user)

With **one** relevant item per user the Task 13 definitions collapse to `HR@K = Recall@K = 1[rank ≤ K]` and `NDCG@K = 1[rank ≤ K] / log2(rank + 1)`. Ties are resolved **against** the model (a pessimistic rank), and an unscorable held-out movie is a miss for every model.
Two readings of the same quantity are used in this notebook:

* **Retrieval stage:** *Recall@R* = the share of users whose held-out movie is inside the R candidates the stage hands over. It is a **ceiling** for everything downstream — a movie that is not retrieved can never be scored or shown.
* **Final list:** `HR@10`, `NDCG@10` of the 10 movies the user finally sees.

Uncertainty is shown as 95% bootstrap intervals over *users* (only 610 test users, so single-model intervals are wide).

```python
def ranks_from_scores(S, target, exclude, big=np.inf):
    """Full-ranking rank (1 = best) of `target[u]` among all non-excluded active items. Pessimistic ties. Unscorable target -> inf (a miss)."""
    S = np.array(S, dtype=np.float64, copy=True)
    S[exclude] = -np.inf
    S[:, n_active:] = -np.inf                                      # items without training data cannot be scored fairly by an ID-based model
    ar = np.arange(len(S))
    st = S[ar, target]
    r = 1 + (S > st[:, None]).sum(1) + ((S == st[:, None]).sum(1) - 1)
    return np.where(np.isfinite(st) & (target < n_active), r, big).astype(np.float64)

def metrics_from_ranks(r, ks=KS):
    r = np.asarray(r, dtype=np.float64)
    out = {}
    for k in ks:
        hit = r <= k
        out[f"HR@{k}"] = float(hit.mean())
        out[f"NDCG@{k}"] = float(np.where(hit, 1.0 / np.log2(np.where(hit, r, 2.0) + 1.0), 0.0).mean())
    out["MRR"] = float(np.where(np.isfinite(r), 1.0 / np.where(np.isfinite(r), r, 1.0), 0.0).mean())
    return out

def per_user_ndcg(r, k=K):
    hit = np.asarray(r) <= k
    return np.where(hit, 1.0 / np.log2(np.where(hit, r, 2.0) + 1.0), 0.0)

def ranks_from_lists(lists, target):
    """Rank of the held-out movie inside each user's final ordered list (1-based); not in the list -> inf. `lists`: (n_users, N) item ids, best first (-1 = padding)."""
    lists = np.asarray(lists)
    hit = lists == np.asarray(target)[:, None]
    return np.where(hit.any(1), hit.argmax(1) + 1, np.inf).astype(np.float64)

def boot_ci(per_user_values, n_boot=2000, seed=0, alpha=0.05):
    v = np.asarray(per_user_values, dtype=np.float64)
    rs = np.random.default_rng(seed)
    m = np.array([v[rs.integers(0, len(v), len(v))].mean() for _ in range(n_boot)])
    return float(np.quantile(m, alpha / 2)), float(np.quantile(m, 1 - alpha / 2))

def paired_boot_diff(a, b, n_boot=4000, seed=0):
    """CI of mean(a - b) over users (paired bootstrap) -> (mean_diff, lo, hi)."""
    d = np.asarray(a, float) - np.asarray(b, float)
    rs = np.random.default_rng(seed)
    m = np.array([d[rs.integers(0, len(d), len(d))].mean() for _ in range(n_boot)])
    return float(d.mean()), float(np.quantile(m, 0.025)), float(np.quantile(m, 0.975))

# ---- sampled protocol (1 positive + 99 random negatives): used ONLY for early stopping of the pairwise models, as in Task 14 ----
def make_candidates(target, seed, n_neg=99):
    rs = np.random.default_rng(seed)
    all_mat = TRAIN_MAT.copy(); all_mat[np.arange(n_users), VAL_ITEM] = True; all_mat[np.arange(n_users), TEST_ITEM] = True
    cand = np.zeros((n_users, 1 + n_neg), dtype=np.int32)
    for u in range(n_users):
        pool = np.flatnonzero(~all_mat[u, :n_active])
        cand[u, 0] = target[u]
        cand[u, 1:] = rs.choice(pool, n_neg, replace=False)
    return cand
CAND_VAL = make_candidates(VAL_ITEM, SEED + 1)

def cand_ranks(scores, target):
    """scores: (n_users, 100) with the positive in column 0; pessimistic ties; unscorable positive -> miss."""
    r = 1 + (scores[:, 1:] >= scores[:, [0]]).sum(1)
    return np.where(target < n_active, r, np.inf).astype(np.float64)

def val_ndcg_s100(cand_scores_fn):
    return metrics_from_ranks(cand_ranks(cand_scores_fn(CAND_VAL), VAL_ITEM), (K,))[f"NDCG@{K}"]

# self-test of the metric code against the textbook definition
_r = np.array([1, 2, 10, 11, 50, np.inf])
_m = metrics_from_ranks(_r, (10,))
assert abs(_m["HR@10"] - 3 / 6) < 1e-12 and abs(_m["NDCG@10"] - (1 + 1 / np.log2(3) + 1 / np.log2(11)) / 6) < 1e-12
_l = np.array([[5, 7, 9], [1, 2, 3]]); assert (ranks_from_lists(_l, np.array([9, 4])) == np.array([3, np.inf])).all()
print("evaluator ready | metric self-tests passed")
```

### 0.4 Item content vectors (Task 11) and query features (Task 14, Phase 6)

* **Item content vectors `F`** — Task 11's featuriser: TF-IDF-weighted genres plus the release decade, L2-normalised. Used (i) as the *candidate-tower side features* of the two-tower model, (ii) as the item–item similarity inside the diversity re-ranker (the same feature space Task 13 used for its diversity metric), (iii) by the content-based baseline.
* **Query features** — for every interaction the user's last `H = 10` movies *before* it (ids, and their average genre vector). Built strictly from the past, so nothing leaks from the target. These are the side features of Task 14's best softmax models and the input of the query tower in Phase 3.

```python
class ItemFeaturizer:
    """Task 11: item vectors = TF-IDF-weighted genres + release decade, L2-normalised."""
    def __init__(self, decade_weight=0.5):
        self.decade_weight = decade_weight
    @staticmethod
    def _genres(s):
        return [g for g in str(s).split("|") if g and g != "(no genres listed)"]
    @staticmethod
    def _decade(title):
        m = re.search(r"\((\d{4})\)\s*$", str(title))
        return (int(m.group(1)) // 10) * 10 if m else None
    def fit(self, movies_df):
        gl = movies_df.genres.map(self._genres)
        self.vocab_ = sorted({g for l in gl for g in l})
        self.gidx_ = {g: j for j, g in enumerate(self.vocab_)}
        df_g = np.zeros(len(self.vocab_))
        for l in gl:
            for g in l: df_g[self.gidx_[g]] += 1
        self.idf_ = np.log((1 + len(movies_df)) / (1 + df_g)) + 1.0
        decs = sorted({d for d in movies_df.title.map(self._decade) if d is not None})
        self.decades_ = decs; self.didx_ = {d: j for j, d in enumerate(decs)}
        return self
    def transform(self, movies_df):
        X = np.zeros((len(movies_df), len(self.vocab_) + len(self.decades_)), dtype=np.float32)
        for r, (g, t) in enumerate(zip(movies_df.genres, movies_df.title)):
            for name in self._genres(g):
                if name in self.gidx_: X[r, self.gidx_[name]] = self.idf_[self.gidx_[name]]
            d = self._decade(t)
            if d in self.didx_: X[r, len(self.vocab_) + self.didx_[d]] = self.decade_weight
        return X / np.maximum(np.linalg.norm(X, axis=1, keepdims=True), 1e-12)

featurizer = ItemFeaturizer().fit(movies)
F = featurizer.transform(movies)                          # (n_items, f) L2-normalised content vectors, rows follow the popularity-ordered item ids
GENRES = featurizer.vocab_
G_MH = np.zeros((n_items, len(GENRES)), dtype=np.float32)  # plain multi-hot genre matrix (query-tower side feature)
for r, g in enumerate(movies.genres):
    for name in ItemFeaturizer._genres(g): G_MH[r, featurizer.gidx_[name]] = 1.0
PRIMARY_GENRE = np.array([(ItemFeaturizer._genres(g) or ["(none)"])[0] for g in movies.genres])
print("content vectors:", F.shape, "| genres:", len(GENRES), "| movies with no genre listed:", int((G_MH.sum(1) == 0).sum()))

# ---------------- query features: the last H items before each interaction ----------------
from numpy.lib.stride_tricks import sliding_window_view
H = 10
G_PAD = np.vstack([np.zeros((1, len(GENRES)), np.float32), G_MH])          # row 0 = padding, row j+1 = item j

def windows(seq, h=H):
    """Row t = the h items BEFORE position t (ids stored as item+1, 0 = padding, left-padded)."""
    padded = np.concatenate([np.zeros(h, np.int32), np.asarray(seq, np.int32) + 1])
    return sliding_window_view(padded, h)[:len(seq)]

W_ALL = np.concatenate([windows(g.i.values) for _, g in ratings.groupby("u", sort=True)]).astype(np.int32)     # aligned with `ratings` rows
assert len(W_ALL) == len(ratings)
split_arr = ratings.split.values
hist = {s: W_ALL[split_arr == s] for s in ("train", "val", "test")}
ts = {s: ratings.timestamp.values[split_arr == s] for s in ("train", "val", "test")}                          # timestamp of the target interaction (= the "query time")
assert (ratings.u.values[split_arr == "val"] == np.arange(n_users)).all() and (ratings.u.values[split_arr == "test"] == np.arange(n_users)).all()

def genre_mix(h):
    return (G_PAD[h].sum(1) / np.maximum((h > 0).sum(1, keepdims=True), 1)).astype(np.float32)

FEATS = {}
for sp_, u_arr in (("train", tr_u), ("val", np.arange(n_users, dtype=np.int32)), ("test", np.arange(n_users, dtype=np.int32))):
    FEATS[sp_] = {"user": u_arr.astype(np.int32)[:, None], "hist": hist[sp_], "genre_mix": genre_mix(hist[sp_])}
KEEP = (hist["train"] > 0).any(1)                                           # drop the first interaction of each user (no history)
print(f"training rows with >=1 history item: {KEEP.sum():,} of {len(KEEP):,} | window H = {H} | genre-mix dim = {len(GENRES)}")

# leakage check: every history item of a training row happened strictly earlier in that user's sequence
_rows = ratings[split_arr == "train"].reset_index(drop=True)
_chk = np.random.default_rng(0).choice(len(_rows), 400, replace=False)
_seq_pos = {(u, i): p for u, i, p in zip(ratings.u.values, ratings.i.values, ratings.pos.values)}
for r in _chk:
    u_, p_ = int(_rows.u[r]), int(_rows.pos[r])
    for h_ in FEATS["train"]["hist"][r]:
        assert h_ == 0 or _seq_pos[(u_, int(h_) - 1)] < p_, "history window contains a non-past item"
print("history-window leakage check passed (400 random training rows)")

class MaskedMean(L.Layer):
    """Mean of the embeddings of the non-padding history items (padding id = 0)."""
    def call(self, inputs):
        emb, ids = inputs
        m = tf.cast(ids > 0, emb.dtype)[..., None]
        return tf.reduce_sum(emb * m, 1) / tf.maximum(tf.reduce_sum(m, 1), 1.0)
    def compute_output_shape(self, shapes):
        return (shapes[0][0], shapes[0][2])
```

### 0.5 Two reference retrievers that need no training
* **Popularity** — rank by the number of training interactions (the hard baseline of Tasks 13–14).
* **Content-based (Task 11)** — a user profile = the average content vector of the movies the user has interacted with (training items; for test queries also the validation item, i.e. everything before the test movie), scored by cosine similarity.

```python
pop_row = (train_cnt + 1e-9 * (1.0 - np.arange(n_items) / n_items)).astype(np.float32)
S_POP = np.tile(pop_row, (n_users, 1))

def content_scores(split):
    """Task 11 content-based scores for the query at `split` ('val' or 'test'): cosine(profile, F[j])."""
    M = SEEN[split].astype(np.float32)                              # items the user had consumed before that query
    prof = M @ F
    prof /= np.maximum(np.linalg.norm(prof, axis=1, keepdims=True), 1e-12)
    return (prof @ F.T).astype(np.float32)
S_CONTENT = {s: content_scores(s) for s in ("val", "test")}

def eval_scores(S, split="test", ks=(10, 100, 500)):
    r = ranks_from_scores(S, TARGET[split], SEEN[split])
    return r, metrics_from_ranks(r, ks)

for nm, S in (("Popularity", S_POP), ("Content-based (Task 11)", S_CONTENT["test"])):
    _, mt = eval_scores(S)
    print(f"{nm:26s} test HR@10 {mt['HR@10']:.4f}  NDCG@10 {mt['NDCG@10']:.4f}  Recall@100 {mt['HR@100']:.4f}  Recall@500 {mt['HR@500']:.4f}")
```

### 0.6 The models of Tasks 12 and 14, rebuilt here (so the notebook is self-contained)

Every stage of the pipeline needs something from the earlier tasks, so the models are re-trained here with the hyper-parameters that were chosen on the validation split in Tasks 12–14 (no new tuning, no test-set peeking):

| Model | Task | Role in this notebook |
|---|---|---|
| **BPR-MF** (`d=32`, L2 `1e-3`) | 12 | an embedding model with a dot-product score — a natural *retrieval* candidate; also the "MF" of Phase 2's popularity-bias discussion |
| **Softmax DNN** (user id + recent genre mix + history bag, `d=32`) | 14 | its query network `ψ(x)` and item matrix `V` are the *Task 14 item embeddings* of Phase 2, and it is the blueprint of the two-tower model (a two-tower whose item tower is a plain lookup) |
| **NeuMF** (`d=16`, L2 gmf `1e-5` / mlp `1e-4`) | 14 | the **scoring** model of Phase 5: user and item interact *inside* the network, so it can only score a (user, item) pair you already have |

Popularity and the content-based model of Task 11 (§0.5) complete the set.

```python
# ---------- shared helpers for the pairwise (user, item) neural models ----------
D_NCF = 16                                                         # NeuMF "predictive factors" (Task 14)
LR_NCF = 1e-3
NEUMF_REG = (1e-5, 1e-4)                                           # (GMF-branch L2, MLP-branch L2): the values selected on the validation split in Task 14
BPR_L2 = 1e-3                                                      # BPR-MF L2 selected on the validation split in Task 14

def sample_train_arrays(n_neg, rs):
    """Every training interaction is a positive; for each, `n_neg` negatives are drawn uniformly from the active items the user has NOT interacted with in train.
    Negatives are re-drawn every epoch (as in the NCF paper), which also acts as a regulariser."""
    n = len(tr_u)
    u_neg = np.repeat(tr_u, n_neg)
    i_neg = rs.integers(0, n_active, size=n * n_neg).astype(np.int32)
    for _ in range(12):                                   # rejection-resample the (few) negatives that are actually positives
        bad = TRAIN_MAT[u_neg, i_neg]
        if not bad.any(): break
        i_neg[bad] = rs.integers(0, n_active, size=int(bad.sum()))
    u = np.concatenate([tr_u, u_neg]); i = np.concatenate([tr_i, i_neg])
    y = np.concatenate([np.ones(n, np.float32), np.zeros(n * n_neg, np.float32)])
    p = rs.permutation(len(u))
    return u[p], i[p], y[p]

def _forward(model, inputs, bs=131072):
    out = []
    for s in range(0, len(inputs[0]), bs):
        out.append(np.asarray(model([np.asarray(x[s:s + bs]).reshape(-1, 1) if np.ndim(x) == 1 else np.asarray(x[s:s + bs]) for x in inputs], training=False)).reshape(-1))
    return np.concatenate(out)

def pairwise_cand_scores(model, cand):
    U, C = cand.shape
    return _forward(model, [np.repeat(np.arange(U, dtype=np.int32), C), cand.reshape(-1).astype(np.int32)]).reshape(U, C)

def pairwise_full_scores(model):
    """(n_users, n_items) NeuMF-style scores for EVERY (user, active item) pair: this is the brute-force scan that production cannot afford."""
    S = np.full((n_users, n_items), -np.inf, dtype=np.float32)
    items = np.arange(n_active, dtype=np.int32)
    for u0 in range(0, n_users, 40):
        us = np.arange(u0, min(u0 + 40, n_users), dtype=np.int32)
        S[us, :n_active] = _forward(model, [np.repeat(us, n_active), np.tile(items, len(us))]).reshape(len(us), n_active)
    return S

def fit_pairwise(model, n_neg=4, epochs=None, batch_size=1024, patience=None, seed=0, label=""):
    """Train a [user, item] -> P(interaction) Keras model with binary cross-entropy on fresh negatives every epoch.
    Early stopping and model selection use the sampled-100 NDCG@10 on the VALIDATION item; the best epoch's weights are restored."""
    epochs = epochs or MAX_EPOCHS; patience = patience or PATIENCE + 1
    rs = np.random.default_rng(seed)
    best, best_w, best_ep, bad, rows = -1.0, None, 0, 0, []
    t_all = time.time()
    for ep in range(1, epochs + 1):
        u, i, y = sample_train_arrays(n_neg, rs)
        t0 = time.time()
        h = model.fit([u[:, None], i[:, None]], y, batch_size=batch_size, epochs=1, verbose=0, shuffle=False)
        sec = time.time() - t0
        v = val_ndcg_s100(lambda c: pairwise_cand_scores(model, c))
        rows.append({"epoch": ep, "train loss": h.history["loss"][0], "val NDCG@10 (s100)": v, "train sec": sec})
        if v > best + 1e-6: best, best_w, best_ep, bad = v, model.get_weights(), ep, 0
        else: bad += 1
        if bad >= patience: break
    model.set_weights(best_w)
    return pd.DataFrame(rows), best_ep, time.time() - t_all
```

```python
# ---------- Task 12: BPR matrix factorization (pairwise ranking loss, implicit feedback) ----------
class BPRMF:
    """x_ui = p_u . q_i + b_i ;  loss = -log sigmoid(x_ui - x_uj) + l2 * (|p_u|^2 + |q_i|^2 + |q_j|^2), j = a random item the user did not interact with."""
    def __init__(self, d=32, lr=5e-3, l2=1e-4, seed=0):
        self.d, self.lr, self.l2, self.seed = d, lr, l2, seed
        rs = np.random.default_rng(seed)
        self.P = tf.Variable(rs.normal(0, 0.05, (n_users, d)).astype(np.float32))
        self.Q = tf.Variable(rs.normal(0, 0.05, (n_items, d)).astype(np.float32))
        self.b = tf.Variable(np.zeros(n_items, np.float32))
        self.opt = keras.optimizers.Adam(lr)
        self._step = tf.function(self._step_impl)
    def _step_impl(self, u, i, j):
        with tf.GradientTape() as tape:
            pu, qi, qj = tf.gather(self.P, u), tf.gather(self.Q, i), tf.gather(self.Q, j)
            x = tf.reduce_sum(pu * (qi - qj), 1) + tf.gather(self.b, i) - tf.gather(self.b, j)
            loss = -tf.reduce_mean(tf.math.log_sigmoid(x)) + self.l2 * tf.reduce_mean(tf.reduce_sum(pu ** 2 + qi ** 2 + qj ** 2, 1))
        vs = [self.P, self.Q, self.b]
        self.opt.apply_gradients(zip(tape.gradient(loss, vs), vs))
        return loss
    def full_scores(self):
        S = (self.P.numpy() @ self.Q.numpy().T + self.b.numpy()[None, :]).astype(np.float32)
        S[:, n_active:] = -np.inf
        return S
    def cand_scores(self, cand):
        P, Q, b = self.P.numpy(), self.Q.numpy(), self.b.numpy()
        return np.einsum("ud,ucd->uc", P, Q[cand]) + b[cand]
    def fit(self, epochs=60, batch_size=1024, patience=6):
        rs = np.random.default_rng(self.seed)
        best, best_w, best_ep, bad = -1.0, None, 0, 0
        for ep in range(1, epochs + 1):
            j = rs.integers(0, n_active, size=len(tr_u)).astype(np.int32)
            for _ in range(12):
                bd = TRAIN_MAT[tr_u, j]
                if not bd.any(): break
                j[bd] = rs.integers(0, n_active, size=int(bd.sum()))
            p = rs.permutation(len(tr_u))
            for s in range(0, len(p), batch_size):
                sl = p[s:s + batch_size]
                self._step(tr_u[sl], tr_i[sl], j[sl])
            v = val_ndcg_s100(self.cand_scores)
            if v > best + 1e-6: best, best_w, best_ep, bad = v, [x.numpy() for x in (self.P, self.Q, self.b)], ep, 0
            else: bad += 1
            if bad >= patience: break
        for var, w in zip((self.P, self.Q, self.b), best_w): var.assign(w)
        self.best_epoch, self.best_val = best_ep, best
        return self

t = time.time()
bpr = BPRMF(d=32, lr=5e-3, l2=BPR_L2, seed=0).fit(epochs=5 if QUICK else 60)
S_BPR = {"test": bpr.full_scores()}
print(f"BPR-MF trained in {time.time() - t:.0f}s (best epoch {bpr.best_epoch})")
```

```python
# ---------- Task 14: softmax DNN with side features (user id + recent genre mix + history bag): query network psi(x) and item matrix V ----------
D = 32                                                    # query / item embedding size

def build_query_model(use_id=True, use_genre=False, use_hist=False, d=D, hidden=64, name="query_tower"):
    """psi(x): concatenates whichever features are switched on, then Dense(hidden, relu) -> Dense(d, relu). Returns (model, ordered input names)."""
    ins, feats, names = [], [], []
    if use_id:
        x = keras.Input(shape=(1,), dtype="int32", name="user"); ins.append(x); names.append("user")
        feats.append(L.Flatten()(L.Embedding(n_users, d, name="user_emb")(x)))
    if use_genre:
        x = keras.Input(shape=(len(GENRES),), name="genre_mix"); ins.append(x); names.append("genre_mix"); feats.append(x)
    if use_hist:
        x = keras.Input(shape=(H,), dtype="int32", name="hist"); ins.append(x); names.append("hist")
        feats.append(MaskedMean(name="hist_bag")([L.Embedding(n_items + 1, d, name="hist_item_emb")(x), x]))
    h = feats[0] if len(feats) == 1 else L.Concatenate(name="query_concat")(feats)
    h = L.Dense(hidden, activation="relu", name="hidden")(h)
    psi = L.Dense(d, activation="relu", name="psi")(h)
    return keras.Model(ins, psi, name=name), names

def build_softmax_head(n_classes=n_items, d=D):
    out = L.Dense(n_classes, name="item_softmax")                   # linear: logits = psi . V^T + b (softmax is applied inside the loss)
    out.build((None, d)); return out

def inputs_for(names, split, rows=None):
    arrs = [FEATS[split][n] for n in names]
    return [a[rows] for a in arrs] if rows is not None else arrs

def softmax_forward_scores(psi_model, out_layer, query_inputs, n_classes):
    """(n_users, n_items) logits psi(x).V^T + b; classes outside the model's output range get -inf."""
    psi = psi_model.predict(query_inputs, batch_size=1024, verbose=0)
    W, b = out_layer.kernel.numpy(), out_layer.bias.numpy()
    S = np.full((n_users, n_items), -np.inf, dtype=np.float32)
    S[:, :n_classes] = psi @ W + b
    return S

def fit_softmax(psi_model, out_layer, X, y, n_classes, epochs=None, batch_size=512, lr=2e-3, patience=3, seed=0, val_inputs=None):
    """Full-softmax training loop with early stopping on the VALIDATION full-ranking NDCG@10 (best weights restored)."""
    epochs = epochs or MAX_EPOCHS
    tf.random.set_seed(seed); rs = np.random.default_rng(seed)
    opt = keras.optimizers.Adam(lr)
    tvars = psi_model.trainable_variables + out_layer.trainable_variables

    @tf.function
    def step(xs, yb):
        with tf.GradientTape() as tape:
            psi = psi_model(xs, training=True)
            logits = tf.matmul(psi, out_layer.kernel) + out_layer.bias
            loss = tf.reduce_mean(tf.nn.sparse_softmax_cross_entropy_with_logits(labels=yb, logits=logits))
        opt.apply_gradients(zip(tape.gradient(loss, tvars), tvars))
        return loss

    best, best_w, best_ep, bad, rows = -1.0, None, 0, 0, []
    n = len(y)
    for ep in range(1, epochs + 1):
        p = rs.permutation(n); t0 = time.time(); losses = []
        for s in range(0, n, batch_size):
            sl = p[s:s + batch_size]
            losses.append(float(step([tf.constant(a[sl]) for a in X], tf.constant(y[sl].astype(np.int64)))))
        S = softmax_forward_scores(psi_model, out_layer, val_inputs, n_classes)
        v = metrics_from_ranks(ranks_from_scores(S, VAL_ITEM, SEEN_FOR_VAL), (K,))[f"NDCG@{K}"]
        rows.append({"epoch": ep, "train loss": np.mean(losses), "val NDCG@10": v, "train sec": time.time() - t0})
        if v > best + 1e-6: best, best_w, best_ep, bad = v, psi_model.get_weights() + out_layer.get_weights(), ep, 0
        else: bad += 1
        if bad >= patience: break
    n_psi = len(psi_model.get_weights())
    psi_model.set_weights(best_w[:n_psi]); out_layer.set_weights(best_w[n_psi:])
    return pd.DataFrame(rows), best_ep

t = time.time()
keras.utils.set_random_seed(400)
sm_psi, sm_names = build_query_model(use_id=True, use_genre=True, use_hist=True, name="softmax_query_network")
sm_head = build_softmax_head()
Xtr = [a[KEEP] for a in inputs_for(sm_names, "train")]
sm_curve, sm_best_ep = fit_softmax(sm_psi, sm_head, Xtr, tr_i[KEEP].astype(np.int64), n_items, val_inputs=inputs_for(sm_names, "val"))
S_SOFTMAX = {s: softmax_forward_scores(sm_psi, sm_head, inputs_for(sm_names, s), n_items) for s in ("val", "test")}
print(f"softmax DNN (Task 14) trained in {time.time() - t:.0f}s (best epoch {sm_best_ep} of {len(sm_curve)} run)")
```

```python
# ---------- Task 14: NeuMF (GMF branch + MLP branch), the scoring model ----------
def _emb(n, dim, reg, name):
    return L.Embedding(n, dim, embeddings_regularizer=keras.regularizers.l2(reg) if reg else None, name=name)

def gmf_branch(u, i, d, reg):
    pu = L.Flatten()(_emb(n_users, d, reg, "gmf_user_emb")(u)); qi = L.Flatten()(_emb(n_items, d, reg, "gmf_item_emb")(i))
    return L.Multiply(name="gmf_vector")([pu, qi])                # psi_GMF(p, q) = p (.) q

def mlp_branch(u, i, d, reg):
    pu = L.Flatten()(_emb(n_users, 2 * d, reg, "mlp_user_emb")(u)); qi = L.Flatten()(_emb(n_items, 2 * d, reg, "mlp_item_emb")(i))
    x = L.Concatenate(name="mlp_concat")([pu, qi])                # [p ; q]: user and item meet INSIDE the network (this is what a two-tower model avoids)
    for w, nm in ((4 * d, "mlp_1"), (2 * d, "mlp_2"), (d, "mlp_3")):
        x = L.Dense(w, activation="relu", name=nm)(x)
    return x

def build_neumf(d=D_NCF, reg=NEUMF_REG, name="NeuMF"):
    r_g, r_m = reg
    u = keras.Input(shape=(1,), dtype="int32", name="user"); i = keras.Input(shape=(1,), dtype="int32", name="item")
    fused = L.Concatenate(name="neumf_concat")([gmf_branch(u, i, d, r_g), mlp_branch(u, i, d, r_m)])      # [psi_GMF ; psi_MLP]
    return keras.Model([u, i], L.Dense(1, activation="sigmoid", name="out")(fused), name=name)

t = time.time()
keras.utils.set_random_seed(2000)
neumf = build_neumf(); neumf.compile(optimizer=keras.optimizers.Adam(LR_NCF), loss="binary_crossentropy")
neu_curve, neu_best_ep, _ = fit_pairwise(neumf, seed=0)
print(f"NeuMF (Task 14) trained in {time.time() - t:.0f}s (best epoch {neu_best_ep} of {len(neu_curve)} run)")
t = time.time()
S_NEUMF = {"test": pairwise_full_scores(neumf)}                  # the brute-force scan over the WHOLE catalog (610 users x 9,681 items)
print(f"NeuMF scored all {n_users:,} x {n_active:,} (user, movie) pairs in {time.time() - t:.0f}s")

# both branches must carry signal (a collapsed embedding table would silently turn NeuMF into a one-branch model; see the Task 14 notes)
_med = lambda nm: float(np.median(np.linalg.norm(neumf.get_layer(nm).embeddings.numpy()[:n_active], axis=1)))
NEUMF_BRANCH_NORMS = {"GMF-branch item embeddings": _med("gmf_item_emb"), "MLP-branch item embeddings": _med("mlp_item_emb")}
NEUMF_BRANCH_ALIVE = all(v > 0.05 for v in NEUMF_BRANCH_NORMS.values())
print("median item-embedding norms:", {k: round(v, 3) for k, v in NEUMF_BRANCH_NORMS.items()}, "->", "both branches alive" if NEUMF_BRANCH_ALIVE else "WARNING: a branch has collapsed")
```

# Phase 1 — The three-stage architecture

In Tasks 11–14 every model was judged on its own: *given a user, is this model's top-10 any good?* A real product has a different problem — a catalog of millions of items, about 100 ms per request, and rules (never show what the user already saw, keep the list varied, promote new releases) that have nothing to do with model accuracy. No single model can do all that, so large recommenders (YouTube, Google Play, Netflix, …) split the job into **three stages**. Each stage is allowed to be cheap *or* precise over the items it sees, never both over the whole catalog.

| Stage | The question it answers | Works on | Cost per item | Produces |
|---|---|---|---|---|
| **① Retrieval** (candidate generation) | "Which ~100 items could this user plausibly like?" | the **whole catalog** | tiny: one dot product with a *pre-computed* item vector (or less, with an ANN index) | `R` candidates |
| **② Scoring** (ranking) | "How good is each of these candidates, exactly?" | only the **R candidates** | large: a neural network in which user and item interact (NeuMF) | scores → a shortlist of 50 |
| **③ Re-ranking** | "Which 10 should we show, as a *list*?" | only the **50 shortlist** | medium, but acts on the whole list: diversity, freshness, business rules | the final top-10 |

**Why retrieval is fast and coarse.** It runs over every item for every request, so the per-item work must be almost nothing. That forces a *two-tower* shape (Phase 3): the item side is computed **once, offline**, and a request only needs one query vector and a nearest-neighbour lookup (Phase 4). The price is that user and item cannot interact before the final dot product, so it is a rougher judge — it is allowed to be, because its only job is to **not lose the good items** (its quality is measured by *recall*, not by the order of its list).

**Why scoring is slow and precise.** It only sees the ~100 items retrieval kept, so a model that is far too expensive for the whole catalog is affordable here — and it can use *cross-features* (user × item interactions), which is exactly what NeuMF's MLP branch does.

**Why re-ranking is a separate last stage.** "Don't show two near-identical movies back to back" or "give new releases a small push this week" are *properties of a list*, not of one (user, item) pair. A point-wise model trained on labels cannot see them, and encoding them as training labels is brittle (the rule changes every week; retraining does not). As a final pass over 50 items they are trivial to apply, easy to switch off, and easy to A/B test.

### 1.2 Why no single stage can do the whole job — the cost of "score everything"

If NeuMF (the precise model) were used on the whole catalog, every request would need one forward pass per catalog item. The table times exactly that on this machine (a compiled TensorFlow function, one user, `N` items; for the very large `N` the item ids are re-used — the cost per pair does not depend on which id it is).

```python
@tf.function(input_signature=[tf.TensorSpec([None, 1], tf.int32), tf.TensorSpec([None, 1], tf.int32)])
def neumf_serve(u, i):
    """Compiled NeuMF forward pass: (user ids, item ids) -> P(interaction). Used for every timing in this notebook."""
    return neumf([u, i], training=False)

def neumf_score_pairs(user, items, chunk=131072):
    """Scores one user against many items, in chunks (bounded memory)."""
    items = np.asarray(items, dtype=np.int32)
    out = []
    for s in range(0, len(items), chunk):
        it = items[s:s + chunk]
        out.append(neumf_serve(tf.constant(np.full((len(it), 1), user, dtype=np.int32)), tf.constant(it[:, None])).numpy().reshape(-1))
    return np.concatenate(out)

# correctness: the compiled function must reproduce the plain Keras scores
_chk = np.random.default_rng(0).choice(n_active, 500, replace=False)
_a = neumf_score_pairs(7, _chk); _b = _forward(neumf, [np.full(500, 7, np.int32), _chk.astype(np.int32)])
assert np.allclose(_a, _b, atol=1e-5), "compiled NeuMF differs from the Keras model"

rng_t = np.random.default_rng(1)
cost_rows = []
for N in (R, 1_000, n_active, 100_000, 1_000_000):
    items = rng_t.integers(0, n_active, N).astype(np.int32)
    neumf_score_pairs(0, items)                                       # warm-up (graph tracing, memory allocation)
    reps = 20 if N <= 10_000 else (5 if N <= 100_000 else 2)
    ts_ = []
    for _ in range(reps):
        t0 = time.perf_counter(); neumf_score_pairs(0, items); ts_.append((time.perf_counter() - t0) * 1000)
    cost_rows.append({"items scored per request": N, "what that equals": {R: "retrieval shortlist", 1_000: "", n_active: "this catalog (brute force)", 100_000: "10x a mid-size catalog", 1_000_000: "a million-item catalog"}[N],
                      "NeuMF time (ms)": float(np.median(ts_))})
cost_tab = pd.DataFrame(cost_rows)
cost_tab["requests per second (this machine, 1 stream)"] = 1000.0 / cost_tab["NeuMF time (ms)"]
display(cost_tab.style.format({"items scored per request": "{:,.0f}", "NeuMF time (ms)": "{:,.1f}", "requests per second (this machine, 1 stream)": "{:,.1f}"}).hide(axis="index"))
NEUMF_MS = dict(zip(cost_tab["items scored per request"], cost_tab["NeuMF time (ms)"]))
MACS = int(sum(np.prod(l.kernel.shape) for l in neumf.layers if isinstance(l, L.Dense)) + D_NCF)      # multiply-adds per (user, item) pair in this NeuMF
display(Markdown(
    f"Scoring the {R} retrieved candidates takes **{NEUMF_MS[R]:.1f} ms**; scoring the whole {n_active:,}-movie catalog takes **{NEUMF_MS[n_active]:.1f} ms** "
    f"({NEUMF_MS[n_active] / NEUMF_MS[R]:.1f}x more), and a million-item catalog would take **{NEUMF_MS[1_000_000]:,.0f} ms per request** — "
    f"{NEUMF_MS[1_000_000] / 100:.1f}x the whole 100 ms budget, for *one* request, and every page view is a request. Below a few thousand items the call is dominated by a fixed overhead; "
    f"beyond that the time grows in direct proportion to the catalog. And this NeuMF is tiny (only about {MACS:,} multiply-adds per pair); production rankers with hundreds of input features cost far more per pair, "
    f"so the gap there is much larger. The precise model can therefore only ever see a short list, and something much cheaper has to choose that list. "
    f"(For *this* catalog brute force is still affordable — that is exactly why Phase 5 can use it as the reference answer.)"))
```

# Phase 2 — Candidate generation and similarity measures revisited

Retrieval is "find the items whose vectors are closest to the query vector". *Closest* needs a definition, and Task 12's three measures give three different answers:

| Measure | Formula | Reads as |
|---|---|---|
| dot product | `dot(q,x) = ⟨q,x⟩` | direction **and** length of both vectors |
| cosine | `cosine(q,x) = ⟨q,x⟩ / (‖q‖ ‖x‖)` | direction only |
| Euclidean | `euclidean(q,x) = ‖q − x‖` | absolute distance between the two points |

**The key practical fact.** If both vectors are L2-normalised (`‖q‖ = ‖x‖ = 1`) the three coincide: `dot = cosine`, and `½‖q−x‖² = ½(‖q‖² − 2⟨q,x⟩ + ‖x‖²) = 1 − ⟨q,x⟩`, so ranking by smallest distance is ranking by largest dot product. For *un-normalised* vectors, expanding the same square shows how they differ for a fixed query `q` (only the terms that depend on the item `x` matter):

* ranking by **dot** ↔ `⟨q,x⟩`
* ranking by **cosine** ↔ `⟨q,x⟩ / ‖x‖` — the dot product **divided by the item norm** (long vectors lose their advantage)
* ranking by **Euclidean** ↔ `⟨q,x⟩ − ½‖x‖²` — the dot product **minus a norm penalty** (long vectors are punished)

So the three measures are one dot product seen through three different filters of the *item norm* `‖x‖`. That is where the bias in candidate generation comes from.

**The textbook expectation** (to be tested, not assumed): popular items get more training signal, so their vectors tend to be longer; the dot product then ranks them high whatever the fit, which reproduces Task 12's popularity-bias finding at the retrieval stage; cosine ignores length; Euclidean punishes it.

```python
from scipy.stats import spearmanr

# ---- the Task 14 item embeddings: the item matrix V of the softmax DNN (logits = psi(x) . v_item + b_item), plus its query vectors psi(x) for the 610 test queries ----
V14 = sm_head.kernel.numpy().T.astype(np.float32)[:n_active]                 # (n_active, 32)
B14 = sm_head.bias.numpy().astype(np.float32)[:n_active]
Q14 = sm_psi.predict(inputs_for(sm_names, "test"), batch_size=1024, verbose=0).astype(np.float32)   # (610, 32)
# ---- for comparison, the Task 12 BPR-MF embeddings: user factors P (the "queries") and item factors Q ----
VBPR, QBPR = bpr.Q.numpy().astype(np.float32)[:n_active], bpr.P.numpy().astype(np.float32)
EMBEDDINGS = {"Task 14 softmax DNN (item matrix V)": (Q14, V14), "Task 12 BPR-MF (item factors)": (QBPR, VBPR)}

# ---- 1) equivalence on L2-normalised vectors, checked numerically on the real embeddings ----
def unit(a): return a / np.maximum(np.linalg.norm(a, axis=1, keepdims=True), 1e-12)
OK14 = np.linalg.norm(Q14, axis=1) > 1e-8              # a query vector can be exactly 0 when every ReLU unit of the query network is off -> cosine is undefined (0/0)
print(f"{int((~OK14).sum())} of the {len(Q14)} test queries have an all-zero query vector psi(x): for them the softmax logits reduce to the item bias b_item (= a popularity ranking) and cosine is undefined. "
      "They are left out of the similarity comparisons below (a production system needs an explicit guard for this case).")
Qk = Q14[OK14]
qh, vh = unit(Qk), unit(V14)
dot_u = qh @ vh.T
cos_u = (Qk @ V14.T) / (np.linalg.norm(Qk, axis=1, keepdims=True) * np.linalg.norm(V14, axis=1)[None, :])
half_sq = 0.5 * ((qh ** 2).sum(1, keepdims=True) - 2 * dot_u + (vh ** 2).sum(1)[None, :])
print(f"unit vectors: max |dot - cosine| = {np.abs(dot_u - cos_u).max():.2e} | max |1/2||q-x||^2 - (1 - <q,x>)| = {np.abs(half_sq - (1 - dot_u)).max():.2e}")
tops = [np.argsort(-S, axis=1)[:, :10] for S in (dot_u, cos_u, -half_sq)]
print("after L2-normalisation the top-10 lists of dot / cosine / Euclidean are identical for",
      f"{np.mean([len(set(a) | set(b) | set(c)) == len(set(a)) for a, b, c in zip(*tops)]):.0%} of the {len(Qk)} queries")
assert np.abs(dot_u - cos_u).max() < 1e-4 and np.abs(half_sq - (1 - dot_u)).max() < 1e-4
```

```python
# ---- 2) un-normalised vectors: the three measures give different top-10s. Aggregate over the test queries with a non-zero query vector (already-seen movies excluded) ----
SEEN_T = SEEN["test"][:, :n_active]
LOGPOP = np.log1p(train_cnt[:n_active])
TOP10 = {}                                     # (embedding name, measure) -> (610, 10) item ids
rows = []
N_Q = {}
for name, (Qm, Vm) in EMBEDDINGS.items():
    ok_q = np.linalg.norm(Qm, axis=1) > 1e-8; N_Q[name] = int(ok_q.sum())
    Qm, seen_m = Qm[ok_q], SEEN_T[ok_q]
    nrm_v = np.linalg.norm(Vm, axis=1)
    dot = Qm @ Vm.T
    cosm = dot / (np.linalg.norm(Qm, axis=1, keepdims=True) * nrm_v[None, :] + 1e-12)
    euc = -np.sqrt(np.maximum((Qm ** 2).sum(1, keepdims=True) - 2 * dot + (nrm_v ** 2)[None, :], 0.0))
    rho = spearmanr(nrm_v, LOGPOP)[0]
    for meas, S in (("dot product", dot), ("cosine", cosm), ("Euclidean", euc)):
        S = S.copy(); S[seen_m] = -np.inf
        top = np.argsort(-S, axis=1)[:, :10]; TOP10[(name, meas)] = top
        rows.append({"embeddings": name, "measure": meas, "corr(item norm, log popularity)": rho,
                     "median train count of the top-10": float(np.median(train_cnt[top])), "share from the 100 most popular movies": float((top < 100).mean()),
                     "mean item norm of the top-10": float(nrm_v[top].mean()), "median norm, all movies": float(np.median(nrm_v))})
sim_tab = pd.DataFrame(rows)
display(sim_tab.style.format({"corr(item norm, log popularity)": "{:+.2f}", "median train count of the top-10": "{:.0f}", "share from the 100 most popular movies": "{:.1%}",
                              "mean item norm of the top-10": "{:.2f}", "median norm, all movies": "{:.2f}"}).hide(axis="index"))

def jac(a, b): return float(np.mean([len(set(x) & set(y)) / len(set(x) | set(y)) for x, y in zip(a, b)]))
ov = pd.DataFrame([{"embeddings": name, "dot vs cosine": jac(TOP10[(name, "dot product")], TOP10[(name, "cosine")]),
                    "dot vs Euclidean": jac(TOP10[(name, "dot product")], TOP10[(name, "Euclidean")]),
                    "cosine vs Euclidean": jac(TOP10[(name, "cosine")], TOP10[(name, "Euclidean")])} for name in EMBEDDINGS])
print(f"\naverage overlap (Jaccard) of the top-10 lists produced by two measures, over the queries used ({N_Q[list(EMBEDDINGS)[0]]} for Task 14, {N_Q[list(EMBEDDINGS)[1]]} for Task 12):")
display(ov.style.format({c: "{:.2f}" for c in ov.columns[1:]}).hide(axis="index"))
```

### 2.1 One sample query, three measures (the Phase 2 deliverable)

The sample query is the test user with the **median** amount of training history (so it is neither a heavy nor a light user). The table shows the union of the three top-10 lists; `—` means the movie is not in that measure's top-10. Movies already seen by the user are excluded for all three.

```python
act = np.bincount(tr_u, minlength=n_users)
U0 = int(np.argsort(act, kind="stable")[n_users // 2])
assert OK14[U0], "sample query has an all-zero query vector"
q0 = Q14[U0]; nV = np.linalg.norm(V14, axis=1)
s_dot = V14 @ q0
s_cos = s_dot / (np.linalg.norm(q0) * nV + 1e-12)
s_euc = -np.linalg.norm(V14 - q0[None, :], axis=1)
lists, rank_of = {}, {}
for meas, s in (("dot product", s_dot), ("cosine", s_cos), ("Euclidean", s_euc)):
    s = s.copy(); s[SEEN_T[U0]] = -np.inf
    lists[meas] = np.argsort(-s)[:10]
    rank_of[meas] = {int(j): r + 1 for r, j in enumerate(lists[meas])}
union = sorted(set(int(j) for l in lists.values() for j in l), key=lambda j: (min(rank_of[m].get(j, 99) for m in rank_of), j))
sample = pd.DataFrame({
    "movie": [str(movies.title[j])[:46] for j in union], "popularity rank": [j + 1 for j in union], "train count": [int(train_cnt[j]) for j in union],
    "‖v‖ (item norm)": [float(nV[j]) for j in union],
    **{f"rank: {m}": [rank_of[m].get(j, np.nan) for j in union] for m in rank_of}})
hist_titles = [str(movies.title[h - 1])[:40] for h in FEATS["test"]["hist"][U0] if h > 0][-5:]
print(f"sample query: MovieLens user {user_ids[U0]} ({act[U0]} training interactions). Most recent movies before the query: {hist_titles}")
display(sample.style.format({"‖v‖ (item norm)": "{:.2f}", **{f"rank: {m}": "{:.0f}" for m in rank_of}}, na_rep="—").hide(axis="index"))

# which movies are unique to one measure, and what do they have in common?
only = {m: [j for j in union if j in rank_of[m] and not any(j in rank_of[o] for o in rank_of if o != m)] for m in rank_of}
in_all = [j for j in union if all(j in rank_of[m] for m in rank_of)]
def med(js, arr): return float(np.median([arr[j] for j in js])) if js else float("nan")
print(f"\nin all three lists: {len(in_all)} movies | union: {len(union)} movies | only under dot: {len(only['dot product'])}, only under cosine: {len(only['cosine'])}, only under Euclidean: {len(only['Euclidean'])}")
for m in rank_of:
    if only[m]:
        print(f"  only under {m:11s}: " + "; ".join(f"{str(movies.title[j])[:34]} (pop. rank {j + 1}, norm {nV[j]:.2f})" for j in only[m]))
```

```python
# ---- the written explanation, generated from the numbers above so it cannot drift from them ----
r14 = sim_tab[sim_tab.embeddings.str.startswith("Task 14")].set_index("measure")
rb = sim_tab[sim_tab.embeddings.str.startswith("Task 12")].set_index("measure")
rho14, rho12 = r14["corr(item norm, log popularity)"].iloc[0], rb["corr(item norm, log popularity)"].iloc[0]
def share(df, m): return df.loc[m, "share from the 100 most popular movies"]
def direction(rho): return "longer vectors belong to RARER movies" if rho < -0.2 else ("longer vectors belong to MORE POPULAR movies" if rho > 0.2 else "norm and popularity are almost unrelated")
def most_popular_measure(df): return df["share from the 100 most popular movies"].idxmax()
bias_rho14 = spearmanr(B14, LOGPOP)[0]
bias_rho12 = spearmanr(bpr.b.numpy()[:n_active], LOGPOP)[0]
def fmt_med(js, arr, f="{:.2f}"): return f.format(med(js, arr)) if js else "n/a (no such movie)"
RANKS = np.arange(1, n_active + 1)
pairs = {"dot vs cosine": "dot product and cosine", "dot vs Euclidean": "dot product and Euclidean", "cosine vs Euclidean": "cosine and Euclidean"}
close14 = pairs[ov.iloc[0][list(pairs)].astype(float).idxmax()]; close12 = pairs[ov.iloc[1][list(pairs)].astype(float).idxmax()]
txt = f"""
**What the numbers say**

* **Norm and popularity.** For the Task 14 item matrix the rank correlation between item norm and log-popularity is **{rho14:+.2f}** ({direction(rho14)}); for Task 12's BPR-MF item factors it is **{rho12:+.2f}** ({direction(rho12)}). The textbook picture ("popular = long vector") is therefore **not a law** — it depends on where a model *stores* popularity. Both models also have an item bias `b_item` (rank correlation with log-popularity: softmax DNN **{bias_rho14:+.2f}**, BPR-MF **{bias_rho12:+.2f}**); a plain dot product `⟨q,v⟩` ignores that bias, so whatever popularity the bias carries is *not* in the vector — a plausible reason why the softmax DNN's rare movies need *longer* vectors to win when they fit. Always measure this correlation on your own embedding table before choosing the retrieval metric.
* **Which measure favours popular movies.** Share of top-10 slots filled by the 100 most popular movies — Task 14 embeddings ({N_Q[list(EMBEDDINGS)[0]]} queries with a non-zero query vector): dot **{share(r14, 'dot product'):.1%}**, cosine **{share(r14, 'cosine'):.1%}**, Euclidean **{share(r14, 'Euclidean'):.1%}** (the most popularity-heavy measure is *{most_popular_measure(r14)}*); Task 12 embeddings ({N_Q[list(EMBEDDINGS)[1]]} queries): dot **{share(rb, 'dot product'):.1%}**, cosine **{share(rb, 'cosine'):.1%}**, Euclidean **{share(rb, 'Euclidean'):.1%}** (most popularity-heavy: *{most_popular_measure(rb)}*). The textbook claim (the dot product pulls toward popular movies) {"holds" if most_popular_measure(rb) == "dot product" else "does not hold"} for the BPR-MF factors (longer vector = more popular movie) and {"holds" if most_popular_measure(r14) == "dot product" else "does not hold"} for the softmax DNN. Cosine divides by the norm and Euclidean subtracts `½‖x‖²`, so they push the list in the opposite direction to the dot product's norm preference: the sign of the norm–popularity correlation decides whether that means more or fewer popular movies.
* **The lists really differ.** Average top-10 overlap (Jaccard; 1.0 = identical) — Task 14: dot vs cosine **{ov.iloc[0]['dot vs cosine']:.2f}**, dot vs Euclidean **{ov.iloc[0]['dot vs Euclidean']:.2f}**, cosine vs Euclidean **{ov.iloc[0]['cosine vs Euclidean']:.2f}** (closest pair: {close14}); Task 12: **{ov.iloc[1]['dot vs cosine']:.2f}**, **{ov.iloc[1]['dot vs Euclidean']:.2f}**, **{ov.iloc[1]['cosine vs Euclidean']:.2f}** (closest pair: {close12}). Which two measures agree changes from one embedding table to the other, so the choice of measure is a real modelling decision, not a detail.
* **The sample query.** {len(in_all)} of the {len(union)} movies are in all three lists and {sum(len(v) for v in only.values())} appear under one measure only (dot only: {len(only['dot product'])}, cosine only: {len(only['cosine'])}, Euclidean only: {len(only['Euclidean'])}). The dot-product-only movies have a median item norm of {fmt_med(only['dot product'], nV)} (median popularity rank {fmt_med(only['dot product'], RANKS, '{:.0f}')}); the cosine-only movies {fmt_med(only['cosine'], nV)} (rank {fmt_med(only['cosine'], RANKS, '{:.0f}')}). They moved because cosine **divides** the dot product by the item norm — short vectors are lifted, long vectors demoted — whereas the Euclidean distance only **subtracts** `½‖x‖²`, a much gentler correction here, so its list stays closer to the dot product's.
* **What this means for retrieval.** The metric used to *search* must be the metric the model was *trained* with. A model trained with a dot product (the two-tower model of Phase 3) should be searched by inner product; normalising the vectors (cosine / Euclidean) is a modelling decision that has to be made at training time, otherwise the index silently answers a different question. Phase 3 checks this empirically on the two-tower embeddings (§3.5).
"""
display(Markdown(txt))
```

# Phase 3 — Two-tower retrieval models

### 3.1 The idea

A **two-tower model** has two separate neural networks that only meet at the very end:

* the **query tower** `q = f(user, context)` — here: user id + the user's recent genre mix + a bag of their last `H = 10` movies (the side features of Task 14, Phase 6) → a `d`-dimensional vector;
* the **candidate tower** `c = g(item)` — here: item id + the Task 11 content vector (TF-IDF genres + decade) → a `d`-dimensional vector.

The score of a pair is just `⟨q, c⟩`. Training pushes `⟨q, c⁺⟩` of the movie the user really watched above `⟨q, c⁻⟩` of other movies (§3.4).

### 3.2 Why this shape is the one retrieval needs

Because the towers do not interact until that final dot product, **`g(item)` does not depend on the query**. Every item vector can be computed **once, offline, for the whole catalog**, put in an index, and reused for every request. A request then costs *one* query-tower pass plus a nearest-neighbour search — independent of how many items exist (Phase 4 makes the search sub-linear).

NeuMF (Task 14) is built the other way round: its MLP branch takes `[user vector ; item vector]` **together** as input, so "the NeuMF score" does not exist until both the user *and* the item are fed in. Nothing about the item can be pre-computed on its own; scoring a new user against the whole catalog means one forward pass *per item* (the cost table of §1.2). That is why NeuMF is a **scorer** and the two-tower model is a **retriever**.

| | Two-tower | NeuMF |
|---|---|---|
| score of (user, item) | `⟨f(user), g(item)⟩` | `MLP([p_user ; q_item])` (+ GMF branch) |
| item side computable before any request? | **yes** → index it | no |
| cost to rank the whole catalog for one user | 1 query pass + 1 vector search | one network pass per item |
| can learn "this user *and* this item *together*" (cross-features)? | only through the final dot product | yes — that is the MLP's job |
| role | **retrieval** | **scoring** |

```python
D_TT = 32                              # embedding size of both towers (as in the task description)
TT_HIDDEN = 64
TT_OOV = n_items                       # an extra "unknown item" id: used for brand-new items and as id-dropout during training (Exercise 4)
TT_NEG = 1024                          # sampled negatives per training step (shared by the whole batch)
TT_PDROP = 0.10                        # probability that the candidate tower sees the OOV id instead of the real id in training
TT_QUERY_INPUTS = ["user", "genre_mix", "hist"]
F_TF = tf.constant(F)                  # (n_items, 44) Task 11 content vectors, gathered inside the training step

def build_query_tower(d=D_TT, hidden=TT_HIDDEN):
    u = keras.Input(shape=(1,), dtype="int32", name="user")
    g = keras.Input(shape=(len(GENRES),), name="genre_mix")
    h = keras.Input(shape=(H,), dtype="int32", name="hist")
    ue = L.Flatten()(L.Embedding(n_users, d, name="user_emb")(u))
    he = MaskedMean(name="hist_bag")([L.Embedding(n_items + 1, d, name="hist_item_emb")(h), h])
    x = L.Dense(hidden, activation="relu", name="q_hidden")(L.Concatenate(name="q_concat")([ue, g, he]))
    return keras.Model([u, g, h], L.Dense(d, name="query_embedding")(x), name="query_tower")        # linear output: embeddings may point in any direction

def build_candidate_tower(d=D_TT, hidden=TT_HIDDEN):
    i = keras.Input(shape=(1,), dtype="int32", name="item")
    c = keras.Input(shape=(F.shape[1],), name="content")
    ie = L.Flatten()(L.Embedding(n_items + 1, d, name="item_emb")(i))
    x = L.Dense(hidden, activation="relu", name="c_hidden")(L.Concatenate(name="c_concat")([ie, c]))
    return keras.Model([i, c], L.Dense(d, name="candidate_embedding")(x), name="candidate_tower")

keras.utils.set_random_seed(7)
query_tower, candidate_tower = build_query_tower(), build_candidate_tower()
print(f"query tower: {query_tower.count_params():,} parameters (inputs: user id, recent genre mix, last-{H} history bag) | candidate tower: {candidate_tower.count_params():,} parameters (inputs: item id, content vector)")
```

### 3.3 Training: a sampled-softmax retrieval loss (Task 14, Phase 3, applied to two towers)

A full softmax over the catalog would need the candidate tower on **all** 9,681 movies for every batch. The sampled softmax of Task 14 avoids that: for each batch,

1. take the true movie `i⁺` of every row, and draw `S = 1,024` negative movies from a **log-uniform (Zipfian) sampler** over the popularity-ordered ids — frequent movies are drawn more often, as in the real item distribution;
2. run the candidate tower on those `B + S` movies only, and form logits `s(q, i) = ⟨q, c_i⟩ − log Q(i)`, where `Q(i)` is the probability that the sampler draws movie `i`. The `− log Q(i)` **correction** stops popular movies from being punished merely because they are sampled more often (without it the model would be biased *against* popularity);
3. remove *accidental hits* (a "negative" that happens to be the row's own positive), and apply the softmax cross-entropy with the true movie as the target.

**Item-id dropout.** With probability `0.10` the candidate tower is given the *unknown-item id* instead of the real id, so it also learns to build a vector from the **content features alone**. That is what lets a brand-new movie (no interactions, no trained id vector) be embedded and indexed in Exercise 4.

Early stopping uses **validation Recall@100** (full ranking, seen movies excluded) — the quantity a retrieval stage is for — and the best epoch's weights are restored.

```python
def cand_forward(ids, training):
    """Candidate tower on item ids (+ their content vectors). In training, each id is replaced by the OOV id with probability TT_PDROP."""
    ids = tf.cast(ids, tf.int32)
    in_ids = tf.where(tf.random.uniform(tf.shape(ids)) < TT_PDROP, tf.fill(tf.shape(ids), TT_OOV), ids) if (training and TT_PDROP > 0) else ids
    return candidate_tower([in_ids[:, None], tf.gather(F_TF, ids)], training=training)

tt_opt = keras.optimizers.Adam(2e-3)
tt_vars = query_tower.trainable_variables + candidate_tower.trainable_variables

@tf.function
def tt_train_step(user, genre, hist, pos):
    pos64 = tf.cast(pos, tf.int64)
    sampled, true_ec, samp_ec = tf.random.log_uniform_candidate_sampler(true_classes=pos64[:, None], num_true=1, num_sampled=TT_NEG, unique=True, range_max=n_active)
    sampled = tf.cast(sampled, tf.int32)
    with tf.GradientTape() as tape:
        q = query_tower([user, genre, hist], training=True)                               # (B, d)
        cp = cand_forward(pos, True)                                                       # (B, d)   true movies
        cn = cand_forward(sampled, True)                                                   # (S, d)   sampled negatives, shared by the batch
        pos_logit = tf.reduce_sum(q * cp, 1) - tf.math.log(true_ec[:, 0])                  # <q,c+> - log Q(i+)
        neg_logit = tf.matmul(q, cn, transpose_b=True) - tf.math.log(samp_ec)[None, :]     # <q,c-> - log Q(i-)
        neg_logit = tf.where(tf.equal(pos[:, None], sampled[None, :]), tf.constant(-1e9, neg_logit.dtype), neg_logit)    # accidental hits
        logits = tf.concat([pos_logit[:, None], neg_logit], axis=1)
        loss = tf.reduce_mean(tf.nn.sparse_softmax_cross_entropy_with_logits(labels=tf.zeros_like(pos), logits=logits))
    tt_opt.apply_gradients(zip(tape.gradient(loss, tt_vars), tt_vars))
    return loss

def tt_item_matrix(ids=None, batch=4096):
    """Inference path of the candidate tower (no id-dropout) for `ids` (default: the whole catalog; items with NO training interaction use the OOV id + content)."""
    ids = np.arange(n_items) if ids is None else np.asarray(ids)
    use_id = np.where(train_cnt[ids] > 0, ids, TT_OOV).astype(np.int32)
    out = [candidate_tower.predict([use_id[s:s + batch, None], F[ids[s:s + batch]]], batch_size=batch, verbose=0) for s in range(0, len(ids), batch)]
    return np.vstack(out).astype(np.float32)

def tt_queries(split):
    return query_tower.predict([FEATS[split][k] for k in TT_QUERY_INPUTS], batch_size=1024, verbose=0).astype(np.float32)

def tt_scores(split, E=None, Q=None):
    E = tt_item_matrix(np.arange(n_active)) if E is None else E[:n_active]
    Q = tt_queries(split) if Q is None else Q
    S = np.full((n_users, n_items), -np.inf, dtype=np.float32); S[:, :n_active] = Q @ E.T
    return S

def tt_val_recall100():
    r = ranks_from_scores(tt_scores("val"), VAL_ITEM, SEEN_FOR_VAL)
    return metrics_from_ranks(r, (100,))["HR@100"]

Xtt = [FEATS["train"][k][KEEP] for k in TT_QUERY_INPUTS]; ytt = tr_i[KEEP].astype(np.int32)
rs_tt = np.random.default_rng(11); tf.random.set_seed(11)
best, best_w, best_ep, bad, tt_rows = -1.0, None, 0, 0, []
t_start = time.time()
for ep in range(1, (MAX_EPOCHS if not QUICK else 3) + 1):
    t0 = time.time(); p = rs_tt.permutation(len(ytt)); ls = []
    for s in range(0, len(p), 512):
        sl = p[s:s + 512]
        ls.append(float(tt_train_step(*[tf.constant(a[sl]) for a in Xtt], tf.constant(ytt[sl]))))
    v = tt_val_recall100()
    tt_rows.append({"epoch": ep, "train loss": float(np.mean(ls)), "val Recall@100": v, "sec": time.time() - t0})
    if v > best + 1e-6: best, best_w, best_ep, bad = v, query_tower.get_weights() + candidate_tower.get_weights(), ep, 0
    else: bad += 1
    if bad >= PATIENCE + 1: break
nq = len(query_tower.get_weights()); query_tower.set_weights(best_w[:nq]); candidate_tower.set_weights(best_w[nq:])
tt_curve = pd.DataFrame(tt_rows)
print(f"two-tower trained in {time.time() - t_start:.0f}s: best epoch {best_ep} of {len(tt_curve)} run (validation Recall@100 = {best:.4f})")
display(tt_curve.style.format({"train loss": "{:.3f}", "val Recall@100": "{:.4f}", "sec": "{:.1f}"}).hide(axis="index"))
```

### 3.4 Retrieval quality — against the retrievers of Tasks 11–14

`Recall@R` = share of the 610 test users whose held-out movie is among the model's `R` best (seen movies excluded, full catalog). **`Recall@100` is the number that matters for this stage**: it is the ceiling for everything downstream. Intervals are 95% bootstrap CIs over users; the last two columns are paired bootstrap differences against the two-tower model.

```python
E_ACT = tt_item_matrix(np.arange(n_active))
Q_TT = {s: tt_queries(s) for s in ("val", "test")}
S_TT = {s: tt_scores(s, E_ACT, Q_TT[s]) for s in ("val", "test")}
tt_rank = ranks_from_scores(S_TT["test"], TEST_ITEM, SEEN_FOR_TEST)

rows = []
cands = {"Two-tower (this task)": tt_rank, **{nm: ZOO_RANKS[nm] for nm in ("Popularity", "BPR-MF (Task 12)", "Softmax DNN (Task 14)", "NeuMF (Task 14)", "Content-based (Task 11)")}}
for nm, r in cands.items():
    hit100 = (r <= 100).astype(float)
    lo, hi = boot_ci(hit100)
    d_, dlo, dhi = paired_boot_diff((tt_rank <= 100).astype(float), hit100) if nm != "Two-tower (this task)" else (np.nan, np.nan, np.nan)
    row = {"retriever": nm, **{f"Recall@{k}": float((r <= k).mean()) for k in (10, 50, 100, 200, 500)}, "Recall@100 95% CI": f"[{lo:.3f}, {hi:.3f}]",
           "two-tower minus this (Recall@100)": d_, "paired CI": f"[{dlo:+.3f}, {dhi:+.3f}]" if nm != "Two-tower (this task)" else ""}
    rows.append(row)
tt_tab = pd.DataFrame(rows)
display(tt_tab.style.format({**{f"Recall@{k}": "{:.3f}" for k in (10, 50, 100, 200, 500)}, "two-tower minus this (Recall@100)": "{:+.3f}"}, na_rep="").hide(axis="index"))

TT_R100 = float((tt_rank <= 100).mean())
tt_m = metrics_from_ranks(tt_rank, (10, 100)); TT_NDCG10 = tt_m["NDCG@10"]
d_bpr = paired_boot_diff((tt_rank <= 100).astype(float), (ZOO_RANKS["BPR-MF (Task 12)"] <= 100).astype(float))
d_sm = paired_boot_diff((tt_rank <= 100).astype(float), (ZOO_RANKS["Softmax DNN (Task 14)"] <= 100).astype(float))
d_pop = paired_boot_diff((tt_rank <= 100).astype(float), (ZOO_RANKS["Popularity"] <= 100).astype(float))
def verdict(dd):
    return "clearly above" if dd[1] > 0 else ("clearly below" if dd[2] < 0 else "not distinguishable from")
display(Markdown(
    f"**Reading.** The two-tower model's Recall@100 is **{TT_R100:.3f}** (it also reaches NDCG@10 {TT_NDCG10:.4f} when used as a stand-alone ranker, with no scorer at all). "
    f"It is **{verdict(d_pop)}** popularity (Δ {d_pop[0]:+.3f}, CI [{d_pop[1]:+.3f}, {d_pop[2]:+.3f}]), **{verdict(d_bpr)}** BPR-MF (Δ {d_bpr[0]:+.3f}, CI [{d_bpr[1]:+.3f}, {d_bpr[2]:+.3f}]) "
    f"and **{verdict(d_sm)}** the softmax DNN (Δ {d_sm[0]:+.3f}, CI [{d_sm[1]:+.3f}, {d_sm[2]:+.3f}]) at Recall@100. "
    f"Accuracy is only part of the argument: the two-tower **shape** also gives a pre-computable item side (§3.2) and, through the content vector, a way to embed movies that have no interactions (Exercise 4) — which the plain look-up item matrix of the Task 14 softmax DNN cannot do."))
```

### 3.5 Which similarity should the index use? (Phase 2, checked on the two-tower vectors)

The model was **trained** with the dot product. The table searches the same vectors by dot product, cosine and (negative) Euclidean distance. Validation decides, test reports.

```python
def three_measures(Q, E):
    nq_, ne_ = np.linalg.norm(Q, axis=1, keepdims=True), np.linalg.norm(E, axis=1)[None, :]
    dot = Q @ E.T
    return {"dot product (as trained)": dot, "cosine": dot / np.maximum(nq_ * ne_, 1e-12), "Euclidean": -np.sqrt(np.maximum(nq_ ** 2 - 2 * dot + ne_ ** 2, 0.0))}

def pad_scores(S):
    out = np.full((n_users, n_items), -np.inf, dtype=np.float32); out[:, :n_active] = S; return out

rows, rk = [], {}
meas_val, meas_test = three_measures(Q_TT["val"], E_ACT), three_measures(Q_TT["test"], E_ACT)
for nm in meas_val:
    rv = ranks_from_scores(pad_scores(meas_val[nm]), VAL_ITEM, SEEN_FOR_VAL)
    rt = ranks_from_scores(pad_scores(meas_test[nm]), TEST_ITEM, SEEN_FOR_TEST); rk[nm] = rt
    lo, hi = boot_ci((rt <= 100).astype(float))
    rows.append({"measure used for search": nm, "val Recall@100": float((rv <= 100).mean()), "test Recall@100": float((rt <= 100).mean()), "test 95% CI": f"[{lo:.3f}, {hi:.3f}]",
                 "test NDCG@10": metrics_from_ranks(rt, (10,))["NDCG@10"]})
meas_tab = pd.DataFrame(rows)
display(meas_tab.style.format({"val Recall@100": "{:.3f}", "test Recall@100": "{:.3f}", "test NDCG@10": "{:.4f}"}).hide(axis="index"))

nrm_e = np.linalg.norm(E_ACT, axis=1)
rho_tt = spearmanr(nrm_e, LOGPOP)[0]
best_val_measure = meas_tab.loc[meas_tab["val Recall@100"].idxmax(), "measure used for search"]
dd_c = paired_boot_diff((rk["dot product (as trained)"] <= 100).astype(float), (rk["cosine"] <= 100).astype(float))
dd_e = paired_boot_diff((rk["dot product (as trained)"] <= 100).astype(float), (rk["Euclidean"] <= 100).astype(float))
display(Markdown(
    f"**Reading.** On the validation split the best measure is **{best_val_measure}**. On test, dot minus cosine = {dd_c[0]:+.3f} Recall@100 (paired CI [{dd_c[1]:+.3f}, {dd_c[2]:+.3f}]) and dot minus Euclidean = {dd_e[0]:+.3f} "
    f"([{dd_e[1]:+.3f}, {dd_e[2]:+.3f}]). The two-tower item vectors have **no bias term**, so a movie's popularity can only live in its vector: the rank correlation between item norm and log-popularity is **{rho_tt:+.2f}** "
    f"({'longer vectors = more popular movies, the textbook case' if rho_tt > 0.2 else ('longer vectors = rarer movies' if rho_tt < -0.2 else 'norm and popularity are almost unrelated')}). "
    f"The index in Phase 4 therefore uses **inner product**, the measure the model was trained with."))
```

Every movie of the catalog — including the movies with no training interaction, which go through the content-only (OOV id) path — is embedded once and written to disk; the query tower and the trained weights are saved next to it so the pipeline can be re-loaded without re-training.

# Phase 4 — Approximate nearest-neighbour (ANN) search

### 4.1 Why exact search stops scaling

Exact retrieval compares the query with **every** item vector: `O(n·d)` multiply-adds per request, and — more importantly on real hardware — the whole `n × d` matrix has to be streamed through memory each time. With `n = 1,000,000` movies and `d = 32` that is 128 MB read per request (on the order of 10 ms on one core; the scale test in 4.6 measures it); with `n = 100 M` items and `d = 128` it is 51 GB per request, for every request, thousands of times per second. Splitting the work across more machines only multiplies the bill: the *work per request* is still proportional to the catalog.

### 4.2 The ANN idea

Build an **index** ahead of time that lets a query look at only a small, well-chosen part of the catalog, and accept that the answer may miss a few of the true top-k:

* **IVFFlat** (inverted file): k-means groups the item vectors into `nlist` cells. A query finds its `nprobe` nearest cells and scans only the vectors inside them (≈ `nprobe / nlist` of the catalog). Bigger `nprobe` → higher recall, slower.
* **HNSW** (hierarchical navigable small-world graph): items are nodes of a layered neighbour graph; a query walks greedily from a coarse top layer down to the finest one. `efSearch` plays the role of `nprobe` (Exercise 1 compares the two).

Quality is measured by **recall@100 of the index**: of the true top-100 (from exact search), how many did the ANN index return?

### 4.3 Build the indexes over the Phase 3 item embeddings

```python
import faiss
from threadpoolctl import threadpool_limits

D_IDX = D_TT
E_IDX = np.ascontiguousarray(ITEM_EMB[:n_active].astype("float32"))       # ids 0..n_active-1 == movie ids (the movies the scoring model knows)
TOPK = 100

# gotcha in the reference snippet of the task: IndexIVFFlat defaults to the L2 metric even when its quantizer is an inner-product index
_snippet_default = faiss.IndexIVFFlat(faiss.IndexFlatIP(D_IDX), D_IDX, 100)
print("reference-snippet IndexIVFFlat metric is L2:", _snippet_default.metric_type == faiss.METRIC_L2, "-> we pass METRIC_INNER_PRODUCT explicitly (the model was trained with a dot product)")

faiss.omp_set_num_threads(2); t0 = time.perf_counter()
flat = faiss.IndexFlatIP(D_IDX); flat.add(E_IDX); t_flat = time.perf_counter() - t0
NLIST = int(round(np.sqrt(n_active)))                                      # rule of thumb: nlist ~ sqrt(n)  (98 -> 100)
t0 = time.perf_counter()
ivf = faiss.IndexIVFFlat(faiss.IndexFlatIP(D_IDX), D_IDX, NLIST, faiss.METRIC_INNER_PRODUCT)
ivf.train(E_IDX); ivf.add(E_IDX); t_ivf = time.perf_counter() - t0
faiss.omp_set_num_threads(1)                                               # latency of ONE request: one thread per query (a server runs many requests in parallel instead)
sz = lambda idx: faiss.serialize_index(idx).nbytes / 1e6
cell_sizes = np.array([ivf.invlists.list_size(c) for c in range(NLIST)])
print(f"IndexFlatIP    : exact, {flat.ntotal:,} vectors, built in {t_flat * 1000:.0f} ms, {sz(flat):.2f} MB")
print(f"IndexIVFFlat   : nlist={NLIST}, {ivf.ntotal:,} vectors, trained+built in {t_ivf * 1000:.0f} ms, {sz(ivf):.2f} MB | vectors per cell: min {cell_sizes.min()}, median {int(np.median(cell_sizes))}, max {cell_sizes.max()}")
```

### 4.4 Latency and recall: ANN vs exact vs a brute-force NumPy scan

Every query is timed **separately** (single thread, as in a real request path) after a warm-up, with the 610 test-user query embeddings; the table gives the median (p50) and p95. *Index recall@100* compares the ANN answer with the exact FAISS answer for the same query. The NumPy baseline is what most people write first: `scores = E @ q`, `argpartition` for the top-100, sort.

```python
def numpy_top100(q1):
    s = E_IDX @ q1[0]
    idx = np.argpartition(-s, TOPK)[:TOPK]
    return idx[np.argsort(-s[idx])]

def time_each(fn, Q, warm=30):
    for q in Q[:warm]: fn(q[None, :])
    out = np.empty(len(Q))
    for k, q in enumerate(Q):
        q1 = q[None, :]; t0 = time.perf_counter(); fn(q1); out[k] = (time.perf_counter() - t0) * 1000
    return out

def pct(a): return {"p50": float(np.percentile(a, 50)), "p95": float(np.percentile(a, 95)), "p99": float(np.percentile(a, 99))}

def index_recall(I_ann, I_ex):
    return float(np.mean([len(set(a) & set(b)) / TOPK for a, b in zip(I_ann, I_ex)]))

QT, QV = Q_TT["test"], Q_TT["val"]
_, I_ex_test = flat.search(QT, TOPK); _, I_ex_val = flat.search(QV, TOPK)
# sanity: exact FAISS == exact NumPy
_np_ids = np.array([numpy_top100(q[None, :]) for q in QT[:50]])
assert all(set(a) == set(b) for a, b in zip(_np_ids, I_ex_test[:50])), "FAISS flat and NumPy disagree"

with threadpool_limits(limits=1):
    lat_np = time_each(numpy_top100, QT)
lat_flat = time_each(lambda q1: flat.search(q1, TOPK), QT)

NPROBES = [1, 2, 4, 8, 16, 32, 64, NLIST]
rows = [{"method": "NumPy brute force (E @ q + argpartition)", "nprobe": "—", "cells scanned": "100%", **{f"latency {k} (ms)": v for k, v in pct(lat_np).items()}, "index recall@100 (test)": 1.0, "index recall@100 (val)": 1.0},
        {"method": "FAISS IndexFlatIP (exact)", "nprobe": "—", "cells scanned": "100%", **{f"latency {k} (ms)": v for k, v in pct(lat_flat).items()}, "index recall@100 (test)": 1.0, "index recall@100 (val)": 1.0}]
IVF_LAT = {}
for npb in NPROBES:
    ivf.nprobe = npb
    lat = time_each(lambda q1: ivf.search(q1, TOPK), QT); IVF_LAT[npb] = lat
    _, I_t = ivf.search(QT, TOPK); _, I_v = ivf.search(QV, TOPK)
    rows.append({"method": "FAISS IndexIVFFlat", "nprobe": npb, "cells scanned": f"{npb / NLIST:.0%}", **{f"latency {k} (ms)": v for k, v in pct(lat).items()},
                 "index recall@100 (test)": index_recall(I_t, I_ex_test), "index recall@100 (val)": index_recall(I_v, I_ex_val)})
ann_tab = pd.DataFrame(rows)
display(ann_tab.style.format({"latency p50 (ms)": "{:.3f}", "latency p95 (ms)": "{:.3f}", "latency p99 (ms)": "{:.3f}", "index recall@100 (test)": "{:.3f}", "index recall@100 (val)": "{:.3f}"}).hide(axis="index"))
assert abs(ann_tab.loc[ann_tab.nprobe == NLIST, "index recall@100 (test)"].iloc[0] - 1.0) < 1e-9, "scanning all cells must reproduce the exact answer"
```

```python
# ---- choose nprobe on the VALIDATION queries with a rule stated in advance: the cheapest setting whose index recall@100 is at least 0.95 ----
RECALL_TARGET = 0.95
ivf_rows = ann_tab[ann_tab.method == "FAISS IndexIVFFlat"]
NPROBE = int(ivf_rows[ivf_rows["index recall@100 (val)"] >= RECALL_TARGET].nprobe.iloc[0])
ivf.nprobe = NPROBE
pick = ivf_rows[ivf_rows.nprobe == NPROBE].iloc[0]
np_row, fl_row = ann_tab.iloc[0], ann_tab.iloc[1]
print(f"chosen: nprobe = {NPROBE} (scans {NPROBE / NLIST:.0%} of the cells). Test index recall@100 = {pick['index recall@100 (test)']:.3f}; p50 latency {pick['latency p50 (ms)']:.3f} ms "
      f"vs NumPy brute force {np_row['latency p50 (ms)']:.3f} ms and FAISS exact {fl_row['latency p50 (ms)']:.3f} ms on this {n_active:,}-movie catalog "
      f"(speed-up vs NumPy: {np_row['latency p50 (ms)'] / pick['latency p50 (ms)']:.1f}x).")
```

### 4.5 Does missing some of the "true top-100" matter? — the production path with the seen-movie filter

*Index recall* only says how close ANN is to exact search. What the **user** cares about is whether the movie they actually watched next is still among the 100 candidates. The production path also has to **skip movies the user has already seen**; FAISS can do that *inside* the search with an ID selector (so we still get 100 unseen candidates rather than throwing away some afterwards). The table runs exactly that, per user, and measures **held-out Recall@100** (test split) for the exact index and for IVFFlat at each `nprobe`.

```python
SEEN_IDS = {sp: [np.flatnonzero(SEEN[sp][u, :n_active]).astype("int64") for u in range(n_users)] for sp in ("val", "test")}

def make_sel(seen_ids):
    return faiss.IDSelectorNot(faiss.IDSelectorBatch(seen_ids))              # "allow everything except these ids"

WIDEN = {"searches": 0}                                                       # how many IVF searches had to scan more cells to find k unseen movies (ensure_k=True only)

def search_filtered(index, q1, seen_ids, k=TOPK, nprobe=None, ensure_k=False):
    """Index search that skips the user's seen movies inside the search. With ensure_k=True an IVF search whose nprobe cells hold fewer than k unseen movies is repeated with twice as many cells
    (until k movies are found or every cell is scanned); without it the missing slots stay -1, which is what the nprobe sweeps of Phase 4 want to measure."""
    sel = make_sel(seen_ids)
    npb = nprobe or (index.nprobe if isinstance(index, faiss.IndexIVF) else None)
    widened = False
    while True:
        if isinstance(index, faiss.IndexIVF): params = faiss.SearchParametersIVF(sel=sel, nprobe=npb)
        elif isinstance(index, faiss.IndexHNSW): params = faiss.SearchParametersHNSW(sel=sel, efSearch=nprobe or index.hnsw.efSearch)
        else: params = faiss.SearchParameters(sel=sel)
        D, I = index.search(q1, k, params=params)
        if not (ensure_k and isinstance(index, faiss.IndexIVF)) or (I >= 0).all() or npb >= index.nlist:
            if widened: WIDEN["searches"] += 1
            return D, I
        npb = min(index.nlist, npb * 2); widened = True

def retrieve_all(index, Q, split, nprobe=None, timed=False, k=TOPK, ensure_k=False):
    ids = np.full((len(Q), k), -1, dtype=np.int64); lat = np.empty(len(Q))
    for u in range(len(Q)):
        t0 = time.perf_counter(); _, I = search_filtered(index, Q[u:u + 1], SEEN_IDS[split][u], k=k, nprobe=nprobe, ensure_k=ensure_k); lat[u] = (time.perf_counter() - t0) * 1000
        ids[u] = I[0]
    return (ids, lat) if timed else ids

def heldout_recall(ids, target):
    return float((ids == np.asarray(target)[:, None]).any(1).mean())

ids_exact_f, lat_exact_f = retrieve_all(flat, QT, "test", timed=True)
assert not np.isin(ids_exact_f[0], SEEN_IDS["test"][0]).any() and (ids_exact_f >= 0).all(), "the seen-movie filter let a seen movie through"
rows = [{"index": "exact (IndexFlatIP)", "nprobe": "—", "held-out Recall@100": heldout_recall(ids_exact_f, TEST_ITEM), "overlap with exact top-100": 1.0, "latency p50 incl. filter (ms)": float(np.median(lat_exact_f))}]
FILTERED = {"exact": ids_exact_f}
for npb in NPROBES:
    ids_f, lat_f = retrieve_all(ivf, QT, "test", nprobe=npb, timed=True); FILTERED[npb] = ids_f
    rows.append({"index": "IndexIVFFlat", "nprobe": npb, "held-out Recall@100": heldout_recall(ids_f, TEST_ITEM), "overlap with exact top-100": index_recall(ids_f, ids_exact_f),
                 "latency p50 incl. filter (ms)": float(np.median(lat_f))})
filt_tab = pd.DataFrame(rows)
display(filt_tab.style.format({"held-out Recall@100": "{:.3f}", "overlap with exact top-100": "{:.3f}", "latency p50 incl. filter (ms)": "{:.3f}"}).hide(axis="index"))
R_EXACT_F = float(filt_tab.iloc[0]["held-out Recall@100"]); R_IVF_F = float(filt_tab[filt_tab.nprobe == NPROBE]["held-out Recall@100"].iloc[0])
lo_e, hi_e = boot_ci((ids_exact_f == TEST_ITEM[:, None]).any(1).astype(float));
dd_ivf = paired_boot_diff((FILTERED[NPROBE] == TEST_ITEM[:, None]).any(1).astype(float), (ids_exact_f == TEST_ITEM[:, None]).any(1).astype(float))
print(f"\nexact search: held-out Recall@100 = {R_EXACT_F:.3f} [{lo_e:.3f}, {hi_e:.3f}] | IVFFlat nprobe={NPROBE}: {R_IVF_F:.3f} (paired difference vs exact {dd_ivf[0]:+.3f}, CI [{dd_ivf[1]:+.3f}, {dd_ivf[2]:+.3f}])")
```

### 4.6 Does ANN pay off? Latency against catalog size

On a **9.7k-movie** catalog almost nothing can be gained from an index — the whole matrix fits in the CPU cache and a plain scan is already microseconds-fast. The point of ANN is what happens when the catalog grows, so the benchmark below makes the catalog bigger: `n` = 10k, 100k and 1M synthetic items, built by re-sampling the real item vectors and adding noise (so the distribution looks like the real one; **recall numbers at scale are indicative, latency numbers are what this test is for**). For each size the IVF index uses `nlist ≈ √n` and the smallest `nprobe` that reaches index recall ≥ 0.95 on 200 queries.

```python
def synth_catalog(n, seed=0):
    rs = np.random.default_rng(seed)
    base = E_IDX[rs.integers(0, n_active, n)]
    sd = E_IDX.std(0, keepdims=True)
    return (base + 0.25 * sd * rs.standard_normal((n, D_IDX)).astype(np.float32)).astype(np.float32)

SCALE_N = [10_000, 100_000] if QUICK else [10_000, 100_000, 1_000_000]
QS = QT[:200]
scale_rows = []
SCALE_SYNTH = {}
for n in SCALE_N:
    X = synth_catalog(n)
    faiss.omp_set_num_threads(2)
    fl = faiss.IndexFlatIP(D_IDX); fl.add(X)
    nl = int(round(np.sqrt(n))); t0 = time.perf_counter()
    iv = faiss.IndexIVFFlat(faiss.IndexFlatIP(D_IDX), D_IDX, nl, faiss.METRIC_INNER_PRODUCT); iv.train(X[: min(n, 100_000)]); iv.add(X); build_s = time.perf_counter() - t0
    faiss.omp_set_num_threads(1)
    _, I_ex = fl.search(QS, TOPK)
    nq_t = 100 if n >= 1_000_000 else 200
    def np_top(q1, X=X):
        s = X @ q1[0]; idx = np.argpartition(-s, TOPK)[:TOPK]; return idx[np.argsort(-s[idx])]
    with threadpool_limits(limits=1):
        l_np = time_each(np_top, QS[:nq_t], warm=5)
    l_fl = time_each(lambda q1: fl.search(q1, TOPK), QS[:nq_t], warm=5)
    chosen = None
    for npb in (1, 2, 4, 8, 16, 32, 64, 128):
        iv.nprobe = npb; _, I_a = iv.search(QS, TOPK); rec = index_recall(I_a, I_ex)
        if rec >= RECALL_TARGET: chosen = (npb, rec); break
    if chosen is None: chosen = (128, rec)
    iv.nprobe = chosen[0]
    l_iv = time_each(lambda q1: iv.search(q1, TOPK), QS[:nq_t], warm=5)
    scale_rows.append({"catalog size n": n, "NumPy brute force p50 (ms)": np.median(l_np), "FAISS exact p50 (ms)": np.median(l_fl), "IVFFlat nlist": nl, "IVFFlat nprobe": chosen[0],
                       "IVFFlat index recall@100": chosen[1], "IVFFlat p50 (ms)": np.median(l_iv), "IVFFlat p95 (ms)": np.percentile(l_iv, 95), "build time (s)": build_s})
    del X, fl, iv
scale_tab = pd.DataFrame(scale_rows)
scale_tab["speed-up vs NumPy"] = scale_tab["NumPy brute force p50 (ms)"] / scale_tab["IVFFlat p50 (ms)"]
display(scale_tab.style.format({"catalog size n": "{:,.0f}", "NumPy brute force p50 (ms)": "{:.3f}", "FAISS exact p50 (ms)": "{:.3f}", "IVFFlat index recall@100": "{:.3f}", "IVFFlat p50 (ms)": "{:.3f}",
                                "IVFFlat p95 (ms)": "{:.3f}", "build time (s)": "{:.1f}", "speed-up vs NumPy": "{:.1f}x"}).hide(axis="index"))

fig, ax = plt.subplots(figsize=(7.4, 4.3))
ax.loglog(scale_tab["catalog size n"], scale_tab["NumPy brute force p50 (ms)"], "o-", label="NumPy brute force", color="#b8660a")
ax.loglog(scale_tab["catalog size n"], scale_tab["FAISS exact p50 (ms)"], "s-", label="FAISS exact (IndexFlatIP)", color="#777")
ax.loglog(scale_tab["catalog size n"], scale_tab["IVFFlat p50 (ms)"], "^-", label="FAISS IVFFlat (recall ≥ 0.95)", color="#1f5fa8")
ax.axhline(15, color="#c0392b", ls="--", lw=1); ax.text(scale_tab["catalog size n"].iloc[0], 17, "15 ms retrieval-stage budget (Phase 7; also covers the query tower)", color="#c0392b", fontsize=8.5)   # same 15 ms as BUDGET in Phase 7
ax.set_ylim(top=35)
ax.set_xlabel("number of items in the catalog"); ax.set_ylabel("top-100 latency per query, p50 (ms)"); ax.set_title("Exact search grows with the catalog; ANN barely does", fontsize=11)
ax.grid(alpha=0.3, which="both"); ax.legend(fontsize=8.5); plt.tight_layout(); plt.show()
```

### 4.7 The trade-off I chose, and why this makes retrieval fit the budget

```python
big = scale_tab.iloc[-1]
_np0 = ivf.nprobe; ivf.nprobe = NPROBE; _, I_ap_t = ivf.search(QT, TOPK); ivf.nprobe = _np0          # which members of the exact top-100 does the chosen index miss?
_miss = np.concatenate([np.flatnonzero(~np.isin(I_ex_test[i], I_ap_t[i])) for i in range(len(QT))]) + 1
miss_pos = float(_miss.mean()) if len(_miss) else float("nan")
display(Markdown(f"""
**Chosen setting.** `IndexIVFFlat`, `nlist = {NLIST}`, `nprobe = {NPROBE}`, inner-product metric, with the seen-movie filter applied inside the search. The rule was fixed *before* looking at test: the cheapest `nprobe` whose index recall@100 on the **validation** queries reaches {RECALL_TARGET:.2f}.

* **What we give up.** The index returns {pick['index recall@100 (test)']:.1%} of the true top-100 on the test queries; the ~{(1 - pick['index recall@100 (test)']) * 100:.0f} movies per request that it misses sit on average at position {miss_pos:.0f} of the exact top-100 (a movie drawn at random from that list would sit at 50.5), so {'they tend to be its less similar members' if miss_pos > 55 else 'they are spread over the whole list'}. What matters to the user is the held-out Recall@100: **{R_IVF_F:.3f}** with IVFFlat vs **{R_EXACT_F:.3f}** exact — a paired difference of {dd_ivf[0]:+.3f} (95% CI [{dd_ivf[1]:+.3f}, {dd_ivf[2]:+.3f}]), i.e. {"no measurable loss" if dd_ivf[1] <= 0 <= dd_ivf[2] else "a small but measurable loss" if dd_ivf[0] < 0 else "no loss"} from approximating.
* **What we gain — on this catalog: very little, and it does not matter.** The p50 latency is {pick['latency p50 (ms)']:.3f} ms (IVFFlat) vs {np_row['latency p50 (ms)']:.3f} ms (NumPy) vs {fl_row['latency p50 (ms)']:.3f} ms (FAISS exact): about {np_row['latency p50 (ms)'] / pick['latency p50 (ms)']:.1f}x in relative terms but only {np_row['latency p50 (ms)'] - pick['latency p50 (ms)']:.3f} ms in absolute terms, i.e. a tiny fraction of a 100 ms budget — with {n_active:,} movies the whole matrix sits in the CPU cache and per-call overhead, not arithmetic, is what is being timed. **The gain appears with scale:** at {int(big['catalog size n']):,} items NumPy needs **{big['NumPy brute force p50 (ms)']:.1f} ms**, exact FAISS {big['FAISS exact p50 (ms)']:.1f} ms, IVFFlat only **{big['IVFFlat p50 (ms)']:.2f} ms** (p95 {big['IVFFlat p95 (ms)']:.2f} ms) at index recall {big['IVFFlat index recall@100']:.2f} — a **{big['speed-up vs NumPy']:.0f}x** speed-up — and exact search keeps growing in proportion to the catalog.
* **Why this is what makes candidate generation (Phase 1) affordable.** Retrieval has to look at the *entire* catalog on every request. An index turns "compare against every item" into "compare against the few percent of items that live near the query", so the cost per request stops being proportional to the catalog size, while the part of the answer that is lost is the part that retrieval does not need to be perfect about (the scorer re-orders the candidates anyway). Cost of the speed: a one-off **offline build** (here {t_ivf * 1000:.0f} ms; {big['build time (s)']:.1f} s for {int(big['catalog size n']):,} items) and a **stale index** between rebuilds (Phase 7).
"""))
```

# Phase 5 — Scoring: the ranking model

### 5.1 What the scoring stage does

Retrieval hands over ~100 candidates per request. The scoring stage ranks **only those** as precisely as it can, with a model that would be far too slow for the whole catalog. It can afford a model in which user and item **interact directly** — cross-features such as "this user's taste vector *and* this movie's vector together" — instead of being confined to two independent towers that only meet in a dot product. That is exactly what NeuMF's MLP branch (Task 14) does, so the Task 14 NeuMF is reused here **as it is** (no re-tuning, trained on the training split only).

The experiment of this phase has three arms, all on the same 610 test users (seen movies excluded everywhere):

| Pipeline | What happens | Movies the scorer touches per request |
|---|---|---|
| **A. brute force** | NeuMF scores the **whole catalog**; top-10 | every unseen movie (≈ 9.5k here) |
| **B. two-stage** (the architecture) | two-tower + IVFFlat retrieves 100 candidates → NeuMF scores only those → top-10 | 100 |
| **C. retrieval only** | the two-tower order of the 100 candidates, no scorer | 0 |

```python
R = R_DEFAULT
CAND = {"val": retrieve_all(ivf, QV, "val", nprobe=NPROBE, k=R), "test": FILTERED[NPROBE][:, :R].copy()}      # (610, R) candidate ids, best two-tower score first, seen movies already excluded
assert (CAND["test"] >= 0).all() and (CAND["val"] >= 0).all()

def neumf_on(cand):
    """NeuMF probability of every (user, candidate) pair; user = row index. Padding (-1) gets -inf."""
    s = pairwise_cand_scores(neumf, np.where(cand >= 0, cand, 0))
    return np.where(cand >= 0, s, -np.inf)

def order_by(cand, scores):
    o = np.argsort(-scores, axis=1, kind="stable")
    return np.take_along_axis(cand, o, 1), np.take_along_axis(scores, o, 1)

SC_NEUMF = {s: neumf_on(CAND[s]) for s in ("val", "test")}
LIST_B, SORTED_B = order_by(CAND["test"], SC_NEUMF["test"])                    # B: retrieve then score (best first)
LIST_C = CAND["test"]                                                           # C: retrieval order
S_A = S_NEUMF["test"].copy(); S_A[SEEN_FOR_TEST] = -np.inf; S_A[:, n_active:] = -np.inf
LIST_A = np.argsort(-S_A, axis=1)[:, :100]                                      # A: brute force over the whole catalog
n_scored_A = int((~SEEN_FOR_TEST[:, :n_active]).sum(1).mean())

PIPE = {"A. brute force: NeuMF on the whole catalog": LIST_A, "B. two-stage: two-tower (R=100) → NeuMF": LIST_B, "C. retrieval only: two-tower order": LIST_C}
RK = {nm: ranks_from_lists(l, TEST_ITEM) for nm, l in PIPE.items()}
RK["Popularity (reference)"] = ZOO_RANKS["Popularity"]
rows = []
for nm, r in RK.items():
    m = metrics_from_ranks(r, (10,)); lo, hi = boot_ci(per_user_ndcg(r))
    rows.append({"pipeline": nm, "HR@10": m["HR@10"], "NDCG@10": m["NDCG@10"], "NDCG@10 95% CI": f"[{lo:.4f}, {hi:.4f}]",
                 "movies scored by NeuMF per request": {"A": f"{n_scored_A:,}", "B": f"{R}", "C": "0"}.get(nm[0], "—") if nm[1] == "." else "—"})
p5_tab = pd.DataFrame(rows)
display(p5_tab.style.format({"HR@10": "{:.4f}", "NDCG@10": "{:.4f}"}).hide(axis="index"))

nd = {k[0]: per_user_ndcg(RK[k]) for k in PIPE}
D_BA = paired_boot_diff(nd["B"], nd["A"]); D_BC = paired_boot_diff(nd["B"], nd["C"]); D_CA = paired_boot_diff(nd["C"], nd["A"])
print(f"paired NDCG@10 differences (mean, 95% CI): B - A = {D_BA[0]:+.4f} [{D_BA[1]:+.4f}, {D_BA[2]:+.4f}] | B - C = {D_BC[0]:+.4f} [{D_BC[1]:+.4f}, {D_BC[2]:+.4f}] | C - A = {D_CA[0]:+.4f} [{D_CA[1]:+.4f}, {D_CA[2]:+.4f}]")
```

### 5.2 Does the two-stage list match the brute-force list?

Two different questions hide here, and both are measured:

* **Agreement with the scorer's own answer** — is B's top-10 the same set of movies as A's top-10? (The task asks to confirm this.)
* **Agreement with reality** — does B find the movie the user actually watched next as often as A does?

```python
top10 = {k[0]: PIPE[k][:, :10] for k in PIPE}
def overlap_rows(a, b): return np.array([len(set(x) & set(y)) / 10 for x, y in zip(a, b)])
ov_BA = overlap_rows(top10["B"], top10["A"])
same_set = np.mean([set(x) == set(y) for x, y in zip(top10["B"], top10["A"])])
top1_in = np.mean([top10["A"][u][0] in set(CAND["test"][u]) for u in range(n_users)])
ceiling = float(np.mean([TEST_ITEM[u] in set(CAND["test"][u]) for u in range(n_users)]))                          # share of users whose held-out movie is a candidate at all
hitA, hitB, hitC = (RK[k] <= 10 for k in PIPE)
hr_B, hr_A, hr_C = hitB.mean(), hitA.mean(), hitC.mean()
agree = pd.DataFrame([
    {"measure": "agreement with the scorer's own answer: average overlap of B's top-10 with A's top-10 (movies in common / 10)", "value": f"{ov_BA.mean():.3f}"},
    {"measure": "    (= the share of A's top-10 movies that retrieval handed over at all)", "value": f"{ov_BA.mean():.3f}"},
    {"measure": "    share of users whose B top-10 is exactly A's top-10 set", "value": f"{same_set:.3f}"},
    {"measure": "    share of users whose A top-1 movie is among the 100 candidates", "value": f"{top1_in:.3f}"},
    {"measure": "agreement with reality: retrieval ceiling — the held-out movie is among the 100 candidates (Recall@100)", "value": f"{ceiling:.3f}"},
    {"measure": "    users (of 610) who hit within the top-10 in both A and B", "value": f"{int((hitA & hitB).sum())}"},
    {"measure": "    users who hit within the top-10 only with A", "value": f"{int((hitA & ~hitB).sum())}"},
    {"measure": "    users who hit within the top-10 only with B", "value": f"{int((hitB & ~hitA).sum())}"},
])
display(agree.style.hide(axis="index"))
# the first two rows are one quantity by construction: A's top-10 movies that are candidates keep the highest NeuMF scores inside the candidate list, so B returns exactly those
assert abs(ov_BA.mean() - np.mean([np.isin(top10["A"][u], CAND["test"][u]).mean() for u in range(n_users)])) < 1e-9
```

```python
# a fully ranked list for one sample user (the same median-activity user as Phase 2): retrieve-then-score vs brute force
def titles(ids): return [str(movies.title[j])[:38] for j in ids]
u0 = U0
lst = pd.DataFrame({"rank": range(1, 11),
                    "B: two-stage (retrieve -> score)": titles(LIST_B[u0, :10]), "NeuMF prob.": SORTED_B[u0, :10],
                    "A: brute-force NeuMF": titles(LIST_A[u0, :10])})
lst["B movie also in A's top-10?"] = ["yes" if LIST_B[u0, r] in set(LIST_A[u0, :10]) else "no" for r in range(10)]
rb_ = RK["B. two-stage: two-tower (R=100) → NeuMF"][u0]; ra_ = RK["A. brute force: NeuMF on the whole catalog"][u0]
print(f"sample user: MovieLens user {user_ids[u0]}; held-out movie: {str(movies.title[TEST_ITEM[u0]])[:50]!r} | among the 100 candidates: {'yes' if TEST_ITEM[u0] in set(CAND['test'][u0]) else 'no'}; "
      f"its rank in B: {'not in the list' if not np.isfinite(rb_) else int(rb_)}; in A (top-100 shown): {'not in the list' if not np.isfinite(ra_) else int(ra_)}")
display(lst.style.format({"NeuMF prob.": "{:.3f}"}).hide(axis="index"))
```

```python
def cmp_phrase(d, a="B", b="A"):
    return f"{a} is better than {b}" if d[1] > 0 else (f"{a} is worse than {b}" if d[2] < 0 else f"{a} and {b} cannot be told apart")
low_ov = ov_BA.mean() < 0.5
zoo_nd = zoo_tab["NDCG@10"]
neumf_weakest = bool(zoo_nd["NeuMF (Task 14)"] <= zoo_nd[["BPR-MF (Task 12)", "Softmax DNN (Task 14)", "NeuMF (Task 14)"]].min())
if low_ov:
    ov_head = "low"
    ov_body = (f"the answer is *no, not most of it* — the two-tower model and NeuMF disagree about what is plausible, so the {R} candidates contain only {ov_BA.mean():.0%} of the scorer's favourites. "
               "The task text expects a \"close match\"; with these two models it does **not** hold, and that is a result, not an error.")
else:
    ov_head = "high"
    ov_body = f"the answer is *mostly yes* — the {R} candidates contain {ov_BA.mean():.0%} of the scorer's favourites, which is the \"close match\" the task text expects."
real_head = ("Agreement with reality is better, not worse." if D_BA[1] > 0 else
             "Agreement with reality is worse." if D_BA[2] < 0 else
             "Agreement with reality is the same within the noise.")
real_dir = "more often than" if hr_B > hr_A else ("less often than" if hr_B < hr_A else "as often as")
in_sample = " in this sample" if (D_BA[1] <= 0 <= D_BA[2] and hr_B != hr_A) else ""   # point estimates differ, but the interval includes 0
guard = ((f"NeuMF is the weakest of the learned models as a stand-alone ranker on this data (§1.3); restricting it" if neumf_weakest else "Restricting NeuMF")
         + " to movies that a different model already finds plausible can remove some of its confident mistakes. Retrieval acts as a *guard rail* for the scorer, not only as a speed trick. "
         "(A plausible reading of the numbers — this notebook does not isolate it with an ablation.)")
if not low_ov:
    why_bullet = ""
elif D_BA[1] > 0:
    why_bullet = f"* **Why a pipeline that agrees so little can still win.** {guard}\n"
elif D_BA[2] >= 0:
    why_bullet = f"* **Why a pipeline that agrees so little can still keep up.** {guard}\n"
else:
    why_bullet = (f"* **Why this pipeline loses to brute force.** The {R} candidates leave out {1 - ov_BA.mean():.0%} of the scorer's favourites, and where the scorer is the better judge, "
                  "the retrieval ceiling costs accuracy (see the practical lesson below).\n")
bc_expl = ("With only ids and genres as inputs this dataset gives a scorer no information that the retriever does not already have; a scorer's expensive cross-features pay off when it sees many more signals (watch time, device, time of day, price, …). §5.4 checks whether a different scorer would do better here."
           if D_BC[1] <= 0 else "Here the scorer does add something over the retriever's own order. §5.4 checks whether a different scorer would do better still.")
display(Markdown(f"""
* **Agreement with the scorer's own answer is {ov_head}.** {'Only ' if low_ov else ''}**{ov_BA.mean():.0%}** of the movies in A's top-10 appear in B's top-10 (average), and {'only ' if low_ov else ''}**{same_set:.1%}** of users get the identical set. This is exactly the quantity "does retrieval hand over what the scorer would have picked?": {ov_body}
* **{real_head}** B finds the held-out movie {real_dir} A{in_sample}: HR@10 {hr_B:.4f} vs {hr_A:.4f}; NDCG@10 {nd['B'].mean():.4f} vs {nd['A'].mean():.4f} (paired Δ {D_BA[0]:+.4f}, 95% CI [{D_BA[1]:+.4f}, {D_BA[2]:+.4f}] → **{cmp_phrase(D_BA)}**), while scoring **{n_scored_A / R:.0f}x fewer movies** per request ({R} instead of ≈{n_scored_A:,}). Counting users: {int((hitB & ~hitA).sum())} hit only with B, {int((hitA & ~hitB).sum())} only with A, {int((hitA & hitB).sum())} with both.
{why_bullet}* **Retrieval alone vs retrieve-then-score.** C (the two-tower order, no scorer) reaches NDCG@10 {nd['C'].mean():.4f}; B − C = {D_BC[0]:+.4f} (CI [{D_BC[1]:+.4f}, {D_BC[2]:+.4f}] → **{cmp_phrase(D_BC, 'B', 'C')}**). {bc_expl}
* **The practical lesson.** The final list can only contain what retrieval hands over: a movie outside the {R} candidates can never be scored, so **retrieval recall is the ceiling** — here the held-out movie is among the candidates for {ceiling:.1%} of the users, and no scorer can lift the top-10 hit rate above that. A good retrieval stage makes an expensive scorer affordable *as long as the candidate set contains what the scorer would pick*; that is why Exercise 2 grows `R`, and why Phase 8's monitoring plan tracks "is the scorer's top pick inside the candidate set?".
"""))
```

### 5.4 Is NeuMF the right scorer on this data? (the same 100 candidates, five different scorers)

The architecture does not care which model scores: the scorer can be swapped without touching retrieval or re-ranking. Here the same candidate sets are re-ordered by each model of the zoo, and by a simple *fusion* — the average of NeuMF's and the two-tower's within-list z-scores (equal weights, fixed in advance, **not tuned**). Validation is shown so that a choice could be made without looking at test.

```python
def gather(S, cand): return np.take_along_axis(S, np.where(cand >= 0, cand, 0), 1)
def zrow(x): return (x - x.mean(1, keepdims=True)) / np.maximum(x.std(1, keepdims=True), 1e-9)
NEU, TWO, FUS = "NeuMF (Task 14) — the scorer of this notebook", "two-tower score (= retrieval order)", "fusion: ½ NeuMF + ½ two-tower (z-scores)"
SCORERS = {}
for sp in ("val", "test"):
    c = CAND[sp]
    SCORERS[sp] = {TWO: gather(S_TT[sp], c), NEU: SC_NEUMF[sp], "BPR-MF (Task 12)": bpr.cand_scores(np.where(c >= 0, c, 0)), "Softmax DNN (Task 14)": gather(S_SOFTMAX[sp], c)}
    SCORERS[sp][FUS] = 0.5 * zrow(SCORERS[sp][NEU]) + 0.5 * zrow(SCORERS[sp][TWO])
rows, nd_sc = [], {}
for nm in SCORERS["test"]:
    out = {}
    for sp in ("val", "test"):
        l, _ = order_by(CAND[sp], SCORERS[sp][nm]); r = ranks_from_lists(l, TARGET[sp]); out[sp] = (metrics_from_ranks(r, (10,)), per_user_ndcg(r))
    nd_sc[nm] = out["test"][1]
    lo, hi = boot_ci(out["test"][1])
    rows.append({"scorer (re-orders the same 100 candidates)": nm, "val NDCG@10": out["val"][0]["NDCG@10"], "test HR@10": out["test"][0]["HR@10"], "test NDCG@10": out["test"][0]["NDCG@10"], "test NDCG@10 95% CI": f"[{lo:.4f}, {hi:.4f}]"})
sc_tab = pd.DataFrame(rows)
sc_tab["Δ vs NeuMF (test, paired CI)"] = [("" if nm == NEU else "{:+.4f} [{:+.4f}, {:+.4f}]".format(*paired_boot_diff(nd_sc[nm], nd_sc[NEU]))) for nm in SCORERS["test"]]
display(sc_tab.style.format({"val NDCG@10": "{:.4f}", "test HR@10": "{:.4f}", "test NDCG@10": "{:.4f}"}).hide(axis="index"))
col = "scorer (re-orders the same 100 candidates)"
best_val_scorer = sc_tab.loc[sc_tab["val NDCG@10"].idxmax(), col]
weakest_test = sc_tab.loc[sc_tab["test NDCG@10"].idxmin(), col]
d_fz_neu = paired_boot_diff(nd_sc[FUS], nd_sc[NEU]); d_fz_tt = paired_boot_diff(nd_sc[FUS], nd_sc[TWO])
display(Markdown(
    f"**Reading.** On validation the best single re-orderer is *{best_val_scorer}*; on test the weakest is *{weakest_test}*. "
    f"The fusion of NeuMF and the two-tower score reaches NDCG@10 {sc_tab.loc[sc_tab[col] == FUS, 'test NDCG@10'].iloc[0]:.4f}: Δ vs NeuMF {d_fz_neu[0]:+.4f} (CI [{d_fz_neu[1]:+.4f}, {d_fz_neu[2]:+.4f}]), Δ vs the two-tower order {d_fz_tt[0]:+.4f} (CI [{d_fz_tt[1]:+.4f}, {d_fz_tt[2]:+.4f}]). "
    f"Most differences between scorers are small next to the sampling noise of 610 users, so the honest summary is: **on MovieLens-small the scoring stage adds little over a good retriever** — its value here is architectural (it is where richer features, a different objective or a heavier model go once the data has them), and *which* scorer to ship is a question for an A/B test (Phase 8), not for this table. "
    f"The rest of the notebook keeps **NeuMF** as the scorer, as specified, and lists the fusion as the first experiment to run."))
```

# Phase 6 — Re-ranking: diversity, freshness, fairness and business rules

### 6.1 Why a separate last stage

The scorer answers "how likely is this movie for this user?", one movie at a time. A product also wants things that are properties of the **whole list**: no ten near-identical movies, a few new releases, nothing the user has already seen, no category flooding the page, exposure for items that have no history yet. These goals are awkward to encode as training labels (they change with the week, the season, a campaign) but trivial to apply as a final pass over the 50-movie shortlist — which is also cheap (50 items) and switchable without retraining.

**What is implemented** (each step is measured on its own, then all together):

| Step | Rule | Formula / mechanism |
|---|---|---|
| diversity | greedy **MMR** over the scored top-50 | `score'(i) = λ·rel(i) − (1−λ)·max_{j∈selected} sim(i,j)`, `sim` = cosine of the Task 11 content vectors (the same space as Task 13's diversity metric); `rel` = NeuMF probability rescaled to [0, 1] inside the shortlist |
| freshness | bonus for recently released movies | `rel'(i) = rel(i) + β·exp(−age(i)/τ)`, `age` = (year of the request − release year), `τ` = 3 years |
| business rule 1 | never show a movie the user has already seen | hard filter (also applied inside retrieval; kept here as the safety net) |
| business rule 2 | at most 3 movies of the same primary genre in one list | hard cap inside the greedy loop — a stand-in for "no two items from the same creator / brand" |

`λ` and `β` are chosen on the **validation** split with a rule fixed in advance (below); the test split is evaluated once.

```python
PRIMARY_ID = pd.factorize(PRIMARY_GENRE)[0]                                    # integer code of each movie's primary genre (for the genre cap)
QYEAR = {sp: pd.to_datetime(ts[sp], unit="s").year.values.astype(float) for sp in ("val", "test")}      # year of the request, per user
N_SHORT_ = N_SHORT; TAU = 3.0; GENRE_CAP = 3

def mmr_select(rel, S, lam, n_out, groups=None, cap=None):
    """Greedy Maximal Marginal Relevance. rel: (m,) relevance, S: (m,m) similarity. Returns the indices (into rel) of the n_out selected items, in selection order."""
    m = len(rel); avail = np.ones(m, bool); max_sim = np.zeros(m); chosen = []; counts = {}
    for _ in range(min(n_out, m)):
        score = lam * rel - (1.0 - lam) * max_sim
        if cap is not None:
            full = [g for g, c in counts.items() if c >= cap]
            ok = avail & ~np.isin(groups, full) if full else avail
            if not ok.any(): ok = avail                                           # nothing satisfies the cap -> relax it rather than return a short list
        else:
            ok = avail
        j = int(np.argmax(np.where(ok, score, -np.inf)))
        chosen.append(j); avail[j] = False; max_sim = np.maximum(max_sim, S[j])
        if cap is not None: counts[groups[j]] = counts.get(groups[j], 0) + 1
    return chosen

def rerank_user(cand_u, scores_u, seen_row, qyear, lam=1.0, beta=0.0, cap=None, drop_seen=True, n_short=N_SHORT_, n_out=K):
    """Re-ranks ONE request: shortlist = best n_short by score -> drop seen -> relevance (+freshness bonus) -> greedy MMR (+genre cap). Returns the chosen item ids in display order."""
    o = np.argsort(-scores_u, kind="stable")[:n_short]
    ids, s = cand_u[o], scores_u[o]
    ok = (ids >= 0) & np.isfinite(s)
    if drop_seen: ok &= ~seen_row[np.where(ids >= 0, ids, 0)]
    ids, s = ids[ok], s[ok]
    if len(ids) == 0: return ids
    rel = (s - s.min()) / max(s.max() - s.min(), 1e-9)
    if beta > 0:
        age = np.maximum(0.0, qyear - REL_YEAR[ids])
        rel = rel + beta * np.where(np.isfinite(age), np.exp(-age / TAU), 0.0)
    Fc = F[ids]
    sel = mmr_select(rel, Fc @ Fc.T, lam, n_out, groups=PRIMARY_ID[ids] if cap else None, cap=cap)
    return ids[sel]

def rerank(cand, scores, split, **kw):
    """Re-ranks every user's scored candidates (see rerank_user). Returns (U, n_out) item ids (-1 = padding)."""
    n_out = kw.get("n_out", K)
    out = np.full((len(cand), n_out), -1, dtype=np.int64)
    for u in range(len(cand)):
        sel = rerank_user(cand[u], scores[u], SEEN[split][u], QYEAR[split][u], **kw)
        out[u, :len(sel)] = sel
    return out

# ---- list-level measures (Task 13 diversity metric + freshness + exposure) ----
def ild_rows(lists):
    """Intra-list diversity: mean pairwise (1 - cosine) of the content vectors in each list (Task 13)."""
    out = np.zeros(len(lists))
    for u, l in enumerate(lists):
        l = l[l >= 0]; Fc = F[l]; G = Fc @ Fc.T; k = len(l)
        out[u] = 1.0 - (G.sum() - np.trace(G)) / max(k * (k - 1), 1)
    return out

def fresh_share_rows(lists, split, within=5.0):
    return np.array([np.mean((QYEAR[split][u] - REL_YEAR[l[l >= 0]]) <= within) for u, l in enumerate(lists)])

def relevance_kept_rows(lists, split):
    """Label-free cost of re-ranking, on the scale MMR itself works in: every shortlisted movie's NeuMF probability is rescaled to [0, 1] inside the user's top-50 (best = 1, 50th = 0);
    per user: the share of the RAW top-10's total rescaled relevance that the final list still carries (1.0 = nothing given up)."""
    out = np.zeros(len(lists))
    for u in range(len(lists)):
        sc = SC_NEUMF[split][u]; top = np.sort(sc)[::-1][:N_SHORT_]; lo_, hi_ = top.min(), top.max()
        d = dict(zip(CAND[split][u].tolist(), ((sc - lo_) / max(hi_ - lo_, 1e-9)).tolist()))
        out[u] = sum(d[j] for j in lists[u] if j >= 0) / sum(d[j] for j in RAW[split][u])
    return out

def relevance_kept(lists, split):
    """Mean over users of relevance_kept_rows (the number used for choosing lambda and beta)."""
    return float(relevance_kept_rows(lists, split).mean())

def list_report(lists, split="test"):
    flat = lists[lists >= 0]
    r = ranks_from_lists(lists, TARGET[split]); nd_ = per_user_ndcg(r); lo, hi = boot_ci(nd_)
    return {"HR@10": float((r <= 10).mean()), "NDCG@10": float(nd_.mean()), "NDCG@10 95% CI": f"[{lo:.4f}, {hi:.4f}]", "relevance kept": relevance_kept(lists, split), "ILD@10": float(ild_rows(lists).mean()),
            "share released ≤5 yrs before the request": float(fresh_share_rows(lists, split).mean()), "catalog coverage (distinct movies shown)": int(len(np.unique(flat))),
            "slots filled by the 100 most popular movies": float((flat < 100).mean()), "mean primary-genre count of the most common genre": float(np.mean([np.bincount(PRIMARY_ID[l[l >= 0]]).max() for l in lists]))}, nd_

# ---- the raw scored list (Phase 5, pipeline B) as the reference ----
RAW = {"val": order_by(CAND["val"], SC_NEUMF["val"])[0][:, :K], "test": LIST_B[:, :K]}
rep_raw_val, _ = list_report(RAW["val"], "val"); rep_raw, nd_raw = list_report(RAW["test"], "test")
assert abs(rep_raw["relevance kept"] - 1.0) < 1e-9
assert np.allclose(rerank(CAND["test"], SC_NEUMF["test"], "test", lam=1.0, beta=0.0), RAW["test"]), "lambda=1, beta=0 must reproduce the raw scored top-10"
print(f"sanity: re-ranking with lambda=1, beta=0, no cap reproduces the raw top-10 exactly | raw list: NDCG@10 {rep_raw['NDCG@10']:.4f}, ILD@10 {rep_raw['ILD@10']:.3f} (test)")
```

### 6.2 Diversity: MMR over the scored top-50

The table shows the relevance/diversity curve as `λ` goes from 1 (pure relevance) down to 0.3. Two cost measures are shown: **relevance kept** — the share of the raw top-10's total NeuMF probability that the new list still carries (label-free: it needs no held-out movie, so it has none of the sampling noise of NDCG@10 on 610 users) — and the real **NDCG@10**.

**Selection rule, fixed before looking at test:** take the *smallest* `λ` (the most diverse list) whose **validation relevance kept** is at least **95%** — "we accept giving up at most 5% of the scorer's relevance for variety". NDCG@10 is deliberately *not* used to choose: on 610 users its standard error is about ±15% of its value, so a tolerance rule on it would pick whatever noise favours.

```python
LAMS = [1.0, 0.9, 0.8, 0.7, 0.6, 0.5, 0.4, 0.3]
rows, LAM_LISTS = [], {}
for lam in LAMS:
    lv = rerank(CAND["val"], SC_NEUMF["val"], "val", lam=lam); lt = rerank(CAND["test"], SC_NEUMF["test"], "test", lam=lam); LAM_LISTS[lam] = lt
    rv, _ = list_report(lv, "val"); rt, nd_t = list_report(lt, "test")
    d = paired_boot_diff(nd_t, nd_raw) if lam < 1 else (0.0, 0.0, 0.0)
    rows.append({"λ": lam, "val relevance kept": rv["relevance kept"], "val ILD@10": rv["ILD@10"], "test relevance kept": rt["relevance kept"], "test ILD@10": rt["ILD@10"], "test NDCG@10": rt["NDCG@10"],
                 "test ΔNDCG@10 vs raw (paired CI)": "" if lam == 1 else "{:+.4f} [{:+.4f}, {:+.4f}]".format(*d), "test HR@10": rt["HR@10"], "movies changed vs raw top-10": 1 - float(np.mean([len(set(a) & set(b)) / 10 for a, b in zip(lt, RAW["test"])]))})
mmr_tab = pd.DataFrame(rows)
display(mmr_tab.style.format({"λ": "{:.1f}", "val relevance kept": "{:.1%}", "val ILD@10": "{:.3f}", "test relevance kept": "{:.1%}", "test ILD@10": "{:.3f}", "test NDCG@10": "{:.4f}", "test HR@10": "{:.4f}",
                              "movies changed vs raw top-10": "{:.0%}"}).hide(axis="index"))

REL_KEEP_MMR = 0.95
ok_lams = mmr_tab[mmr_tab["val relevance kept"] >= REL_KEEP_MMR]
LAMBDA = float(ok_lams["λ"].min())
print(f"rule: smallest λ with validation relevance kept >= {REL_KEEP_MMR:.0%} -> λ = {LAMBDA}")

fig, ax = plt.subplots(figsize=(6.6, 4.2))
ax.plot(mmr_tab["test ILD@10"], mmr_tab["test NDCG@10"], "o-", color="#2a7d3b")
for _, r_ in mmr_tab.iterrows(): ax.annotate(f"λ={r_['λ']:.1f}", (r_["test ILD@10"], r_["test NDCG@10"]), textcoords="offset points", xytext=(5, 5), fontsize=8)
ax.set_xlabel("intra-list diversity ILD@10 (higher = more varied lists)"); ax.set_ylabel("NDCG@10 (test)"); ax.set_title("What diversity costs: relevance vs variety (MMR)", fontsize=11); ax.grid(alpha=0.3)
plt.tight_layout(); plt.show()
```

### 6.3 Freshness: a bonus for recently released movies

With the chosen `λ` fixed, `β` (the size of the bonus, on the same 0–1 relevance scale) is increased. **Selection rule, fixed before looking at test:** the *largest* `β` for which the **validation relevance kept** — measured against the original raw list, so MMR's share of the cost is already included — is still at least **90%** (MMR may spend up to 5 points of relevance, freshness a further 5).

```python
BETAS = [0.0, 0.05, 0.1, 0.2, 0.3, 0.5, 0.75, 1.0]
rows, BETA_LISTS = [], {}
for beta in BETAS:
    lv = rerank(CAND["val"], SC_NEUMF["val"], "val", lam=LAMBDA, beta=beta); lt = rerank(CAND["test"], SC_NEUMF["test"], "test", lam=LAMBDA, beta=beta); BETA_LISTS[beta] = lt
    rv, _ = list_report(lv, "val"); rt, nd_t = list_report(lt, "test")
    rows.append({"β": beta, "val relevance kept": rv["relevance kept"], "val share ≤5 yrs": rv["share released ≤5 yrs before the request"], "test relevance kept": rt["relevance kept"],
                 "test share ≤5 yrs": rt["share released ≤5 yrs before the request"], "mean release year of the list": float(np.nanmean([REL_YEAR[l[l >= 0]].mean() for l in lt])),
                 "test ILD@10": rt["ILD@10"], "test NDCG@10": rt["NDCG@10"], "test ΔNDCG@10 vs raw (paired CI)": "{:+.4f} [{:+.4f}, {:+.4f}]".format(*paired_boot_diff(nd_t, nd_raw))})
fr_tab = pd.DataFrame(rows)
display(fr_tab.style.format({"β": "{:.2f}", "val relevance kept": "{:.1%}", "val share ≤5 yrs": "{:.1%}", "test relevance kept": "{:.1%}", "test share ≤5 yrs": "{:.1%}", "mean release year of the list": "{:.1f}",
                             "test ILD@10": "{:.3f}", "test NDCG@10": "{:.4f}"}).hide(axis="index"))
REL_KEEP_ALL = 0.90
ok_b = fr_tab[fr_tab["val relevance kept"] >= REL_KEEP_ALL]
BETA = float(ok_b["β"].max())
print(f"rule: largest β with validation relevance kept >= {REL_KEEP_ALL:.0%} -> β = {BETA}")
```

### 6.4 Business rules: never repeat a seen movie, and cap one genre at 3 per list

**Rule 1 — exclude what the user has already seen.** The pipeline filters seen movies *inside* retrieval (Phase 4.5), so the rule never fires on the final list — but it must exist as a hard guard in the last stage, because a seen movie shown again is the most visible bug a recommender can have. The next cell shows what would happen without either filter.

**Rule 2 — a category cap.** At most 3 of the 10 slots may share a *primary genre* (the first genre listed for the movie). It stands in for the rule real products need ("don't show two items from the same creator / brand / franchise back-to-back"); other rules of the same kind are: content-policy or age filters, regional licensing, "not more than N items from one partner", and minimum slots for new or sponsored content.

```python
# --- what would the seen-movie rule do? retrieval WITHOUT the filter, then the same scorer ---
ivf.nprobe = NPROBE
_, I_unf = ivf.search(QT, R)
sc_unf = neumf_on(I_unf.astype(np.int64)); l_unf, _ = order_by(I_unf.astype(np.int64), sc_unf)
seen_share_unf = float(np.mean([SEEN_FOR_TEST[u, l_unf[u, :10]].mean() for u in range(n_users)]))
seen_share_cand = float(np.mean([SEEN_FOR_TEST[u, I_unf[u]].mean() for u in range(n_users)]))
print(f"without any seen-movie filter: {seen_share_cand:.1%} of the 100 retrieved candidates and {seen_share_unf:.1%} of the top-10 slots are movies the user has already seen "
      f"(the user's own history is what the models were trained on); with the filter: 0.0%.")

# --- the guard in the last stage: feed it a list that contains seen movies and check it removes them ---
_chk = rerank(I_unf.astype(np.int64), sc_unf, "test", lam=1.0, beta=0.0, drop_seen=True)
assert not any(SEEN_FOR_TEST[u, _chk[u][_chk[u] >= 0]].any() for u in range(n_users)), "guard failed: a seen movie survived the re-ranker"
print("guard test passed: the re-ranker removes every seen movie even when retrieval did not")

# --- cap: effect on its own, on top of MMR + freshness ---
cap_list = rerank(CAND["test"], SC_NEUMF["test"], "test", lam=LAMBDA, beta=BETA, cap=GENRE_CAP)
nocap_list = BETA_LISTS[BETA]
mx_nocap = np.array([np.bincount(PRIMARY_ID[l]).max() for l in nocap_list]); mx_cap = np.array([np.bincount(PRIMARY_ID[l[l >= 0]]).max() for l in cap_list])
# the cap can only be honoured if the 50-movie shortlist offers enough genres: feasible <=> sum over genres of min(count, cap) >= 10 (otherwise the re-ranker relaxes the cap instead of returning a short list)
def cap_feasible(u):
    o = np.argsort(-SC_NEUMF["test"][u], kind="stable")[:N_SHORT_]; ids = CAND["test"][u][o]; ids = ids[~SEEN_FOR_TEST[u, ids]]
    return np.minimum(np.bincount(PRIMARY_ID[ids]), GENRE_CAP).sum() >= K
feasible = np.array([cap_feasible(u) for u in range(n_users)])
print(f"largest number of movies with the same primary genre in one list: without cap — mean {mx_nocap.mean():.2f}, users above {GENRE_CAP}: {int((mx_nocap > GENRE_CAP).sum())} | "
      f"with cap — mean {mx_cap.mean():.2f}, users above {GENRE_CAP}: {int((mx_cap > GENRE_CAP).sum())}")
print(f"the shortlist makes the cap impossible for {int((~feasible).sum())} users (too few different genres among their 50 shortlisted movies); for those the rule is relaxed rather than the list shortened")
assert (mx_cap[feasible] <= GENRE_CAP).all(), "the genre cap was violated for a user whose shortlist allowed it"
```

```python
FINAL_TEST = cap_list                                                           # diversity + freshness + both business rules
FINAL_VAL = rerank(CAND["val"], SC_NEUMF["val"], "val", lam=LAMBDA, beta=BETA, cap=GENRE_CAP)
steps = {"raw scored top-10 (Phase 5, B)": RAW["test"], f"+ MMR diversity (λ={LAMBDA})": LAM_LISTS[LAMBDA], f"+ freshness (β={BETA})": BETA_LISTS[BETA], f"+ genre cap ≤{GENRE_CAP} = final list": FINAL_TEST}
rows, ndf = [], {}
for nm, l in steps.items():
    rep, nd_ = list_report(l, "test"); ndf[nm] = nd_
    rows.append({"list": nm, **rep, "movies changed vs raw": 1 - float(np.mean([len(set(a) & set(b)) / 10 for a, b in zip(l, RAW["test"])]))})
final_tab = pd.DataFrame(rows)
display(final_tab.style.format({"HR@10": "{:.4f}", "NDCG@10": "{:.4f}", "relevance kept": "{:.1%}", "ILD@10": "{:.3f}", "share released ≤5 yrs before the request": "{:.1%}", "slots filled by the 100 most popular movies": "{:.1%}",
                                "mean primary-genre count of the most common genre": "{:.2f}", "movies changed vs raw": "{:.0%}", "catalog coverage (distinct movies shown)": "{:,.0f}"}).hide(axis="index"))
names = list(steps)
D_FINAL = paired_boot_diff(ndf[names[-1]], ndf[names[0]])
raw_r, fin_r = final_tab.iloc[0], final_tab.iloc[-1]
display(Markdown(f"""
**The trade-off that was made.**
* **Relevance paid:** the final list keeps {fin_r['relevance kept']:.1%} of the raw list's NeuMF relevance; measured against the held-out movies, NDCG@10 {raw_r['NDCG@10']:.4f} → {fin_r['NDCG@10']:.4f} (paired Δ {D_FINAL[0]:+.4f}, 95% CI [{D_FINAL[1]:+.4f}, {D_FINAL[2]:+.4f}] → {"a measurable loss" if D_FINAL[2] < 0 else ("a measurable gain" if D_FINAL[1] > 0 else "within the noise of 610 users — we cannot claim a loss, but also not a gain")}); HR@10 {raw_r['HR@10']:.4f} → {fin_r['HR@10']:.4f}.
* **Variety bought:** ILD@10 {raw_r['ILD@10']:.3f} → {fin_r['ILD@10']:.3f}; the most frequent primary genre fills {raw_r['mean primary-genre count of the most common genre']:.2f} → {fin_r['mean primary-genre count of the most common genre']:.2f} of the 10 slots on average.
* **Freshness bought:** the share of movies released within 5 years of the request {raw_r['share released ≤5 yrs before the request']:.1%} → {fin_r['share released ≤5 yrs before the request']:.1%}.
* **Exposure:** the final lists show **{fin_r['catalog coverage (distinct movies shown)']:,}** distinct movies across the 610 users (raw: {raw_r['catalog coverage (distinct movies shown)']:,}); the 100 most popular movies fill {fin_r['slots filled by the 100 most popular movies']:.1%} of the slots (raw: {raw_r['slots filled by the 100 most popular movies']:.1%}).
* **Changed lists:** {fin_r['movies changed vs raw']:.0%} of the movies in the 10-lists differ from the raw list.

**Each step's own effect** (the table rows are cumulative, so read the row-to-row change): MMR moves ILD@10 {final_tab.iloc[0]['ILD@10']:.3f} → {final_tab.iloc[1]['ILD@10']:.3f}; freshness moves the release-year share {final_tab.iloc[1]['share released ≤5 yrs before the request']:.1%} → {final_tab.iloc[2]['share released ≤5 yrs before the request']:.1%}; the cap moves the most-common-genre count {final_tab.iloc[2]['mean primary-genre count of the most common genre']:.2f} → {final_tab.iloc[3]['mean primary-genre count of the most common genre']:.2f}. The cap is a guarantee rather than a tuning knob: it only changes lists in which one genre already dominated.
"""))
```

### 6.6 Fairness — what re-ranking can and cannot do

A relevance-maximising scorer **starves anything without an engagement history**: a new movie, a small creator, a niche genre has few interactions, so its predicted relevance is low, so it is shown rarely, so it collects no interactions — a feedback loop. Because that loop is a property of the *list-level exposure*, not of any single prediction, the re-ranking stage is where it is usually broken: exposure floors for new or small-creator content, per-group caps, or randomised "exploration slots" that give an item enough impressions to be judged.

What the table above can show here is the *symptom* (coverage and the share of slots taken by the head of the popularity curve), not the cause — MovieLens has no creator or demographic fields, so **no fairness claim about people or creators can be made from this data**. Two honest caveats: (1) coverage and head-share are crude proxies of exposure fairness; (2) a constraint that raises exposure for under-shown items costs relevance *today* in exchange for information and a healthier catalog *tomorrow*, which only an online experiment (Phase 8) can price.

# Phase 7 — Serving infrastructure and latency budgets

### 7.1 A request, stage by stage, and where each stage's cost comes from

| Stage | What runs per request | Where the time goes | Grows with |
|---|---|---|---|
| feature look-up | fetch the user's recent history / profile | network round-trip to a feature store (here: an array read) | — |
| ① retrieval | query tower → one vector; ANN search with the seen-movie filter | a small neural-network call; the index probe is memory-bound | almost nothing (that is what the index is for) |
| ② scoring | NeuMF on the `R` candidates | forward passes of the heaviest model | **linearly in `R`** and in model size |
| ③ re-ranking | MMR over the shortlist: `50 × 50` similarities, 10 greedy steps | list-level logic, mostly interpreter overhead | shortlist size² |
| overhead | network hops between services, auth, (de)serialisation | the network, not the CPU | number of service boundaries |

The code below is a **single-request serving function**: it does exactly what one live request would do (nothing is pre-computed per user except what a real system also stores: the user's history window and the set of movies they have seen).

```python
USER_INDEX = {int(uid): i for i, uid in enumerate(user_ids)}              # MovieLens userId -> internal row

@tf.function(input_signature=[tf.TensorSpec([None, 1], tf.int32), tf.TensorSpec([None, len(GENRES)], tf.float32), tf.TensorSpec([None, H], tf.int32)])
def query_fn(u, g, h):
    """Compiled query tower: (user id, recent genre mix, last-H history window) -> query vector."""
    return query_tower([u, g, h], training=False)

def features_for(u, split="test", hist_ids=None):
    """Query features of user row `u` at request time. Default: the stored history of the split; or build them from `hist_ids` (movie ids, oldest first) — the 'online' path."""
    if hist_ids is None:
        return FEATS[split]["user"][u:u + 1], FEATS[split]["genre_mix"][u:u + 1], FEATS[split]["hist"][u:u + 1]
    h = np.zeros((1, H), np.int32); last = np.asarray(hist_ids, np.int64)[-H:]
    if len(last): h[0, H - len(last):] = last + 1
    return np.array([[u]], np.int32), genre_mix(h), h

def serve(u, split="test", R=R_DEFAULT, top_n=K, compiled=True, hist_ids=None, lam=None, beta=None, cap=None, trace=None, details=None):
    """ONE request: features -> query tower -> filtered ANN search -> NeuMF scoring -> re-ranking. Returns the final movie ids; `trace` (a dict) receives per-stage milliseconds, `details` (a dict) the candidates and their scores."""
    lam = LAMBDA if lam is None else lam; beta = BETA if beta is None else beta; cap = GENRE_CAP if cap is None else cap
    t0 = time.perf_counter()
    feats = features_for(u, split, hist_ids)
    t1 = time.perf_counter()
    q = (query_fn(*feats).numpy() if compiled else query_tower.predict(list(feats), verbose=0)).astype("float32")
    t2 = time.perf_counter()
    _, I = search_filtered(ivf, q, SEEN_IDS[split][u], k=R, nprobe=NPROBE, ensure_k=True)
    cand = I[0][I[0] >= 0]
    t3 = time.perf_counter()
    if compiled: s = neumf_score_pairs(u, cand)
    else: s = neumf.predict([np.full((len(cand), 1), u, np.int32), cand[:, None].astype(np.int32)], batch_size=4096, verbose=0).reshape(-1)
    t4 = time.perf_counter()
    out = rerank_user(cand, s, SEEN[split][u], QYEAR[split][u], lam=lam, beta=beta, cap=cap, n_short=N_SHORT, n_out=top_n)
    t5 = time.perf_counter()
    if details is not None: details.update({"cand": cand, "scores": s, "query": q})
    if trace is not None:
        trace.update({"feature look-up": (t1 - t0) * 1e3, "query tower": (t2 - t1) * 1e3, "ANN search (+ seen filter)": (t3 - t2) * 1e3, "NeuMF scoring": (t4 - t3) * 1e3, "re-ranking": (t5 - t4) * 1e3, "total": (t5 - t0) * 1e3})
    return out

# ---- correctness: the one-request function must reproduce the batch pipeline of Phase 6 (same lists, user by user) ----
_same = np.mean([np.array_equal(serve(u), FINAL_TEST[u][FINAL_TEST[u] >= 0]) for u in range(0, n_users, 5)])
print(f"serving function vs batch pipeline: identical lists for {_same:.1%} of the {len(range(0, n_users, 5))} users checked")
assert _same > 0.97, "the serving function does not reproduce the batch pipeline"
```

### 7.2 Measured end-to-end timing — first the naive implementation, then the compiled one

*Naive* = what one writes first: `model.predict(...)` for the query tower and for NeuMF. *Compiled* = the same models wrapped in a `tf.function` with a fixed input signature (graph traced once, no per-call Python/Keras overhead). Everything else is identical. One thread, one request at a time, warm-up excluded; the naive run uses fewer requests because each one is slow.

```python
STAGES = ["feature look-up", "query tower", "ANN search (+ seen filter)", "NeuMF scoring", "re-ranking", "total"]

def run_timing(compiled, users, reps):
    for u in users[:10]: serve(u, compiled=compiled)                            # warm-up
    rows = []
    for _ in range(reps):
        for u in users:
            tr = {}; serve(u, compiled=compiled, trace=tr); rows.append(tr)
    return pd.DataFrame(rows)

def pct_tab(df):
    return pd.DataFrame({"p50 (ms)": df.quantile(0.50), "p95 (ms)": df.quantile(0.95), "p99 (ms)": df.quantile(0.99), "mean (ms)": df.mean()})

T_NAIVE = run_timing(False, list(range(0, n_users, 6)), 1)                      # ~100 requests
T_COMP = run_timing(True, list(range(n_users)), 3)                              # 1,830 requests
tab_naive, tab_comp = pct_tab(T_NAIVE[STAGES]), pct_tab(T_COMP[STAGES])
display(Markdown("**Naive implementation** (`model.predict` for both neural models):"))
display(tab_naive.style.format("{:.2f}"))
display(Markdown("**Compiled implementation** (`tf.function`; this is the pipeline used from here on):"))
display(tab_comp.style.format("{:.2f}"))
BOTTLE_NAIVE = tab_naive.drop(index="total")["p50 (ms)"].idxmax(); BOTTLE_COMP = tab_comp.drop(index="total")["p50 (ms)"].idxmax()
print(f"\nbiggest stage, naive: {BOTTLE_NAIVE} ({tab_naive.loc[BOTTLE_NAIVE, 'p50 (ms)']:.1f} ms of {tab_naive.loc['total', 'p50 (ms)']:.1f} ms p50) | compiled: {BOTTLE_COMP} ({tab_comp.loc[BOTTLE_COMP, 'p50 (ms)']:.2f} ms of {tab_comp.loc['total', 'p50 (ms)']:.2f} ms p50)")
```

The measured columns are **per-request sums** (e.g. retrieval = query tower + ANN search of the *same* request), so percentiles are taken of real request totals, not added up from stage percentiles.

```python
BUDGET = {"overhead: network, auth, feature store, serialisation": 20.0, "① retrieval: query tower + ANN search": 15.0, "② scoring: NeuMF on the candidates": 40.0, "③ re-ranking": 10.0, "headroom for tails and bursts": 15.0}
assert abs(sum(BUDGET.values()) - 100.0) < 1e-9
GROUP = {"overhead: network, auth, feature store, serialisation": ["feature look-up"], "① retrieval: query tower + ANN search": ["query tower", "ANN search (+ seen filter)"],
         "② scoring: NeuMF on the candidates": ["NeuMF scoring"], "③ re-ranking": ["re-ranking"]}
rows = []
for stage, b in BUDGET.items():
    if stage in GROUP:
        x = T_COMP[GROUP[stage]].sum(1); xn = T_NAIVE[GROUP[stage]].sum(1)
        note = "only the feature-store read is measured; the network is not simulated" if stage.startswith("overhead") else ""
        rows.append({"stage": stage, "budget (ms)": b, "compiled p50": x.median(), "compiled p95": x.quantile(0.95), "compiled p99": x.quantile(0.99), "p99 used": x.quantile(0.99) / b,
                     "naive p50": xn.median(), "naive p95": xn.quantile(0.95), "note": note})
    else:
        rows.append({"stage": stage, "budget (ms)": b, "compiled p50": np.nan, "compiled p95": np.nan, "compiled p99": np.nan, "p99 used": np.nan, "naive p50": np.nan, "naive p95": np.nan, "note": ""})
rows.append({"stage": "TOTAL", "budget (ms)": 100.0, "compiled p50": T_COMP["total"].median(), "compiled p95": T_COMP["total"].quantile(0.95), "compiled p99": T_COMP["total"].quantile(0.99),
             "p99 used": T_COMP["total"].quantile(0.99) / 100.0, "naive p50": T_NAIVE["total"].median(), "naive p95": T_NAIVE["total"].quantile(0.95), "note": "measured total excludes the network"})
bud_tab = pd.DataFrame(rows)
display(bud_tab.style.format({"budget (ms)": "{:.0f}", "compiled p50": "{:.2f}", "compiled p95": "{:.2f}", "compiled p99": "{:.2f}", "p99 used": "{:.1%}", "naive p50": "{:.1f}", "naive p95": "{:.1f}"}, na_rep="").hide(axis="index"))
NAIVE_OVER = float(T_NAIVE["total"].quantile(0.95)) > 100.0 - BUDGET["overhead: network, auth, feature store, serialisation"] - BUDGET["headroom for tails and bursts"]
print(f"compiled: p99 of the whole request {T_COMP['total'].quantile(0.99):.2f} ms = {T_COMP['total'].quantile(0.99):.1f}% of the 100 ms budget | "
      f"naive: p95 {T_NAIVE['total'].quantile(0.95):.1f} ms {'— the compute part alone already exceeds what the budget leaves for it (100 ms minus overhead and headroom)' if NAIVE_OVER else '— fits'}")
```

### 7.4 Offline versus online, and the staleness this creates

| | Offline (batch job, nightly / hourly) | Online (every request) |
|---|---|---|
| what | train the towers and NeuMF; run the candidate tower over the catalog; build the ANN index | features → query tower → ANN search → scoring → re-ranking |
| why there | the candidate tower does not depend on the query, so its output can be computed once and shared by all requests; an index build needs the whole catalog and is expensive | the query depends on the user's **latest** actions, so it must be computed per request |
| time limit | none (minutes–hours) | the 100 ms budget |

```python
off_rows = [("train the two-tower model", float(tt_curve["sec"].sum()), "once per model version (hours at production scale)"),
            ("train NeuMF", float(neu_curve["train sec"].sum()), "once per model version"),
            ("embed the whole catalog with the candidate tower", float(embed_sec), "every index rebuild"),
            ("build the IVFFlat index", float(t_ivf), "every index rebuild")]
big = scale_tab.iloc[-1]
display(pd.DataFrame(off_rows, columns=["offline job (this notebook)", "seconds", "how often"]).style.format({"seconds": "{:.2f}"}).hide(axis="index"))
rebuild_1m = float(embed_sec) * big["catalog size n"] / n_items + float(big["build time (s)"])
print(f"extrapolated to a {int(big['catalog size n']):,}-item catalog: embedding ≈ {float(embed_sec) * big['catalog size n'] / n_items:.0f} s + index build {float(big['build time (s)']):.1f} s ≈ {rebuild_1m:.0f} s per rebuild — a nightly (or hourly) batch job.")

# ---- staleness, measured ----
cold_users = int((TEST_ITEM >= n_active).sum())
print(f"\nSTALENESS, item side: the index was built from the training period. {cold_users} of the {n_users} test users ({cold_users / n_users:.1%}) watched next a movie that was not in the index "
      f"(no training interaction). A frozen index can never retrieve those, so it caps Recall at {1 - cold_users / n_users:.1%} however good the model is.")

# ---- ... while the user side is fresh: the query tower is online, so one new action changes the list at once, with NO rebuild ----
hist_full = [int(h) - 1 for h in FEATS["test"]["hist"][U0] if h > 0]
last_two = hist_full[-1:]
before = serve(U0, hist_ids=hist_full[:-1]); after = serve(U0, hist_ids=hist_full)
print(f"STALENESS, user side: sample user {user_ids[U0]} — the same pipeline, the same index, before vs after ONE new action ({str(movies.title[last_two[0]])[:40]!r}): "
      f"{10 - len(set(before) & set(after))} of the 10 movies change. No index rebuild is involved, because the query tower runs per request.")
```

### 7.5 The biggest bottleneck and the fix

The bottleneck is read off the measurements above — never guessed. Then the fix is *applied and re-measured*, and the budget is projected to a larger catalog, because a pipeline that is fast only on a 9.7k-movie catalog proves little.

```python
sp = tab_naive.loc[BOTTLE_NAIVE, "p50 (ms)"] / tab_comp.loc[BOTTLE_NAIVE, "p50 (ms)"]
nn_ms = tab_naive.loc[["query tower", "NeuMF scoring"], "p50 (ms)"]; nn_share = nn_ms.sum() / tab_naive.loc["total", "p50 (ms)"]
display(Markdown(f"""
**Measured bottleneck.** In the naive pipeline the two `model.predict` calls — the query tower ({nn_ms['query tower']:.1f} ms p50) and NeuMF ({nn_ms['NeuMF scoring']:.1f} ms) — make up **{nn_share:.0%}** of a {tab_naive.loc['total', 'p50 (ms)']:.1f} ms request; the single biggest stage is **{BOTTLE_NAIVE}** ({tab_naive.loc[BOTTLE_NAIVE, 'p50 (ms)']:.1f} ms).
It is **not arithmetic**: the same call, compiled, takes {tab_comp.loc[BOTTLE_NAIVE, 'p50 (ms)']:.2f} ms — **{sp:.0f}x faster**. `model.predict` is built for large offline batches (it sets up a data pipeline, callbacks and progress handling on every call); for one request that framework overhead dwarfs the maths. The index search, the part the whole architecture is designed around, was never the problem: {tab_naive.loc['ANN search (+ seen filter)', 'p50 (ms)']:.2f} ms p50 even in the naive pipeline.

**Fix applied:** wrap the models in `tf.function` with a fixed input signature (done above). The whole request goes from {tab_naive.loc['total', 'p50 (ms)']:.1f} ms to {tab_comp.loc['total', 'p50 (ms)']:.2f} ms p50 ({tab_naive.loc['total', 'p50 (ms)'] / tab_comp.loc['total', 'p50 (ms)']:.0f}x), p99 {tab_comp.loc['total', 'p99 (ms)']:.2f} ms.

**What is the bottleneck now?** After the fix, the largest stage is **{BOTTLE_COMP}** ({tab_comp.loc[BOTTLE_COMP, 'p50 (ms)']:.2f} ms p50). Each remaining stage is a small fraction of its allocation, so on *this* catalog nothing needs further work.
"""))

# ---- projection: how the request latency behaves as the catalog grows (measured pieces; retrieval taken from the Phase 4.6 benchmark) ----
fixed_ms = float(T_COMP[["feature look-up", "query tower", "NeuMF scoring", "re-ranking"]].sum(1).median())          # does not depend on the catalog size
proj = []
for _, r_ in scale_tab.iterrows():
    n = int(r_["catalog size n"])
    proj.append({"catalog size": n, "pipeline p50 (ms): fixed stages + IVFFlat": fixed_ms + float(r_["IVFFlat p50 (ms)"]), "brute force: NeuMF on every movie (ms)": float(np.interp(n, cost_tab["items scored per request"], cost_tab["NeuMF time (ms)"]))})   # interpolated between the sizes timed in §1.2
proj_tab = pd.DataFrame(proj)
proj_tab["brute force / pipeline"] = proj_tab.iloc[:, 2] / proj_tab.iloc[:, 1]
display(proj_tab.style.format({"catalog size": "{:,.0f}", "pipeline p50 (ms): fixed stages + IVFFlat": "{:.2f}", "brute force: NeuMF on every movie (ms)": "{:.1f}", "brute force / pipeline": "{:.0f}x"}).hide(axis="index"))
display(Markdown(f"""
**What the projection says.** The pipeline's cost is almost independent of the catalog (only the ANN probe grows, and slowly): about {proj_tab.iloc[0, 1]:.1f} ms at {proj_tab.iloc[0, 0]:,} movies and {proj_tab.iloc[-1, 1]:.1f} ms at {proj_tab.iloc[-1, 0]:,}. Scoring the whole catalog with NeuMF grows in proportion: {proj_tab.iloc[-1, 2]:.0f} ms at {proj_tab.iloc[-1, 0]:,} movies, {proj_tab.iloc[-1, 3]:.0f}x the pipeline — and with a production-size ranker the factor is far larger.
(Caveat: a production ranker has far more inputs per candidate, so the scoring stage would be the one to watch; the levers are the ones the task names — fewer candidates `R` (Exercise 2 measures the cost in quality), a cheaper scorer, or quantised / batched inference. Real services also add network time that this single-machine test cannot show.)
"""))
```

# Phase 8 — Capstone: the full pipeline, monitoring and the A/B loop

### 8.1 `get_recommendations(user_id, top_n=10)`

One function, the complete pipeline: **look up the user → query tower → filtered ANN retrieval → NeuMF scoring → MMR + freshness + business rules → a ready-to-show table**. It takes a MovieLens `userId` (not an internal row number), falls back to a popularity list for a user it has never seen (a brand-new user has no history to build a query from), and returns the movie, its metadata and why it is in the list. The request is evaluated at the **test-time state**: everything that happened before the held-out movie (so the held-out movie itself is never an input).

```python
def get_recommendations(user_id, top_n=10, *, R=R_DEFAULT, lam=None, beta=None, cap=None, hist_ids=None, split="test", return_trace=False):
    """Complete retrieve -> score -> re-rank pipeline for one request.
    user_id : MovieLens userId.  top_n : list length (<= 50, the shortlist size).  hist_ids : optional override of the user's recent history (movie ids, oldest first).
    Returns a DataFrame (rank, movieId, title, genres, released, NeuMF probability, source)."""
    assert 1 <= top_n <= N_SHORT, f"top_n must be between 1 and the shortlist size ({N_SHORT})"
    trace = {}
    if int(user_id) not in USER_INDEX:                                           # unknown user: no history, no query vector -> popularity fallback (ids are popularity-ordered)
        ids = np.arange(top_n)
        out = pd.DataFrame({"rank": range(1, top_n + 1), "movieId": item_ids[ids], "title": movies.title.values[ids], "genres": movies.genres.values[ids],
                            "released": REL_YEAR[ids], "NeuMF probability": np.nan, "source": "popularity fallback (unknown user)"})
        return (out, trace) if return_trace else out
    u = USER_INDEX[int(user_id)]
    det = {}
    ids = serve(u, split=split, R=R, top_n=top_n, hist_ids=hist_ids, lam=lam, beta=beta, cap=cap, trace=trace, details=det)
    sc = dict(zip(det["cand"].tolist(), det["scores"].tolist()))
    scorer_top = set(det["cand"][np.argsort(-det["scores"], kind="stable")[:top_n]].tolist())
    out = pd.DataFrame({"rank": range(1, len(ids) + 1), "movieId": item_ids[ids], "title": movies.title.values[ids], "genres": movies.genres.values[ids], "released": REL_YEAR[ids],
                        "NeuMF probability": [sc[j] for j in ids.tolist()], "source": ["scorer's own top-n" if j in scorer_top else "promoted by re-ranking" for j in ids.tolist()]})
    return (out, trace) if return_trace else out

# ---- integration test: the function must reproduce the batch pipeline of Phase 6 for every one of the 610 users ----
ITEM_INDEX = {int(m): i for i, m in enumerate(item_ids)}                         # MovieLens movieId -> internal item id
def internal_ids(rec): return rec["movieId"].map(ITEM_INDEX).values
same = [np.array_equal(internal_ids(get_recommendations(int(user_ids[u]))), FINAL_TEST[u][FINAL_TEST[u] >= 0]) for u in range(n_users)]
print(f"get_recommendations() reproduces the batch pipeline's final list for {np.mean(same):.1%} of the {n_users} users")
assert np.mean(same) > 0.99
assert get_recommendations(10 ** 9).shape[0] == 10 and (get_recommendations(10 ** 9)["source"] == "popularity fallback (unknown user)").all()
print("unknown user id -> popularity fallback works")
```

### 8.2 Tested on three different users

Three deliberately different users: the **heaviest** (most training interactions), the **lightest**, and the **median** user (the sample user of Phases 2–6). For each: what they watched last, the ten recommendations, and programmatic sanity checks. The *taste-fit* check measures how well the list matches the genres of the user's own recent history (cosine between the list's average genre vector and the history's), against the same number for plain popularity.

```python
act_tr = np.bincount(tr_u, minlength=n_users)
THREE = {"heaviest user": int(np.argmax(act_tr)), "median user": U0, "lightest user": int(np.argmin(act_tr))}
GM_HIST = FEATS["test"]["genre_mix"]
def list_genre_vec(ids): return G_MH[np.asarray(ids)].mean(0)
def cos_(a, b): return float(a @ b / max(np.linalg.norm(a) * np.linalg.norm(b), 1e-12))
def fit_of(u, ids): return cos_(list_genre_vec(ids), GM_HIST[u])
POP_UNSEEN = lambda u: np.flatnonzero(~SEEN_FOR_TEST[u, :n_active])[:10]                     # the 10 most popular movies the user has not seen
LOWPROMO = {}                                                                               # per demo user: (promoted movies with a NeuMF probability < 0.05, of which released within 5 years of the request)

for label, u in THREE.items():
    rec, tr = get_recommendations(int(user_ids[u]), return_trace=True)
    ids = internal_ids(rec)
    hist_t = [str(movies.title[h - 1])[:34] for h in FEATS["test"]["hist"][u] if h > 0][-5:]
    top_g = [GENRES[j] for j in np.argsort(-GM_HIST[u])[:3]]
    checks = {"10 distinct movies": bool(len(set(ids)) == 10), "none already seen": bool(not SEEN_FOR_TEST[u, ids].any()), f"≤{GENRE_CAP} per primary genre": bool(np.bincount(PRIMARY_ID[ids]).max() <= GENRE_CAP)}
    print(f"\n===== {label}: MovieLens user {user_ids[u]} | {act_tr[u]} training interactions | request took {tr['total']:.1f} ms =====")
    print(f"last movies watched : {hist_t}")
    print(f"strongest genres in recent history: {top_g}")
    display(rec.assign(**{"NeuMF probability": rec["NeuMF probability"].round(3), "released": rec["released"].map(lambda v: f"{v:.0f}"), "title": rec["title"].str[:46]}).drop(columns="movieId").style.hide(axis="index"))
    held = TEST_ITEM[u]
    print(f"sanity: {checks} | taste fit (genre cosine): list {fit_of(u, ids):.2f} vs most-popular-unseen {fit_of(u, POP_UNSEEN(u)):.2f} | "
          f"held-out movie {str(movies.title[held])[:36]!r}: {'in this list' if held in set(ids) else 'not in this list'}")
    assert checks["10 distinct movies"] and checks["none already seen"]
    promo_low = (rec["source"].str.startswith("promoted") & (rec["NeuMF probability"] < 0.05)).to_numpy()
    LOWPROMO[label] = (int(promo_low.sum()), int(((QYEAR["test"][u] - REL_YEAR[ids][promo_low]) <= 5).sum()))

# ---- the same taste-fit check over ALL users, as the quantitative version of the qualitative check ----
fit_final = np.array([fit_of(u, FINAL_TEST[u][FINAL_TEST[u] >= 0]) for u in range(n_users)])
fit_raw = np.array([fit_of(u, RAW["test"][u]) for u in range(n_users)])
fit_pop = np.array([fit_of(u, POP_UNSEEN(u)) for u in range(n_users)])
print(f"\ntaste fit over all {n_users} users (mean genre cosine with the user's own recent history): final lists {fit_final.mean():.3f} | raw scored lists {fit_raw.mean():.3f} | most-popular-unseen {fit_pop.mean():.3f}")

# ---- an honest look at the price of re-ranking for single users ----
rk_users = relevance_kept_rows(FINAL_TEST, "test")
worst = max(LOWPROMO, key=lambda k: LOWPROMO[k][0])
if LOWPROMO[worst][0] >= 3:
    display(Markdown(f"""**A caution from the {worst}'s list.** {LOWPROMO[worst][0]} of its 10 movies were promoted by the re-ranker although NeuMF gives each of them a probability below 0.05 ({LOWPROMO[worst][1]} of these {LOWPROMO[worst][0]} were released within 5 years of the request). The cause is the scale the re-ranker works on: relevance is rescaled to 0–1 *inside the shortlist*, so when one candidate is far ahead of a flat tail of tiny probabilities (for example a user who has already seen most of what is popular), the freshness bonus (up to β = {BETA}) can be larger than the differences inside the tail, and then the release year decides much of the tail's order. Over all {n_users} users the price is smaller — relevance kept: median {np.median(rk_users):.0%}, 10th percentile {np.percentile(rk_users, 10):.0%}, lowest {rk_users.min():.0%} — but a production version would add a floor ("never promote a movie whose score is below a set fraction of the raw top-10") or use a rank-based relevance scale."""))
```

### 8.4 The ongoing A/B feedback loop (Task 13, Phase 8, applied to a two-stage system)

**Ship retrieval and scoring changes independently, and A/B test them separately.**

* They fail in different ways. A new **two-tower model** changes *which movies are even considered* — its risks are recall, coverage and index/feature breakage, and its benefit can be invisible if the scorer cannot use it (Phase 5: a retriever with higher recall did not automatically improve the final list). A new **scorer** changes only the *order of the same candidates* — its risks are calibration and position effects, and a stronger scorer is wasted if retrieval drops the right movie first (retrieval recall is the ceiling).
* They are measured differently. Retrieval: **candidate-set recall / coverage** (offline, against what users later chose) and a *counterfactual-safe* online test, because users only click what was shown, so an arm that retrieves *new* movies has no click data about them until it shows them. Scoring: the final-list metric on identical candidate sets (interleaving is efficient here).
* Testing only the combined system confounds the two: a regression cannot be attributed, and a gain from one stage can hide a loss in the other.

**The loop:** (1) offline gate — beat the current model on Task 13/14-style metrics with a bootstrap interval *and* pass the latency and recall checks above; (2) **shadow mode** — the new stage runs on live traffic without being shown, so its output, latency and failure rate can be compared with production; (3) **A/B on a small slice** with pre-registered primary metric and guardrails (Exercise 5), run for whole weekly cycles; (4) ramp, with the monitoring table above as the alerting layer; (5) the logged impressions and clicks become the next training set — and the next time the loop starts, the retrieval model must be trained on what the *new* system showed, not only on history, or it will keep reproducing the old system's choices.

```python
# A/B arithmetic to ground the design: how many users per arm are needed to see a lift of a given size in a hit-rate-type metric (two-sided 5% test, 80% power)?
from scipy.stats import norm as _norm
def users_per_arm(p0, rel_lift, alpha=0.05, power=0.80):
    p1 = p0 * (1 + rel_lift); zb, za = _norm.ppf(power), _norm.ppf(1 - alpha / 2)
    return int(np.ceil(((za * np.sqrt(2 * (p0 + p1) / 2 * (1 - (p0 + p1) / 2)) + zb * np.sqrt(p0 * (1 - p0) + p1 * (1 - p1))) ** 2) / (p1 - p0) ** 2))
p_base = float((r_final <= 10).mean())
ab = pd.DataFrame({"relative lift to detect": ["+2%", "+5%", "+10%", "+20%"], "users needed per arm": [users_per_arm(p_base, x) for x in (0.02, 0.05, 0.10, 0.20)]})
display(Markdown(f"Baseline for the arithmetic: the pipeline's offline HR@10 = {p_base:.3f} (a stand-in for a per-user success rate such as 'clicked something in the list'). *Offline* we have 610 users — enough to see only very large differences, which is why every comparison in this notebook carries an interval."))
display(ab.style.format({"users needed per arm": "{:,.0f}"}).hide(axis="index"))
```

## Exercise 1 — Swap `IndexIVFFlat` for `IndexHNSWFlat`

**What changes.** `IndexIVFFlat` splits the catalog into `nlist` clusters and scans the `nprobe` clusters closest to the query. `IndexHNSWFlat` builds a *graph* in which every movie is linked to about `M` near neighbours, and a query walks that graph towards its nearest neighbours; the search-effort knob is `efSearch` (build effort: `efConstruction`).

**A catch with inner products.** A graph is built on *distances*. Our two-tower model is trained with a dot product and the movie vectors have very different lengths (Phase 2), so "close in the graph" is not the same as "highest dot product". FAISS accepts `METRIC_INNER_PRODUCT` for HNSW, so that is measured as variant 1. The standard fix is to add one coordinate to every movie vector, √(M² − ‖x‖²) with M the longest vector, and a 0 to every query: the squared L2 distance then equals ‖q‖² + M² − 2·q·x, so *nearest in L2 = highest dot product, exactly*. That is variant 2 (the cell checks that an exhaustive search on the augmented vectors returns the same top-100).

Protocol as in 4.4: one thread, every query timed separately; `efSearch` is chosen on the **validation** queries as the cheapest setting with index recall@100 ≥ 0.95.

```python
M_HNSW, EF_CONSTR, EF_GRID = 32, 200, (16, 32, 64, 128, 256, 512)

def augment_items(X, nmax=None):
    """one extra coordinate so that squared-L2 distance ranks exactly like the inner product"""
    nmax = float(np.linalg.norm(X, axis=1).max()) if nmax is None else nmax
    return np.hstack([X, np.sqrt(np.maximum(nmax ** 2 - (X ** 2).sum(1), 0))[:, None]]).astype("float32"), nmax

def augment_queries(Q): return np.hstack([Q, np.zeros((len(Q), 1), "float32")]).astype("float32")

def build_hnsw(X, metric, threads):
    faiss.omp_set_num_threads(threads); t0 = time.perf_counter()
    h = faiss.IndexHNSWFlat(X.shape[1], M_HNSW, metric); h.hnsw.efConstruction = EF_CONSTR; h.add(X); dt = time.perf_counter() - t0
    faiss.omp_set_num_threads(1)
    return h, dt

E_AUG, NORM_MAX = augment_items(E_IDX); QT_A, QV_A = augment_queries(QT), augment_queries(QV)
hnsw_ip, t_hnsw_ip = build_hnsw(E_IDX, faiss.METRIC_INNER_PRODUCT, 1)          # variant 1: HNSW straight on the inner product
hnsw_l2, t_hnsw_l2 = build_hnsw(E_AUG, faiss.METRIC_L2, 1)                     # variant 2: HNSW (L2) on the augmented vectors

# the augmentation is exact: exhaustive L2 search on the augmented vectors == exhaustive inner-product search
fl_aug = faiss.IndexFlatL2(E_AUG.shape[1]); fl_aug.add(E_AUG); _, I_aug = fl_aug.search(QT_A, TOPK)
assert index_recall(I_aug, I_ex_test) > 0.995, "the augmented-L2 ranking must reproduce the inner-product ranking"

def sweep_hnsw(index, Qt, Qv, label):
    rows = []
    for ef in EF_GRID:
        index.hnsw.efSearch = ef
        _, I_t = index.search(Qt, TOPK); _, I_v = index.search(Qv, TOPK)
        lat = time_each(lambda q1: index.search(q1, TOPK), Qt, warm=20)
        rows.append({"method": label, "setting": f"efSearch={ef}", "knob": ef, **{f"latency {k} (ms)": v for k, v in pct(lat).items()},
                     "index recall@100 (test)": index_recall(I_t, I_ex_test), "index recall@100 (val)": index_recall(I_v, I_ex_val)})
    return pd.DataFrame(rows)

sw_ip = sweep_hnsw(hnsw_ip, QT, QV, "HNSW, native inner product")
sw_l2 = sweep_hnsw(hnsw_l2, QT_A, QV_A, "HNSW, L2 on augmented vectors")

def choose_ef(sw):
    ok = sw[sw["index recall@100 (val)"] >= RECALL_TARGET]
    return (ok.iloc[0], True) if len(ok) else (sw.iloc[-1], False)
ef_ip_row, ef_ip_ok = choose_ef(sw_ip); ef_l2_row, ef_l2_ok = choose_ef(sw_l2)
EF_IP, EF_L2 = int(ef_ip_row["knob"]), int(ef_l2_row["knob"])

fmt = {"latency p50 (ms)": "{:.3f}", "latency p95 (ms)": "{:.3f}", "latency p99 (ms)": "{:.3f}", "index recall@100 (test)": "{:.3f}", "index recall@100 (val)": "{:.3f}"}
display(Markdown("**efSearch sweep** (index recall@100 against the exact top-100 of the same query, no seen filter):"))
display(pd.concat([sw_ip, sw_l2]).drop(columns="knob").style.format(fmt).hide(axis="index"))
print(f"chosen on validation: native inner product -> efSearch={EF_IP} ({'target reached' if ef_ip_ok else 'target NOT reached even at the largest efSearch tried'}); "
      f"augmented L2 -> efSearch={EF_L2} ({'target reached' if ef_l2_ok else 'target NOT reached'}). IVFFlat (Phase 4): nprobe={NPROBE}.")
```

### 1.2 The production path: seen-movie filter inside the search, all indexes side by side

Same rows as 4.5: the exact index, IVFFlat at its chosen `nprobe`, and the two HNSW variants at their chosen `efSearch`, each with the seen-movie filter inside the search. *Overlap* = share of the exact filtered top-100 that the index returns; *held-out Recall@100* = share of users whose real next movie is among the 100 returned.

```python
hit_exact = (ids_exact_f == TEST_ITEM[:, None]).any(1).astype(float)
ex1_rows = []; EX1_D = {}                                                  # paired difference vs exact search, per index (numbers behind the CI strings)
def add_row(name, setting, idx, Q, knob, build_ms, size_mb):
    if isinstance(idx, faiss.IndexHNSW): idx.hnsw.efSearch = knob
    elif isinstance(idx, faiss.IndexIVF): idx.nprobe = knob
    ids, lat = retrieve_all(idx, Q, "test", timed=True, nprobe=knob)
    hit = (ids == TEST_ITEM[:, None]).any(1).astype(float); d = paired_boot_diff(hit, hit_exact); EX1_D[name] = d
    ex1_rows.append({"index": name, "setting": setting, "build (ms)": build_ms, "size (MB)": size_mb, "overlap with exact top-100": index_recall(ids, ids_exact_f),
                     "held-out Recall@100": float(hit.mean()), "Δ vs exact (paired 95% CI)": "—" if name.startswith("exact") else "{:+.3f} [{:+.3f}, {:+.3f}]".format(*d),
                     "p50 (ms)": float(np.percentile(lat, 50)), "p95 (ms)": float(np.percentile(lat, 95)), "p99 (ms)": float(np.percentile(lat, 99))})
    return ids

add_row("exact (IndexFlatIP)", "—", flat, QT, None, t_flat * 1000, sz(flat))
add_row("IVFFlat", f"nprobe={NPROBE}", ivf, QT, NPROBE, t_ivf * 1000, sz(ivf))
add_row("HNSW, native inner product", f"efSearch={EF_IP}", hnsw_ip, QT, EF_IP, t_hnsw_ip * 1000, sz(hnsw_ip))
add_row("HNSW, L2 on augmented vectors", f"efSearch={EF_L2}", hnsw_l2, QT_A, EF_L2, t_hnsw_l2 * 1000, sz(hnsw_l2))
ex1_tab = pd.DataFrame(ex1_rows)
display(ex1_tab.style.format({"build (ms)": "{:.0f}", "size (MB)": "{:.2f}", "overlap with exact top-100": "{:.3f}", "held-out Recall@100": "{:.3f}", "p50 (ms)": "{:.3f}", "p95 (ms)": "{:.3f}", "p99 (ms)": "{:.3f}"}).hide(axis="index"))
ivf.nprobe = NPROBE                                                              # leave the production index exactly as Phase 4 left it

# recall-vs-latency curves (the classic way to compare ANN indexes)
iv_sw = ann_tab[ann_tab.method == "FAISS IndexIVFFlat"]
fig, ax = plt.subplots(figsize=(7.6, 4.5))
ax.plot(iv_sw["latency p50 (ms)"], iv_sw["index recall@100 (test)"], "^-", color="#1f5fa8", label="IVFFlat (nprobe 1 … 98)")
ax.plot(sw_l2["latency p50 (ms)"], sw_l2["index recall@100 (test)"], "o-", color="#2a8a4a", label="HNSW, L2 on augmented vectors (efSearch 16 … 512)")
ax.plot(sw_ip["latency p50 (ms)"], sw_ip["index recall@100 (test)"], "s--", color="#b8660a", label="HNSW, native inner product")
ax.axhline(RECALL_TARGET, color="#c0392b", ls=":", lw=1); ax.text(ax.get_xlim()[0], RECALL_TARGET + 0.012, f"recall target {RECALL_TARGET}", color="#c0392b", fontsize=8.5)
ax.set_xscale("log"); ax.set_xlabel("p50 latency of one top-100 query (ms, log scale)"); ax.set_ylabel("index recall@100 vs exact (test queries)")
ax.set_title("Recall–latency trade-off on the 9.7k-movie catalog", fontsize=11); ax.grid(alpha=0.3, which="both"); ax.legend(fontsize=8.5, loc="lower right"); plt.tight_layout(); plt.show()
```

### 1.3 Does the picture change with a bigger catalog?

Same synthetic catalogs as 4.6 (the real vectors re-sampled with noise), 10k and 100k items; both indexes are rebuilt here so build time, memory and latency come from the same run. Each index uses the cheapest setting with recall ≥ 0.95 on 200 queries (`nprobe` from 1…128, `efSearch` from 16…512; the largest if none reaches it). A 1M-item HNSW index was **not** built (the build-time column below, scaled up, shows why).

```python
ex1_scale_rows = []
for n in (10_000, 100_000):
    X = synth_catalog(n); XA, _ = augment_items(X); QS = QT[:200]; QSA = augment_queries(QS)
    faiss.omp_set_num_threads(2)
    fl = faiss.IndexFlatIP(D_IDX); fl.add(X); _, I_ex = fl.search(QS, TOPK)
    nl = int(round(np.sqrt(n))); t0 = time.perf_counter()
    iv = faiss.IndexIVFFlat(faiss.IndexFlatIP(D_IDX), D_IDX, nl, faiss.METRIC_INNER_PRODUCT); iv.train(X[: min(n, 100_000)]); iv.add(X); t_iv = time.perf_counter() - t0
    hn, t_hn = build_hnsw(XA, faiss.METRIC_L2, 2)
    faiss.omp_set_num_threads(1)
    pick_iv = None
    for npb in (1, 2, 4, 8, 16, 32, 64, 128):
        iv.nprobe = npb; _, I_a = iv.search(QS, TOPK); rec_iv = index_recall(I_a, I_ex)
        if rec_iv >= RECALL_TARGET: pick_iv = npb; break
    pick_iv = pick_iv or 128; iv.nprobe = pick_iv; _, I_a = iv.search(QS, TOPK); rec_iv = index_recall(I_a, I_ex)
    pick_hn = None
    for ef in EF_GRID:
        hn.hnsw.efSearch = ef; _, I_h = hn.search(QSA, TOPK); rec_hn = index_recall(I_h, I_ex)
        if rec_hn >= RECALL_TARGET: pick_hn = ef; break
    pick_hn = pick_hn or EF_GRID[-1]; hn.hnsw.efSearch = pick_hn; _, I_h = hn.search(QSA, TOPK); rec_hn = index_recall(I_h, I_ex)
    l_iv = time_each(lambda q1: iv.search(q1, TOPK), QS[:100], warm=10); l_hn = time_each(lambda q1: hn.search(q1, TOPK), QSA[:100], warm=10)
    ex1_scale_rows.append({"catalog size n": n, "IVFFlat setting": f"nprobe={pick_iv}", "IVFFlat recall": rec_iv, "IVFFlat p50 (ms)": float(np.median(l_iv)), "IVFFlat build (s)": t_iv, "IVFFlat size (MB)": sz(iv),
                           "HNSW setting": f"efSearch={pick_hn}", "HNSW recall": rec_hn, "HNSW p50 (ms)": float(np.median(l_hn)), "HNSW build (s)": t_hn, "HNSW size (MB)": sz(hn)})
    del X, XA, fl, iv, hn
ex1_scale = pd.DataFrame(ex1_scale_rows)
display(ex1_scale.style.format({"catalog size n": "{:,.0f}", "IVFFlat recall": "{:.3f}", "IVFFlat p50 (ms)": "{:.3f}", "IVFFlat build (s)": "{:.2f}", "IVFFlat size (MB)": "{:.1f}",
                                "HNSW recall": "{:.3f}", "HNSW p50 (ms)": "{:.3f}", "HNSW build (s)": "{:.2f}", "HNSW size (MB)": "{:.1f}"}).hide(axis="index"))
```

## Exercise 2 — Candidate set 100 → 500

The retriever hands `R` candidates to the scorer. A larger `R` raises the **ceiling** (the share of users whose real next movie is available to the scorer at all) but costs scoring time, and it only helps if the scorer is good at picking the right movie out of a bigger pile. The sweep below changes **only** `R` (same retriever, same NeuMF, same re-ranker) and records, for every `R`:

* the ceiling and the NDCG@10 of the scorer's list on validation and test (test with a bootstrap interval and a paired difference against `R = 100`),
* the NDCG@10 of the final re-ranked list,
* the end-to-end latency of the serving function of Phase 7 (one request at a time, 610 requests per `R`),
* the median popularity id of the scorer's top-10 (ids are ordered by popularity, so a *lower* number = more popular movies) — a check on *why* quality moves.

```python
R_GRID = (50, 100, 200, 300, 500)
LB_ref, _ = order_by(CAND["test"], SC_NEUMF["test"]); nd_ref = per_user_ndcg(ranks_from_lists(LB_ref, TEST_ITEM))          # the pipeline as built (R = 100)
for Rg in R_GRID:                                                                                      # one warm-up pass per R before anything is timed, so the first R of the sweep is not penalised
    for u in range(20): serve(u, R=Rg)
ex2_rows, EX2_LISTS, ND_V, WIDE_N = [], {}, {}, {}                                   # WIDE_N: per R, the retrieval searches (val + test) that needed more than nprobe cells to find R unseen movies
for Rg in R_GRID:
    out = {}
    for sp in ("val", "test"):
        w0 = WIDEN["searches"]; C = retrieve_all(ivf, Q_TT[sp], sp, nprobe=NPROBE, k=Rg, ensure_k=True); WIDE_N[Rg] = WIDE_N.get(Rg, 0) + WIDEN["searches"] - w0
        assert (C >= 0).all(), "IVF returned fewer candidates than requested even after scanning every cell"
        L, _ = order_by(C, neumf_on(C)); out[sp] = (C, L)
    Ct, Lt = out["test"]
    nd_t = per_user_ndcg(ranks_from_lists(Lt, TEST_ITEM)); nd_v = per_user_ndcg(ranks_from_lists(out["val"][1], VAL_ITEM)); lo, hi = boot_ci(nd_t); dd = paired_boot_diff(nd_t, nd_ref); ND_V[Rg] = nd_v
    for u in range(10): serve(u, R=Rg)                                                                              # warm-up
    trs, fin = [], np.full((n_users, K), -1, np.int64)
    for u in range(n_users):
        tr = {}; o = serve(u, R=Rg, trace=tr); fin[u, :len(o)] = o; trs.append(tr)
    T = pd.DataFrame(trs); EX2_LISTS[Rg] = fin
    ex2_rows.append({"R": Rg, "ceiling (held-out in candidates)": float((Ct == TEST_ITEM[:, None]).any(1).mean()),
                     "val NDCG@10": float(nd_v.mean()), "test NDCG@10": float(nd_t.mean()), "test 95% CI": f"[{lo:.4f}, {hi:.4f}]", "Δ vs R=100 (paired CI)": "—" if Rg == 100 else "{:+.4f} [{:+.4f}, {:+.4f}]".format(*dd),
                     "final list NDCG@10": float(per_user_ndcg(ranks_from_lists(fin, TEST_ITEM)).mean()), "median popularity id of top-10": float(np.median(Lt[:, :10])),
                     "overlap of top-10 with brute force": float(np.mean([len(set(a[:10]) & set(b[:10])) / 10 for a, b in zip(Lt, LIST_A)])),
                     "NeuMF scoring p50 (ms)": float(T["NeuMF scoring"].median()), "total p50 (ms)": float(T["total"].median()), "total p99 (ms)": float(T["total"].quantile(0.99))})
ex2_tab = pd.DataFrame(ex2_rows)
nd_A = per_user_ndcg(ranks_from_lists(LIST_A, TEST_ITEM)); NDCG_A = float(nd_A.mean())
display(ex2_tab.style.format({"ceiling (held-out in candidates)": "{:.3f}", "val NDCG@10": "{:.4f}", "test NDCG@10": "{:.4f}", "final list NDCG@10": "{:.4f}", "median popularity id of top-10": "{:.0f}",
                              "overlap of top-10 with brute force": "{:.2f}", "NeuMF scoring p50 (ms)": "{:.2f}", "total p50 (ms)": "{:.2f}", "total p99 (ms)": "{:.2f}"}).hide(axis="index"))
print(f"reference: brute-force NeuMF over the whole catalog (pipeline A of Phase 5) has test NDCG@10 {NDCG_A:.4f}")

fig, axes = plt.subplots(1, 2, figsize=(11.6, 4.2))
ax = axes[0]; x = ex2_tab["R"].values
lo_ = [float(s.strip("[]").split(",")[0]) for s in ex2_tab["test 95% CI"]]; hi_ = [float(s.strip("[]").split(",")[1]) for s in ex2_tab["test 95% CI"]]
ax.errorbar(x, ex2_tab["test NDCG@10"], yerr=[ex2_tab["test NDCG@10"] - lo_, np.array(hi_) - ex2_tab["test NDCG@10"]], fmt="o-", color="#1f5fa8", capsize=3, label="test NDCG@10 (95% CI)")
ax.plot(x, ex2_tab["val NDCG@10"], "s--", color="#7a7a7a", label="validation NDCG@10"); ax.axhline(NDCG_A, color="#b8660a", ls=":", label="brute force over the whole catalog (test)")
ax.set_xlabel("candidates handed to the scorer, R"); ax.set_ylabel("NDCG@10 of the scorer's list"); ax.set_ylim(bottom=0); ax.grid(alpha=0.3)
ax2 = ax.twinx(); ax2.plot(x, ex2_tab["ceiling (held-out in candidates)"], "^-", color="#2a8a4a", label="ceiling (right axis)"); ax2.set_ylabel("share of users whose real next movie is a candidate"); ax2.set_ylim(0, 1)
h1, l1 = ax.get_legend_handles_labels(); h2, l2 = ax2.get_legend_handles_labels(); ax.legend(h1 + h2, l1 + l2, fontsize=8, loc="lower left"); ax.set_title("More candidates: higher ceiling, not better list", fontsize=10.5)
ax = axes[1]
ax.plot(x, ex2_tab["total p50 (ms)"], "o-", color="#1f5fa8", label="end-to-end p50"); ax.plot(x, ex2_tab["total p99 (ms)"], "o--", color="#c0392b", label="end-to-end p99"); ax.plot(x, ex2_tab["NeuMF scoring p50 (ms)"], "s-", color="#2a8a4a", label="NeuMF scoring p50")
ax.set_xlabel("candidates handed to the scorer, R"); ax.set_ylabel("ms per request (one thread)"); ax.set_ylim(bottom=0); ax.grid(alpha=0.3); ax.legend(fontsize=8.5); ax.set_title("Latency grows slowly with R (budget: 100 ms)", fontsize=10.5)
plt.tight_layout(); plt.show()
```

## Exercise 3 — A second candidate source: "trending in the last 24 hours"

This is the **mixed hybrid** of Task 13 (Phase 7) applied at the candidate stage: two sources each propose movies, the *union* goes to the same NeuMF scorer, and the rest of the pipeline is unchanged.

**The source.** For a request made at time *t* by user *u*, `trending(u, t, W)` = the movies with the most interactions by **other** users during the window `[t − W, t)`, seen movies removed, at most 50 movies. Only **training** interactions before *t* are used, so nothing from the future (or from the held-out movie) can leak in. The merge keeps the two-tower order first, appends the trending movies that are not already there, and the scorer decides.

**An honest data warning.** MovieLens-small has about 100k ratings spread over 22 years, so a 24-hour window is *nearly empty* for many requests (the table below counts how many). The task's 24 h window is implemented as asked; 7-day and 30-day windows are added to show how the source behaves once it has data. **Rule fixed in advance:** the source is adopted only if, on **validation**, the paired 95% interval of ΔNDCG@10 (merged − two-tower only) lies entirely above zero. Test only reports.

```python
_o = np.argsort(train.timestamp.values, kind="stable")
TR_TS, TR_ITEM, TR_USER = train.timestamp.values[_o], train.i.values[_o].astype(np.int64), train.u.values[_o].astype(np.int64)     # training interactions in time order

def trending_lists(split, hours, k_tr=50):
    """per user: the k_tr most-interacted-with movies of OTHER users in [t - hours, t), t = this user's request time; seen movies and cold movies excluded"""
    T = ts[split]; out = np.full((n_users, k_tr), -1, np.int64); n_found = np.zeros(n_users, int); n_inter = np.zeros(n_users, int)
    for u in range(n_users):
        a = np.searchsorted(TR_TS, T[u] - hours * 3600, "left"); b = np.searchsorted(TR_TS, T[u], "left")          # strictly before the request
        items = TR_ITEM[a:b][TR_USER[a:b] != u]; n_inter[u] = len(items)
        if len(items) == 0: continue
        cnt = np.bincount(items, minlength=n_items); cnt[SEEN[split][u]] = 0; cnt[n_active:] = 0
        top = np.lexsort((np.arange(n_items), -cnt))[:k_tr]; top = top[cnt[top] > 0]                               # most interactions first; ties -> the more popular movie (lower id)
        out[u, :len(top)] = top; n_found[u] = len(top)
    return out, n_found, n_inter

def merge_cands(base, extra):
    """union of two candidate matrices (-1 = padding): `base` order first, then the movies of `extra` that are not already there"""
    out = np.full((len(base), base.shape[1] + extra.shape[1]), -1, np.int64)
    for u in range(len(base)):
        b = base[u][base[u] >= 0]; e = extra[u][extra[u] >= 0]; e = e[~np.isin(e, b)]
        row = np.concatenate([b, e]); out[u, :len(row)] = row
    return out

def zrow_valid(x):
    v = np.where(np.isfinite(x), x, np.nan); m = np.nanmean(v, 1, keepdims=True); sd = np.nanstd(v, 1, keepdims=True)
    return np.where(np.isfinite(x), (x - m) / np.maximum(sd, 1e-9), -np.inf)
def score_neumf(C, split): return neumf_on(C)
def score_fusion(C, split): return 0.5 * zrow_valid(neumf_on(C)) + 0.5 * zrow_valid(np.where(C >= 0, gather(S_TT[split], C), -np.inf))      # the fusion scorer of Phase 5, valid entries only

def run_arm(split, C, scorer):
    L, _ = order_by(C, scorer(C, split)); return L, per_user_ndcg(ranks_from_lists(L, TARGET[split]))

WINDOWS = {"24 hours": 24, "7 days": 24 * 7, "30 days": 24 * 30}
EX3, ex3_rows, ex3_data = {}, [], []
BASE = {sp: {"neumf": run_arm(sp, CAND[sp], score_neumf), "fusion": run_arm(sp, CAND[sp], score_fusion)} for sp in ("val", "test")}
assert abs(BASE["test"]["neumf"][1].mean() - per_user_ndcg(RK["B. two-stage: two-tower (R=100) → NeuMF"]).mean()) < 1e-9, "baseline must equal pipeline B of Phase 5"
for label, hrs in WINDOWS.items():
    res = {}
    for sp in ("val", "test"):
        TRN, nf, ni = trending_lists(sp, hrs); MC = merge_cands(CAND[sp], TRN)
        assert not any(SEEN[sp][u][MC[u][MC[u] >= 0]].any() for u in range(0, n_users, 7)), "a seen movie reached the merged candidate set"
        Lm, ndm = run_arm(sp, MC, score_neumf); Lf, ndf = run_arm(sp, MC, score_fusion)
        res[sp] = {"TRN": TRN, "nf": nf, "ni": ni, "MC": MC, "Lm": Lm, "ndm": ndm, "ndf": ndf,
                   "d_neumf": paired_boot_diff(ndm, BASE[sp]["neumf"][1]), "d_fusion": paired_boot_diff(ndf, BASE[sp]["fusion"][1])}
    EX3[label] = res
    t = res["test"]; tg = TEST_ITEM; ni_t, nf_t = t["ni"], t["nf"]
    in_base = np.array([tg[u] in set(CAND["test"][u]) for u in range(n_users)]); in_merged = np.array([tg[u] in set(t["MC"][u]) for u in range(n_users)])
    only_tr = in_merged & ~in_base; in_top10 = np.array([tg[u] in set(t["Lm"][u][:10]) for u in range(n_users)])
    ex3_data.append({"window": label, "requests with ≥1 trending movie": float((nf_t > 0).mean()), "median other-user interactions in window": float(np.median(ni_t)),
                     "mean list length (requests with a list)": float(nf_t[nf_t > 0].mean()) if (nf_t > 0).any() else 0.0})
    ex3_rows.append({"window": label, "ceiling: two-tower only → merged": f"{in_base.mean():.3f} → {in_merged.mean():.3f}", "real next movie found ONLY via trending (users)": int(only_tr.sum()),
                     "…and ranked in the top-10": int((only_tr & in_top10).sum()),
                     "trending-only movies in the final top-10 (avg)": float(np.mean([np.sum(~np.isin(t["Lm"][u][:10], CAND["test"][u])) for u in range(n_users)])),
                     "NDCG@10 two-tower only": float(BASE["test"]["neumf"][1].mean()), "NDCG@10 merged": float(t["ndm"].mean()),
                     "Δ test, NeuMF scorer": "{:+.4f} [{:+.4f}, {:+.4f}]".format(*t["d_neumf"]), "Δ validation, NeuMF scorer": "{:+.4f} [{:+.4f}, {:+.4f}]".format(*res["val"]["d_neumf"]),
                     "Δ test, fusion scorer": "{:+.4f} [{:+.4f}, {:+.4f}]".format(*t["d_fusion"]),
                     "trending list alone, raw order (NDCG@10)": float(per_user_ndcg(ranks_from_lists(t["TRN"], TEST_ITEM)).mean())})
ex3_data = pd.DataFrame(ex3_data); ex3_tab = pd.DataFrame(ex3_rows)
display(Markdown("**How much data does the source have?** (test requests)"))
display(ex3_data.style.format({"requests with ≥1 trending movie": "{:.0%}", "median other-user interactions in window": "{:.0f}", "mean list length (requests with a list)": "{:.1f}"}).hide(axis="index"))
display(Markdown("**What does merging it do?** (Δ = NDCG@10 of merged candidates minus two-tower-only candidates, paired 95% interval; the baseline is pipeline B of Phase 5)"))
display(ex3_tab.style.format({"trending-only movies in the final top-10 (avg)": "{:.2f}", "NDCG@10 two-tower only": "{:.4f}", "NDCG@10 merged": "{:.4f}", "trending list alone, raw order (NDCG@10)": "{:.4f}"}).hide(axis="index"))
```

## Exercise 4 — A brand-new movie: which stage has to change?

**Set-up.** The catalog has **61 movies with no training interaction at all** (they appear only in the validation/test data, or nowhere). They are the realistic stand-in for "a movie added this morning": its content (genres, year) is known, nobody has rated it. The offline index of Phase 4 contains only the 9,681 movies the models were trained on, so the pipeline as built can never recommend these 61. The experiment adds them, one stage at a time, and measures what happens:

| Variant | What is changed |
|---|---|
| 0. as built | nothing: the movie is not in the index |
| 1. embed + add to the index | the candidate tower embeds the movie from its **content** (it was trained with item-id dropout for exactly this: unknown id + genre vector) and the vector is added to the live index (`index.add`, no rebuild) |
| 2. + scorer fallback | NeuMF has no trained embedding for the movie; use the two-tower (content-based) score instead, put on NeuMF's scale by z-scoring both inside the candidate list |
| 3. + freshness | pretend the movie was released in the request year, so the re-ranker's freshness bonus applies in full (*simulated*: the 61 movies are not new in reality) |
| 4. + reserved exploration slot | the last of the 10 positions is given to the best content-matching new movie, outside the score competition |

Everything is measured on the test requests; the 24 users whose **real next movie is one of these 61** are the ones who could be helped, the other 586 are the ones who could be hurt.

```python
cold_ids = np.arange(n_active, n_items); E_COLD = np.ascontiguousarray(ITEM_EMB[n_active:]).astype("float32")
cold_target_users = np.flatnonzero(TEST_ITEM >= n_active); warm_users = np.setdiff1d(np.arange(n_users), cold_target_users)
SEEN_FULL = {sp: [np.flatnonzero(SEEN[sp][u]).astype("int64") for u in range(n_users)] for sp in ("val", "test")}      # seen ids over the WHOLE catalog (the Phase 4 filter only knew the 9,681 indexed movies)
def clone_index(idx): return faiss.deserialize_index(faiss.serialize_index(idx))

# ---- what a new movie costs to bring in: embed it from its content, add the vector to a copy of the live index ----
def embed_new_item(j):
    return candidate_tower([np.array([[TT_OOV]], np.int32), F[j:j + 1]], training=False).numpy().astype("float32")
assert np.allclose(embed_new_item(cold_ids[0]), ITEM_EMB[cold_ids[0]], atol=1e-4), "the online embedding path must equal the offline one"
_tmp = clone_index(ivf); embed_new_item(cold_ids[1])                                                                  # warm-up
t0 = time.perf_counter(); v_new = embed_new_item(cold_ids[2]); t_embed1 = (time.perf_counter() - t0) * 1000
t0 = time.perf_counter(); _tmp.add(v_new); t_add1 = (time.perf_counter() - t0) * 1000
assert _tmp.ntotal == n_active + 1

ivf_c = clone_index(ivf); ivf_c.add(E_COLD); ivf_c.nprobe = NPROBE
assert ivf_c.ntotal == n_items and ivf.ntotal == n_active, "cold movies must be added to the COPY only"

C_before = retrieve_all(ivf, QT, "test", nprobe=NPROBE)
C_after = np.full((n_users, TOPK), -1, np.int64)
for u in range(n_users):
    _, I = search_filtered(ivf_c, QT[u:u + 1], SEEN_FULL["test"][u], k=TOPK, nprobe=NPROBE); C_after[u] = I[0]
hit_b = (C_before == TEST_ITEM[:, None]).any(1); hit_a = (C_after == TEST_ITEM[:, None]).any(1)
S_all = QT @ ITEM_EMB.T                                                                                                 # exact scores over the whole catalog, cold movies included
rank_cold = []
for u in cold_target_users:
    s = S_all[u].copy(); s[SEEN_FULL["test"][u]] = -np.inf; rank_cold.append(int((s > s[TEST_ITEM[u]]).sum()) + 1)
rank_cold = np.array(rank_cold); n_unseen_med = int(np.median([n_items - len(SEEN_FULL["test"][u]) for u in cold_target_users]))
ex4_idx = pd.DataFrame([
    ("movies in the index", f"{ivf.ntotal:,}", f"{ivf_c.ntotal:,}"),
    ("new movies that can be retrieved", f"0 of {len(cold_ids)}", f"{len(cold_ids)} of {len(cold_ids)}"),
    ("held-out Recall@100, all 610 users", f"{hit_b.mean():.3f}", f"{hit_a.mean():.3f}"),
    (f"… the {len(cold_target_users)} users whose real next movie is new", f"{int(hit_b[cold_target_users].sum())} of {len(cold_target_users)}", f"{int(hit_a[cold_target_users].sum())} of {len(cold_target_users)}"),
    (f"… the other {len(warm_users)} users", f"{hit_b[warm_users].mean():.3f}", f"{hit_a[warm_users].mean():.3f}  (lost {int((hit_b[warm_users] & ~hit_a[warm_users]).sum())}, gained {int((~hit_b[warm_users] & hit_a[warm_users]).sum())})"),
    ("new movies per 100-candidate list (average)", "0", f"{(C_after >= n_active).sum(1).mean():.2f}"),
    ("users with at least one new movie among their candidates", "0%", f"{(C_after >= n_active).any(1).mean():.1%}"),
    ("rank of the real (new) next movie under the content-only embedding", "not in the index", f"median {int(np.median(rank_cold)):,} of ~{n_unseen_med:,} unseen; {int((rank_cold <= 100).sum())} in the top-100, {int((rank_cold <= 1000).sum())} in the top-1000 (of {len(rank_cold)})"),
    ("time to bring one new movie in", "—", f"embed {t_embed1:.1f} ms + index.add {t_add1:.3f} ms"),
], columns=["stage ① retrieval", "as built", "after embed + add"])
display(ex4_idx.style.hide(axis="index"))
```

```python
# ---- stage ②: what does the scorer say about a movie it never trained on? ----
cold_raw = np.array([neumf_score_pairs(u, cold_ids) for u in range(0, n_users, 5)])                 # (122 users, 61 new movies): NeuMF probabilities from the untrained embedding rows
warm_raw = np.array([neumf_score_pairs(u, CAND["test"][u]) for u in range(0, n_users, 5)])          # the same users' 100 retrieved candidates
share_above = float(np.mean([(warm_raw[k][:, None] > cold_raw[k][None, :]).mean() for k in range(len(cold_raw))]))
print(f"NeuMF probability of the new movies: mean {cold_raw.mean():.3f}, max {cold_raw.max():.3f} | of the retrieved candidates: median {np.median(warm_raw):.3f}")
print(f"in {share_above:.1%} of (user, new movie, retrieved candidate) comparisons the candidate scores higher than the new movie")
```

```python
# ---- the five pipeline variants (each one adds a change to the previous one) ----
def best_cold(u, split, q, k):
    """exploration source: the k best never-trained movies for this user by content-based two-tower score (unseen only)"""
    s = (q @ ITEM_EMB[cold_ids].T)[0].copy(); s[SEEN[split][u][cold_ids]] = -np.inf
    return cold_ids[np.argsort(-s)[:k]]

def cold_pipeline(u, split="test", fallback=False, explore=0, fresh=False, details=None):
    q = Q_TT[split][u:u + 1]
    _, I = search_filtered(ivf_c, q, SEEN_FULL[split][u], k=R_DEFAULT, nprobe=NPROBE); cand = I[0][I[0] >= 0]
    s = neumf_score_pairs(u, cand).astype("float64"); isc = cand >= n_active
    if fallback and isc.any() and (~isc).sum() > 1:
        sT = (q @ ITEM_EMB[cand].T)[0].astype("float64")
        zN = (s - s[~isc].mean()) / max(s[~isc].std(), 1e-9); zT = (sT - sT[~isc].mean()) / max(sT[~isc].std(), 1e-9)
        s = np.where(isc, zT, zN)
    if fresh:
        saved = REL_YEAR[cold_ids].copy(); REL_YEAR[cold_ids] = QYEAR[split][u]
    try: out = rerank_user(cand, s, SEEN[split][u], QYEAR[split][u], lam=LAMBDA, beta=BETA, cap=GENRE_CAP, n_short=N_SHORT, n_out=K)
    finally:
        if fresh: REL_YEAR[cold_ids] = saved
    if explore > 0:
        extra = [j for j in best_cold(u, split, q, explore + K) if j not in set(out.tolist())][:explore]
        out = np.concatenate([out[:K - len(extra)], np.array(extra, dtype=np.int64)])
    if details is not None: details.update({"cand": cand, "scores": s})
    return out

def run_variant(**kw):
    L = np.full((n_users, K), -1, np.int64)
    for u in range(n_users): o = cold_pipeline(u, **kw); L[u, :len(o)] = o
    return L

_ry = REL_YEAR.copy()
VARIANTS = {"0. as built (new movies not in the index)": FINAL_TEST, "1. embed from content + add to the index": run_variant(),
            "2. + scorer fallback (content score)": run_variant(fallback=True), "3. + freshness (simulated: released this year)": run_variant(fallback=True, fresh=True),
            "4. + one reserved exploration slot": run_variant(fallback=True, fresh=True, explore=1)}
assert np.array_equal(REL_YEAR, _ry, equal_nan=True), "the simulated release year must be restored"
assert not any(SEEN["test"][u][L[u][L[u] >= 0]].any() for L in VARIANTS.values() for u in range(n_users)), "a variant recommended a movie the user has already seen"

SHOWN = {nm: (int((L >= n_active).sum()), int((L >= n_active).any(1).sum())) for nm, L in VARIANTS.items()}
def short_shown(nm): n, uu = SHOWN[nm]; return "0" if n == 0 else f"{n} (to {uu} user{'s' if uu != 1 else ''})"
def fmt_shown(nm): n, uu = SHOWN[nm]; return "no new movie is shown" if n == 0 else f"{n} new movie{'s' if n != 1 else ''} shown, to {uu} user{'s' if uu != 1 else ''}"
base_nd = per_user_ndcg(ranks_from_lists(VARIANTS["0. as built (new movies not in the index)"], TEST_ITEM)); rows = []
for nm, L in VARIANTS.items():
    r = ranks_from_lists(L, TEST_ITEM); nd_k = per_user_ndcg(r); d_w = paired_boot_diff(nd_k[warm_users], base_nd[warm_users])
    rows.append({"variant": nm, "new movies shown in the 610 final lists": short_shown(nm),
                 f"the {len(cold_target_users)} users with a new real next movie: found in top-10": int((r[cold_target_users] <= K).sum()), "NDCG@10, all 610": float(nd_k.mean()),
                 f"NDCG@10, the {len(warm_users)} other users": float(nd_k[warm_users].mean()), "Δ for those vs variant 0 (paired CI)": "—" if nm.startswith("0.") else "{:+.4f} [{:+.4f}, {:+.4f}]".format(*d_w)})
ex4_var = pd.DataFrame(rows)
display(ex4_var.style.format({"NDCG@10, all 610": "{:.4f}", f"NDCG@10, the {len(warm_users)} other users": "{:.4f}"}).hide(axis="index"))
```

### 4.1 One new movie, followed through every stage

Two real cases among the 24 users whose real next movie is new: **(a)** the pair that retrieval gets closest to (best exact rank under the content-only embedding) — it shows how far a new movie gets through the normal path; **(b)** a user who *was* helped by the reserved slot (if there is one) — it shows what finally lets a movie in.

```python
def trace_movie(u, j, split="test"):
    q = Q_TT[split][u:u + 1]
    s_all = (q @ ITEM_EMB.T)[0].copy(); s_all[SEEN_FULL[split][u]] = -np.inf; rank_exact = int((s_all > s_all[j]).sum()) + 1; n_unseen = int(np.isfinite(s_all).sum())
    _, I = search_filtered(ivf_c, q, SEEN_FULL[split][u], k=R_DEFAULT, nprobe=NPROBE); cand = I[0][I[0] >= 0]; pos = np.flatnonzero(cand == j)
    rows = [("0. as built", "the movie is not in the index", "retrieval can never return it, so no later stage ever sees it")]
    if len(pos):
        s_nm = neumf_score_pairs(u, cand); p = float(s_nm[pos[0]]); rk = int((s_nm > p).sum()) + 1
        rows.append(("1. ① retrieval, after embed + add", f"exact rank {rank_exact} of {n_unseen:,} unseen movies; returned as candidate #{pos[0] + 1} of {len(cand)}", "stage ① can now carry it"))
        rows.append(("2. ② NeuMF score", f"probability {p:.3f} vs median {np.median(s_nm):.3f} of the other candidates → position {rk} of {len(cand)}",
                     ("outside the shortlist of " + str(N_SHORT) + ": the re-ranker never sees it") if rk > N_SHORT else "inside the shortlist"))
        det = {}; cold_pipeline(u, split, fallback=True, details=det); sc = det["scores"]; p2 = np.flatnonzero(det["cand"] == j)[0]; rk2 = int((sc > sc[p2]).sum()) + 1
        rows.append(("3. ② content-score fallback", f"z-score {sc[p2]:+.2f} → position {rk2} of {len(sc)}", ("still outside the shortlist of " + str(N_SHORT)) if rk2 > N_SHORT else "inside the shortlist"))
    else:
        rows.append(("1. ① retrieval, after embed + add", f"exact rank {rank_exact} of {n_unseen:,} unseen movies; NOT among the {R_DEFAULT} candidates", "stage ① drops it: scoring and re-ranking never see it"))
    for nm, L in list(VARIANTS.items())[1:]:
        where = np.flatnonzero(L[u] == j)
        rows.append((f"final top-10, variant {nm[0]}", f"position {where[0] + 1} of {K}" if len(where) else "not shown",
                     "the reserved slot: chosen as the best content match among the new movies, not by competing on score" if (len(where) and nm.startswith("4.")) else ""))
    return pd.DataFrame(rows, columns=["step", "what happens to the movie", "consequence"])

u_a = int(cold_target_users[np.argmin(rank_cold)]); j_a = int(TEST_ITEM[u_a])
display(Markdown(f"**(a) best-retrieved case:** user {int(user_ids[u_a])}, new movie *{movies.title[j_a]}*"))
ex4_trace = trace_movie(u_a, j_a); display(ex4_trace.style.hide(axis="index"))
helped = [int(u) for u in cold_target_users if (VARIANTS[list(VARIANTS)[4]][u] == TEST_ITEM[u]).any()]
if helped:
    u_b = helped[0]; j_b = int(TEST_ITEM[u_b])
    display(Markdown(f"**(b) a case the reserved slot helped:** user {int(user_ids[u_b])}, new movie *{movies.title[j_b]}*"))
    ex4_trace_b = trace_movie(u_b, j_b); display(ex4_trace_b.style.hide(axis="index"))
```

### 4.2 Which stages have to change — the answer

The table gives, for each stage, what happens to a brand-new movie today, what has to change, and the measurement above that shows it.

```python
v = ex4_var.set_index("variant"); keys = list(v.index); col_hit = [c for c in v.columns if c.startswith("the ") and "found in top-10" in c][0]
dw4 = v.loc[keys[4], "Δ for those vs variant 0 (paired CI)"]
nd_ex = {k: per_user_ndcg(ranks_from_lists(L, TEST_ITEM)) for k, L in VARIANTS.items()}
base_warm = float(nd_ex[keys[0]][warm_users].mean()); d_w4 = paired_boot_diff(nd_ex[keys[4]][warm_users], nd_ex[keys[0]][warm_users])
d_w1 = paired_boot_diff(nd_ex[keys[1]][warm_users], nd_ex[keys[0]][warm_users])      # variant 1 alone: new vectors in the index, nothing reserved
lost_w = int((hit_b[warm_users] & ~hit_a[warm_users]).sum()); gained_w = int((~hit_b[warm_users] & hit_a[warm_users]).sum())
yr = REL_YEAR[cold_ids]; yr = yr[np.isfinite(yr)]
ex4_stage = pd.DataFrame([
    ("offline: embeddings + index", "the movie has no vector and is not in the index → invisible", "embed it from content when it is published and add it to the live index (or to a small 'fresh' index that is searched alongside the big one)",
     f"{len(cold_ids)} of {len(cold_ids)} retrievable after the change; cost {t_embed1:.1f} ms to embed + {t_add1:.3f} ms to add"),
    ("① retrieval", "cannot return it", "content-only vectors are weak (no taste or popularity signal): the real new movie ranks in the top-100 for only some users",
     f"{int(hit_a[cold_target_users].sum())} of {len(cold_target_users)} users retrieved their real new movie; median exact rank {int(np.median(rank_cold)):,} of ~{n_unseen_med:,}; for the other users the candidate lists lost {lost_w} and gained {gained_w} real next movies"),
    ("② scoring", "NeuMF never trained an embedding for it (negatives are drawn from the trained movies only), so its probability is meaningless", "a fallback score for movies with too few interactions (content/two-tower score, or a bandit prior) — and a rule for when a movie 'graduates' to the normal scorer",
     f"NeuMF gives new movies mean {cold_raw.mean():.3f} vs median {np.median(warm_raw):.3f} for retrieved candidates; with the content-score fallback: {fmt_shown(keys[2])}"),
    ("③ re-ranking", "the freshness bonus only re-orders movies that are already in the shortlist of 50; it cannot add a movie", "an explicit exploration slot (or quota) that is not decided by the score competition",
     f"with simulated freshness added: {fmt_shown(keys[3])}; with one reserved slot: {fmt_shown(keys[4])}, {int(v.loc[keys[4], col_hit])} of {len(cold_target_users)} real new next movies found in the top-10, cost to the other users Δ NDCG@10 {dw4}"),
    ("feedback loop", "no impressions → no clicks → no training signal → still cold at the next retraining", "log every impression of a new movie; after N interactions train its ID embedding and drop the fallback", "not simulated here (needs live traffic)"),
], columns=["stage", "today, for a brand-new movie", "what has to change", "evidence in this notebook"])
display(ex4_stage.style.hide(axis="index"))

display(Markdown(f"""
**Reading.**

* **Nothing reaches a new movie until stage ① changes.** As built it is invisible (variant 0). Embedding it from its content and adding the vector costs {t_embed1:.1f} ms + {t_add1:.3f} ms and makes all {len(cold_ids)} retrievable, but the content-only vector places the user's real new movie in the top-100 for only {int(hit_a[cold_target_users].sum())} of {len(cold_target_users)} users (median exact rank {int(np.median(rank_cold)):,} of about {n_unseen_med:,}).
* **Stage ② is the second wall.** The scorer was never trained on the movie, and new movies score {cold_raw.mean():.3f} on average, against a median of {np.median(warm_raw):.3f} for the retrieved candidates; in {share_above:.0%} of comparisons the old candidate wins. A fallback score lets the movie through the *scoring* step honestly, but with content information alone it still rarely beats established movies — with the fallback (variant 2), {fmt_shown(keys[2])} in the final lists.
* **Stage ③ cannot fix what the earlier stages dropped.** Simulated freshness (variant 3) changes little ({fmt_shown(keys[3])}), because the bonus only re-orders the shortlist. Only a **reserved exploration slot** (variant 4) makes new movies appear reliably ({fmt_shown(keys[4])}); it found {int(v.loc[keys[4], col_hit])} of the {len(cold_target_users)} real new next movies in the top-10 — {"a small absolute number, but far above what the other variants reach" if int(v.loc[keys[4], col_hit]) > int(v.loc[keys[3], col_hit]) else "no more than the other variants"} — at a cost to the other {len(warm_users)} users of ΔNDCG@10 {dw4} ({d_w4[0] / base_warm:+.0%} of their {base_warm:.4f}; {"the interval excludes zero" if (d_w4[1] > 0 or d_w4[2] < 0) else "the interval includes zero"}). (Variant 1 alone — the new vectors in the index, nothing reserved — already moves it by {d_w1[0]:+.4f} [{d_w1[1]:+.4f}, {d_w1[2]:+.4f}]: the new vectors take places in the candidate lists, and {lost_w} of these {len(warm_users)} users lost their real next movie from the candidates that way.) That cost is the price of learning about new movies, and how many slots to spend on it is a product decision to be tested online.
* **Caveats.** The 61 movies are movies nobody rated during training, not necessarily newly released (their release years run from {int(yr.min())} to {int(yr.max())}), so freshness is simulated; with only {len(cold_target_users)} affected users these counts are anecdotes, not estimates; and with a catalog of millions of new movies the exploration source would have to be a proper index search, not a scan of {len(cold_ids)} vectors.
"""))
```

## Exercise 5 — Two A/B tests, two different designs (written)

Phase 8.4 argued *why* retrieval and scoring/re-ranking changes are tested separately. This is the written design for the two cases the exercise names: a **retrieval-model change** and a **re-ranking-rule change**. The sample sizes and the evidence quoted come from the measurements above.

```python
n_pa = {x: users_per_arm(p_base, x) for x in (0.02, 0.05, 0.10, 0.20)}                       # users per arm, from the Phase 8.4 arithmetic
e2 = ex2_tab.set_index("R")
ceil_gain = 100 * (e2.loc[500, "ceiling (held-out in candidates)"] - e2.loc[100, "ceiling (held-out in candidates)"])
ex2_nodiff = bool(e2.loc[500, "test NDCG@10"] <= e2.loc[100, "test NDCG@10"])                     # R = 500 gave no better scorer list than R = 100
ex3_nodiff = bool(all(EX3[w]["test"]["d_neumf"][1] <= 0 for w in WINDOWS))                      # no trending window has a test interval above zero
both_nodiff = ex2_nodiff and ex3_nodiff
stage_note = (f"a retriever with a higher ceiling did **not** give a better final list (Exercise 2: the share of users whose next movie is a candidate rose by {ceil_gain:.0f} points from R=100 to R=500 while NDCG@10 did not improve; Exercise 3: a second source raised it further, again without a better list)"
              if both_nodiff else
              f"a higher ceiling does not automatically give a better final list (Exercise 2: {ceil_gain:+.0f} points of ceiling from R=100 to R=500 moved NDCG@10 {e2.loc[100, 'test NDCG@10']:.4f} → {e2.loc[500, 'test NDCG@10']:.4f}; see also Exercise 3)")
recall_note = "this notebook shows it can rise without any gain in the list (Exercises 2 and 3)" if both_nodiff else "its link to the final list has to be checked, as Exercises 2 and 3 do"
fooled_note = "a higher recall with no gain in the final list — what Exercises 2 and 3 found offline" if both_nodiff else "the link between recall and the final list has to be measured, as Exercises 2 and 3 do"
rel_cost = d_w4[0] / base_warm                                                                # cost of the whole new-movie path (variant 4 vs as built) to the other users (Exercise 4)
display(Markdown(f"""
### What both tests share

* **Randomise by user, not by request.** Recommendations are personalised and people come back: a user must stay in one arm (hash of the user id → bucket) so they never see both lists, and the *user* is the unit of analysis (a request-level analysis makes the intervals too narrow).
* **Decide on the list the user actually sees.** The stages interact. In this notebook {stage_note}. A stage-level metric is a *diagnostic*, never the decision.
* **Before the test:** an A/A run (the same system in both arms) to check the plumbing, and a sample-ratio check during the test (a 50/50 split that drifts to 52/48 means a bug, so stop and debug).
* **Size.** For a per-user success rate like our offline HR@10 ({p_base:.3f}), a two-sided 5% test with 80% power needs about **{n_pa[0.05]:,} users per arm** to detect a +5% relative change, {n_pa[0.10]:,} for +10% and {n_pa[0.20]:,} for +20% (Phase 8.4); +2% would need {n_pa[0.02]:,}. Small relative changes need a lot of traffic, so use the least noisy metric that still reflects the goal and reduce variance with pre-experiment behaviour as a covariate. How long that takes depends on traffic: users per arm divided by the number of eligible users reaching each arm per day.

### Design A — a retrieval-model change
*Example: the two-tower model v2, a different index, or the "trending" source of Exercise 3.*

* **Hypothesis.** The new retriever puts more of what the user will want into the candidate set, and the unchanged scorer and re-ranker turn that into a better final list.
* **Arms.** Control = current retriever; treatment = new retriever. Scorer, re-ranker and filters are identical.
* **Primary metric.** User-level success on the final list: the share of users who play or click a recommended movie during the test period (offline stand-in: HR@10); watch time per user as the continuous companion with more statistical power.
* **Why not candidate recall as the primary metric?** Recall is an intermediate quantity; {recall_note}. It is reported as a *diagnostic*: candidate overlap with control, the share of final top-10 movies that only the new retriever proposed (every candidate carries a source tag), recall@R of the movies users later chose, catalogue coverage.
* **Guardrails.** (1) Latency: retrieval stage p99 within its 15 ms share and end-to-end p99 ≤ 100 ms (Phase 7). (2) Reliability: empty or short candidate lists, timeouts and the popularity-fallback rate must not rise. (3) Hard invariants: no already-seen or ineligible movie in any list (must be exactly 0). (4) Exposure: catalogue coverage, ILD@10 and the head-100 share not worse than control by more than a stated margin — a retriever that collapses onto popular movies can raise clicks and still damage the catalogue. (5) Cost: index build time, memory, CPU per request. (6) Long-run: next-week return rate.
* **Offline gate before any traffic.** Held-out Recall@R with a paired interval not worse than production; index recall against exact search ≥ 0.95; latency inside the budget; zero seen movies. Then **shadow mode** on live traffic (the new retriever runs, its output is not shown) to compare overlap, latency and failures — an offline replay cannot credit movies the old system never showed.
* **Duration.** Ramp 1% → 10% → 50%; at least two full weeks at the final split (weekday/weekend cycle); compare week 1 with week 2 for a novelty effect.
* **Decision rule.** Ship if the primary metric's 95% interval is above zero and every guardrail is inside its margin. If the change is meant to save cost or latency (for example an HNSW swap, Exercise 1), run it as a **non-inferiority** test with a margin set in advance.
* **Ways to be fooled.** The scorer cannot turn the extra candidates into a better list ({fooled_note}); the new retriever surfaces movies with no history that the scorer under-rates (Exercise 4); a shared training pipeline feeds the treatment arm's clicks into both arms' next model.

### Design B — a re-ranking-rule change
*Example: MMR λ {LAMBDA} → 0.4 (more diversity), the freshness weight β {BETA} → 1.0, or the reserved exploration slot of Exercise 4.*

* **Hypothesis.** The rule gives up a little short-term relevance for something the product values — variety, fresh content, exposure for new or long-tail movies — and the exchange is worth it. Offline, the rules of Phase 6 changed the NDCG@10 of the raw scored list by {d_final_raw[0]:+.4f} (CI [{d_final_raw[1]:+.4f}, {d_final_raw[2]:+.4f}]) and raised ILD@10 from {final_tab.iloc[0]['ILD@10']:.3f} to {final_tab.iloc[-1]['ILD@10']:.3f}; the new-movie path of Exercise 4 (new vectors in the index plus one reserved exploration slot) cost the other users {rel_cost:+.0%} of their NDCG@10. Offline accuracy cannot say whether that trade is good.
* **Arms.** Same candidates and same scores in both arms; only the final selection and order differ.
* **Primary metric.** What the rule is meant to improve, measured per user over a longer horizon — sessions or distinct active days per user over 28 days, or a satisfaction signal. If the rule is a business requirement (a freshness quota, say), the test is a **cost measurement** instead: the primary metric is engagement and the question is whether the loss stays inside an agreed budget.
* **Guardrail with a margin (non-inferiority).** Short-term engagement — plays or clicks per user, and click-through of the top-3 slots — must not fall by more than a pre-agreed margin (for example −2% relative): its interval's lower bound has to stay above the margin. A re-ranker trades relevance *by construction*, so "no significant loss" is not enough; the margin is a product decision made before the test.
* **Rule-specific metrics (secondary).** ILD@10, genre entropy, share of movies released within the last N years, exposure share of new and long-tail movies, catalogue coverage.
* **Other guardrails.** Re-ranker latency ≤ 10 ms; hard-rule violations (seen movies, the genre cap, policy filters) = 0; hides, "not interested" and complaint rates; the click distribution by slot (changing the order moves where clicks happen); segments — new vs heavy users — so an average does not hide a group that is hurt.
* **Offline gate.** "Relevance kept" ≥ the Phase 6 threshold, zero seen movies, cap respected. A cheaper first online step is **interleaving** (mix control and treatment lists in one display and count which list's movies are chosen), which needs far fewer users than a classic A/B; confirm with an A/B afterwards.
* **Duration and size.** At least four weeks plus a small long-term holdout that stays on control for months: the benefit of variety shows up as retention, slowly, and a novelty bump in week 1 is typical. A non-inferiority margin of −2% on a binary success rate like ours needs about {n_pa[0.02]:,} users per arm; −5% needs about {n_pa[0.05]:,}. Tight margins are expensive, so use a continuous metric with variance reduction or accept a wider margin.
* **Decision rule.** Ship if the rule-specific target improves by the pre-set amount, the long-run primary metric is not worse, and the engagement guardrail's lower bound is above its margin.
* **Ways to be fooled.** Optimising a diversity number users do not notice; judging only week-1 clicks and killing a rule whose benefit arrives later; a rule that is fine on average but hurts one segment.

### Why the two designs differ

| | retrieval change | re-ranking-rule change |
|---|---|---|
| what it changes | which movies are *available* | which of the *same* movies are shown, and in what order |
| decision metric | end-to-end success of the final list | the long-run metric the rule is meant to improve |
| role of short-term clicks | the primary metric | a guardrail with a non-inferiority margin |
| diagnostics | candidate recall, overlap, share from the new source | ILD, freshness share, long-tail exposure |
| main risks | recall loss, coverage collapse, latency and infrastructure | position effects, policy violations, slow-to-appear user effects |
| duration | ≥ 2 weeks after shadow mode and ramp-up | ≥ 4 weeks plus a long-term holdout |
"""))
EX5_DONE = True
```
